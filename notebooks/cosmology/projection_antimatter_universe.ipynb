A speculative hypothesis, not established physics. Step 2 found that one worldline crossing every time slice gives equal matter and antimatter, while we see about $10^{80}$ baryons and no antimatter. The escape route is a worldline balanced globally, with the antimatter far away or in a mirror universe. This step lists what such an antimatter universe must satisfy and tests each criterion against data. There are no antinuclei in cosmic rays and no annihilation gamma rays from clusters. Domains must be at least as large as the horizon, about 16 Gpc. A random worldline is fantastically unlikely. That leaves the CPT-symmetric universe of Boyle, Finn and Turok, whose predictions (a massless lightest neutrino, $r = 0$) agree with data so far and could fail.

Step 7 of the Projection Hypothesis, a theory step for postulate P3 (all particles are projections of one object). Step 1 (cosmology/projection_postulate) stated the postulates. The question is what such an antimatter universe must satisfy to agree with the postulates, and whether it does. Known physics first. The Sakharov conditions (1967) explain a matter excess by baryon-number violation, C and CP violation and departure from equilibrium. They predict a universe made of matter everywhere, with $\eta = n_B/n_\gamma = 6.1\times10^{-10}$ (Planck 2018, $\Omega_b h^2 = 0.0224$). The alternative, a symmetric universe with separate matter and antimatter regions, was studied from the 1960s (Alfvén and Klein, Omnès). Steigman (ARA&A 14, 339, 1976) reviewed the tests, and Cohen, De Rújula and Glashow (ApJ 495, 539, 1998) closed the case for domains inside the visible universe. All data come from data/antimatter_criteria.csv, which holds cited numbers with their sources. The number of dimensions $D$ the worldline lives in is a parameter. Only its time coordinate enters, so no result depends on $D$. Criterion 1: no antimatter mixed in nearby. BESS-Polar found no antihelium among about 48 million helium nuclei. The 95% limit is antihelium/helium $< 6.9\times10^{-8}$ for a helium-like spectrum and $1.0\times10^{-7}$ without that assumption (Abe et al., PRL 108, 131301, 2012). AMS-02 has published no antinucleus detection. Gamma rays: in the hot gas of a cluster, an antiproton fraction $f$ annihilates at the rate $f\,n_p^2\langle\sigma v\rangle$, with $\langle\sigma v\rangle\approx10^{-15}$ cm$^3$/s (an order-of-magnitude choice). Each annihilation gives 2.1 photons above 100 MeV in the pion-decay spectrum used here. For the Coma cluster (beta model, $n_0 = 3.4\times10^{-3}$ cm$^{-3}$, $r_c = 291$ kpc, $\beta = 0.75$) and the Fermi-LAT limit $F(>100\ \mathrm{MeV}) < 5.2\times10^{-9}$ cm$^{-2}$ s$^{-1}$ (Ackermann et al., ApJ 819, 149, 2016), this gives $f < 9\times10^{-8}$. Satisfied: any antimatter must be outside clusters and beyond the reach of cosmic rays, which are confined to tens of Mpc. Criterion 2: no domains inside our horizon. If matter and antimatter domains of comoving size $d$ touched, annihilation at their walls would be unavoidable after recombination. The model here is the simple version of Cohen, De Rújula and Glashow's argument. The wall area per volume is $3/d$ for cubes of side $d$ ($n/d$ in $n$ dimensions, tested), and baryons flow into the walls at $\kappa$ times the gas sound speed. The gas tracks the CMB temperature down to $z\approx150$ and then cools adiabatically. Each annihilation gives 3 photons from neutral-pion decay. The photons redshift to today, $dJ/dE = (c/4\pi)\int dz\,\dot{\mathcal N}\,S(E(1+z))/H$, and that integral is checked against a closed form. Annihilation is counted from recombination ($z_{max} = 1000$) down to $z = 20$. As in CDG, later epochs are left out because they depend on structure formation; including them would only raise the flux. The flux is exactly $\propto 1/d$ (tested). It is compared with the COMPTEL background, $dJ/dE = 6.4\times10^{-3}(E/\mathrm{MeV})^{-2.38}$ cm$^{-2}$ s$^{-1}$ sr$^{-1}$ MeV$^{-1}$ from 0.8 to 30 MeV (Kappadath et al. 1996), and the Fermi-LAT isotropic background ($7.2\times10^{-6}$ cm$^{-2}$ s$^{-1}$ sr$^{-1}$ above 100 MeV, index 2.32, cutoff 279 GeV; Ackermann et al., ApJ 799, 86, 2015). The photons from $z\sim20$-$300$ arrive at 1 to 10 MeV, where COMPTEL sets the limit. Result: the prediction exceeds the data by a factor 160 for $d = 100$ Mpc, 16 for 1 Gpc and 1.6 for 10 Gpc. The minimum domain size is $d_{min}\approx16$ Gpc (at 8.5 MeV, $\kappa = 1$), larger than the 14.3 Gpc radius of the observable universe. CDG reached the same conclusion: the matter domain must be nearly the whole visible universe. This is a toy model, and $d_{min}$ scales with $\kappa$. CDG argue that annihilation heating speeds up the inflow, which strengthens the bound. Constrained: any antimatter must be beyond our horizon, where no telescope can test it. Criterion 3: fine-tuning. Suppose the antimatter is beyond the horizon on the same slice and nothing arranges it. A balanced random worldline then gives our patch a net charge of order $\sqrt M$ for $M$ crossings. The net-charge variance equals $M$, or $M(T-M)/(T-1)$ when the total of $T$ crossings is balanced (both tested). The chance of a large excess follows Cramér's rate function, $\log_{10}P\approx -M\,\mathrm{KL}/\ln10$, checked against the exact binomial tail. Today we see $N_B\approx9\times10^{79}$ baryons and no antimatter, so every crossing must have the same sign: $P = 2^{-N_B}$, $\log_{10}P\approx-2.7\times10^{79}$. Read instead as the early-universe excess of $9\times10^{79}$ among about $1.5\times10^{89}$ pairs (one per photon), it is $\log_{10}P\approx-1.2\times10^{70}$. Without a mechanism the balanced worldline fails. The mechanism would have to be a boundary condition that correlates the signs of all crossings in our patch: a symmetry, not a fluctuation. Criterion 4: the CPT-symmetric universe. Boyle, Finn and Turok (PRL 121, 251301, 2018) propose that the universe after the Big Bang is the CPT mirror image of a universe before it, so the pair as a whole does not violate CPT. The author's alignment of P3 with this model follows. A worldline that crosses $t = 0$ is matter on our side. Observers in the mirror have the opposite arrow of time, so they see the same stretch as antimatter. The caveat proved in step 2 still applies: on any complete slice one continuous worldline has $|Q|\leq1$, so the mirror alone does not help. The alignment works if the worldline winds through the Big Bang and through the future boundary, with the two ends of conformal time glued together. That is an extra assumption beyond the published model and the step-2 extension C in a new place. A loop that winds $w$ times reads $+w$ on every slice of our side and $-w$ on every mirror slice, read with the mirror's own arrow. The total is zero (tested). The excess is now a topological integer, not a fluctuation, though $w\approx10^{80}$ is still put in by hand. The model's published predictions (from the abstract of Boyle, Finn and Turok), tested against cited numbers: (a) The lightest neutrino is massless. With the NuFIT 6.0 splittings ($\Delta m^2_{21} = 7.49\times10^{-5}$ eV$^2$, $\Delta m^2_{31} = 2.513\times10^{-3}$ eV$^2$ normal, $\Delta m^2_{32} = -2.484\times10^{-3}$ eV$^2$ inverted) this fixes $\Sigma m_\nu = 0.0588$ eV (normal ordering) or $0.0989$ eV (inverted), and $m_\beta = 0.009$ or $0.049$ eV. KATRIN bounds $m_\beta$ below 0.8 eV (Nature Physics 18, 160, 2022) and 0.45 eV (Science 388, 180, 2025), far above both values. Planck 2018 gives $\Sigma m_\nu < 0.12$ eV, which allows both. ## 2. `equation.py` — physics → tensor translation

```python
%%writefile problems/cosmology/projection_antimatter_universe/equation.py
"""The criteria an antimatter universe must meet, as tensor operations.

Shapes in comments: E photon energies, Z redshifts, D domain sizes, J pion energies, B batch members, S samples along
a worldline, L time slices, W windings. Summed indices: the redshift z (photons arriving today), the pion energy j
(photon spectrum of one annihilation), the radius r (emission measure), the segment s (crossings of a slice).
Units: cgs, energies in MeV, distances in Mpc unless stated.
"""

import math

import torch

C_CM = 2.998e10  # speed of light, cm/s
MPC_CM = 3.0857e24  # one megaparsec in cm
K_B = 1.380649e-16  # Boltzmann constant, erg/K
M_P = 1.67262e-24  # proton mass, g
M_PI0 = 134.977  # neutral pion mass, MeV
N_GAMMA = 3.0  # photons per p pbar annihilation (about 1.5 neutral pions, 2 photons each)


# --- criterion 1: nothing mixed in nearby -------------------------------------------------------------------------


def emission_measure(n0: float, rc: float, beta: float, r_max: float, n: int = 4000) -> float:
    """int_0^r_max 4 pi r^2 n(r)^2 dr for the beta model n = n0 (1 + r^2/rc^2)^(-3 beta / 2) (r summed, log grid)."""
    r = torch.logspace(math.log10(1e-4 * rc), math.log10(r_max), n, dtype=torch.float64)
    f = 4 * math.pi * r**2 * n0**2 * (1 + (r / rc) ** 2) ** (-3 * beta)
    core = 4 / 3 * math.pi * (1e-4 * rc) ** 3 * n0**2  # the tiny sphere inside the first grid point
    return float(torch.trapezoid(f, r) + core)


def photons_above(E_min: float) -> float:
    """Photons above E_min (MeV) per annihilation, from the pion-decay spectrum."""
    E = torch.logspace(math.log10(E_min), 3.3, 4000, dtype=torch.float64)
    return float(torch.trapezoid(pion_decay_spectrum(E), E))


def cluster_flux(f: torch.Tensor, c: dict) -> torch.Tensor:
    """Gamma-ray flux above 100 MeV (ph cm^-2 s^-1) from a Coma-like cluster whose gas holds an antiproton fraction f:
    F = N_gamma(>100 MeV) f <sigma v> int n_p^2 dV / (4 pi D^2), with n_p = n_e / 1.17 (helium)."""
    rc = c["coma_rc_kpc"] * 1e-3 * MPC_CM
    em = emission_measure(c["coma_n0_cm3"] / 1.17, rc, c["coma_beta"], r_max=3 * MPC_CM)
    dist = C_CM / 1e5 * c["coma_z"] / (100 * c["planck_h"]) * MPC_CM  # Hubble law, cm
    return photons_above(100.0) * f * c["pbar_p_sigma_v"] * em / (4 * math.pi * dist**2)


def cluster_antimatter_limit(c: dict) -> float:
    """Largest antiproton fraction allowed by the Fermi-LAT upper limit on Coma (the flux is linear in f)."""
    one = cluster_flux(torch.ones(1, dtype=torch.float64), c)
    return float(c["coma_flux_ul"] / one)


# --- criterion 2: domains inside our horizon -----------------------------------------------------------------------


def wall_area_per_volume(d: torch.Tensor, n: int = 3) -> torch.Tensor:
    """Checkerboard of cubes of side d in n dimensions: each cell owns n faces of area d^(n-1), volume d^n -> n / d."""
    return n / d


def domains(noise: torch.Tensor, d: torch.Tensor) -> torch.Tensor:
    """(D, N, N) random matter (+1) / antimatter (-1) domains of typical size d (pixels): the sign of white noise
    smoothed by a Gaussian of width d / 3 (FFT, periodic box)."""
    N = noise.shape[-1]
    k = torch.fft.fftfreq(N, dtype=torch.float64) * 2 * math.pi  # (N,) rad per pixel
    k2 = k[:, None] ** 2 + k[None, :] ** 2  # (N, N)
    sigma = (d / 3)[:, None, None]  # (D, 1, 1)
    field = torch.fft.ifft2(torch.fft.fft2(noise)[None] * torch.exp(-k2 * sigma**2 / 2)).real
    return torch.where(field >= 0, 1, -1)


def walls(sign: torch.Tensor) -> torch.Tensor:
    """(..., N, N) mask of pixels next to a neighbour of the opposite kind (where annihilation happens)."""
    m = torch.zeros_like(sign, dtype=torch.bool)
    for shift, dim in ((1, -1), (-1, -1), (1, -2), (-1, -2)):
        m |= sign != torch.roll(sign, shift, dim)
    return m


def measured_wall_per_area(sign: torch.Tensor) -> torch.Tensor:
    """(...,) wall length per unit area in pixels^-1: unlike nearest-neighbour pairs (x and y bonds) per pixel."""
    bonds = (sign != torch.roll(sign, 1, -1)).double() + (sign != torch.roll(sign, 1, -2)).double()
    return bonds.mean((-1, -2))


def z_grid(z_max: float, n: int, z_min: float = 0.0):
    """Redshifts z (Z,) uniform in ln(1 + z) from z_min to z_max, and trapezoid weights dz (Z,)."""
    x = torch.linspace(math.log1p(z_min), math.log1p(z_max), n, dtype=torch.float64)
    z = torch.expm1(x)
    w = torch.full_like(x, x[1] - x[0])
    w[[0, -1]] /= 2
    return z, w * (1 + z)


def hubble(z: torch.Tensor, c: dict) -> torch.Tensor:
    """H(z) in s^-1 for flat LCDM with radiation (Omega_r h^2 = 4.15e-5, photons + neutrinos)."""
    h = c["planck_h"]
    om, orad = c["planck_omega_m"], 4.15e-5 / h**2
    H0 = 100 * h * 1e5 / MPC_CM
    return H0 * torch.sqrt(om * (1 + z) ** 3 + orad * (1 + z) ** 4 + (1 - om - orad))


def gas_sound_speed(z: torch.Tensor, c: dict) -> torch.Tensor:
    """Adiabatic sound speed (cm/s) of the gas: T = T_CMB (1+z) above z_dec, then adiabatic cooling (1+z)^2,
    and 10^4 K, fully ionised, after reionization. c_s = sqrt(5 k T / (3 mu m_p))."""
    T0, zd, zr = c["cmb_t0_k"], c["z_dec"], c["planck_z_re"]
    T = torch.where(z >= zd, T0 * (1 + z), T0 * (1 + z) ** 2 / (1 + zd))
    ionised = z < zr
    T = torch.where(ionised, torch.clamp(T, min=1e4), T)
    mu = torch.where(ionised, 0.59, 1.22)
    return torch.sqrt(5 * K_B * T / (3 * mu * M_P))


def pion_decay_spectrum(E: torch.Tensor, n_pion: int = 120) -> torch.Tensor:
    """Photons per MeV from one p pbar annihilation (a modelling choice): N_GAMMA photons from neutral pions with total
    energies E_pi ~ Gaussian(350 MeV, 150 MeV); each pion gives a flat box of photons from E_pi(1 - beta)/2 to
    E_pi(1 + beta)/2 of width p_pi (the pion energy j is summed)."""
    Epi = torch.linspace(M_PI0 * 1.001, 1200.0, n_pion, dtype=torch.float64, device=E.device)  # (J,)
    w = torch.exp(-((Epi - 350.0) ** 2) / (2 * 150.0**2))
    w = w / w.sum()
    p = torch.sqrt(Epi**2 - M_PI0**2)
    lo, hi = (Epi - p) / 2, (Epi + p) / 2
    inside = (E[..., None] >= lo) & (E[..., None] <= hi)  # (..., J)
    return N_GAMMA * (inside.double() * (w / p)).sum(-1)


def redshifted_density(E, z, dz, rate, H, spectrum) -> torch.Tensor:
    """Photons per cm^3 per MeV today (E,): n(E) = int dz rate(z) S(E (1+z)) / H(z), summed over z.
    rate (..., Z) is per comoving volume per second, H (Z,), spectrum a function of the emitted energy."""
    S = spectrum(E[:, None] * (1 + z)[None, :])  # (E, Z)
    return torch.einsum("...z,ez->...e", rate / H * dz, S)  # repeated z => summed


def annihilation_rate(z: torch.Tensor, d_mpc: torch.Tensor, c: dict, speed_factor: float) -> torch.Tensor:
    """(D, Z) annihilations per comoving cm^3 per s: (wall area / volume) a^2 n_b(z) v = (3 / d) n_b0 (1 + z) v(z),
    with baryons flowing into the walls at v = speed_factor x the gas sound speed."""
    eta = 273.9e-10 * c["planck_omega_b_h2"]
    nb0 = eta * c["n_gamma_cm3"]
    av = wall_area_per_volume(d_mpc * MPC_CM)  # (D,) comoving cm^-1
    v = speed_factor * gas_sound_speed(z, c)  # (Z,)
    return av[:, None] * nb0 * (1 + z)[None, :] * v[None, :]


def annihilation_intensity(E, d_mpc, c, speed_factor: float, z_max: float, z_min: float = 20.0, nz: int = 1500):
    """(D, E) diffuse gamma-ray intensity dJ/dE today (ph cm^-2 s^-1 sr^-1 MeV^-1) from annihilation at domain walls
    between recombination (z_max) and z_min: J = c n(E) / (4 pi). As in Cohen, De Rujula & Glashow, later epochs
    (structure formation, reionization) are left out, which makes the bound conservative; absorption is neglected."""
    z, dz = z_grid(z_max, nz, z_min)
    R = annihilation_rate(z, d_mpc, c, speed_factor)
    n = redshifted_density(E, z, dz, R, hubble(z, c), pion_decay_spectrum)
    return C_CM * n / (4 * math.pi)


def comptel_intensity(E: torch.Tensor, c: dict) -> torch.Tensor:
    """COMPTEL diffuse background fit dJ/dE (ph cm^-2 s^-1 sr^-1 MeV^-1), NaN outside 0.8-30 MeV."""
    J = c["comptel_norm"] * E ** (-c["comptel_index"])
    return torch.where((E >= c["comptel_emin_mev"]) & (E <= c["comptel_emax_mev"]), J, torch.nan)


def fermi_intensity(E: torch.Tensor, c: dict) -> torch.Tensor:
    """Fermi-LAT isotropic background fit: K (E/100 MeV)^-Gamma exp(-E/E_cut), K fixed by the published intensity
    above 100 MeV; NaN outside 100 MeV - 820 GeV."""
    lo, hi, g, ec = c["fermi_emin_mev"], c["fermi_emax_mev"], c["fermi_index"], c["fermi_cutoff_gev"] * 1e3
    e = torch.logspace(math.log10(lo), math.log10(hi), 4000, dtype=torch.float64)
    shape = lambda x: (x / 100.0) ** (-g) * torch.exp(-x / ec)  # noqa: E731
    K = c["fermi_intensity_100mev"] / torch.trapezoid(shape(e), e)
    return torch.where((E >= lo) & (E <= hi), K * shape(E), torch.nan)


def observed_intensity(E: torch.Tensor, c: dict) -> torch.Tensor:
    """(E,) the measured diffuse background where either instrument covers E (NaN in the gap 30-100 MeV)."""
    return torch.where(torch.isnan(comptel_intensity(E, c)), fermi_intensity(E, c), comptel_intensity(E, c))


def minimum_domain_size(c: dict, speed_factor: float, z_max: float, z_min: float = 20.0) -> dict:
    """Smallest domain size the background allows. The flux is exactly proportional to 1/d, so computing it once for
    d = 1 Mpc gives d_min = 1 Mpc x max_E (predicted / observed) over the energies the instruments cover."""
    E = torch.logspace(math.log10(0.8), math.log10(8.2e5), 300, dtype=torch.float64)
    J1 = annihilation_intensity(E, torch.ones(1, dtype=torch.float64), c, speed_factor, z_max, z_min)[0]
    ratio = J1 / observed_intensity(E, c)
    ratio = torch.nan_to_num(ratio, nan=0.0)
    i = int(ratio.argmax())
    return {"d_min_mpc": float(ratio[i]), "E_worst_mev": float(E[i]), "E": E, "ratio_1mpc": ratio}


# --- criterion 3: fine-tuning of one random worldline ---------------------------------------------------------------


def random_net_charge(M: torch.Tensor, B: int, generator) -> torch.Tensor:
    """(len(M), B) net charge of M crossings with independent random signs +-1 (B samples each)."""
    signs = torch.randint(0, 2, (B, int(M.max())), generator=generator, dtype=torch.int8) * 2 - 1
    run = torch.cumsum(signs, 1, dtype=torch.int32)  # (B, M_max) partial sums
    return run[:, M - 1].T.long()


def patch_net_charge_balanced(T: int, M: torch.Tensor, B: int, generator) -> torch.Tensor:
    """(len(M), B) net charge of the first M of T crossings of a balanced loop (T/2 of each sign, random order)."""
    order = torch.rand(B, T, generator=generator).argsort(1)  # (B, T) random permutations
    signs = torch.where(order < T // 2, 1, -1)
    return torch.cumsum(signs, 1)[:, M - 1].T.long()


def bridge_variance(M: torch.Tensor, T: int) -> torch.Tensor:
    """Variance of the patch net charge for a balanced total: M (T - M) / (T - 1) (sampling without replacement)."""
    return M * (T - M) / (T - 1)


def log10_tail(Q: torch.Tensor, M: torch.Tensor) -> torch.Tensor:
    """log10 P(net >= Q) for M random signs, from Cramer's rate function: -M KL((1 + q)/2 || 1/2) / ln 10, q = Q/M.
    KL = [(1+q) ln(1+q) + (1-q) ln(1-q)] / 2, with the series q^2/2 + q^4/12 for tiny q (no cancellation)."""
    q = Q / M
    exact = 0.5 * (torch.special.xlogy(1 + q, 1 + q) + torch.special.xlogy(1 - q, 1 - q))
    series = q**2 / 2 + q**4 / 12 + q**6 / 30
    kl = torch.where(q.abs() < 1e-3, series, exact)
    return -M * kl / math.log(10)


def log10_tail_exact(Q: torch.Tensor, M: torch.Tensor) -> torch.Tensor:
    """log10 P(net >= Q), exact binomial tail: sum over k >= (M + Q)/2 of C(M, k) 2^-M (lgamma, logsumexp)."""
    out = []
    for q, m in zip(Q.tolist(), M.tolist(), strict=True):
        k = torch.arange(math.ceil((m + q) / 2), int(m) + 1, dtype=torch.float64)
        logc = math.lgamma(m + 1) - torch.lgamma(k + 1) - torch.lgamma(m - k + 1) - m * math.log(2)
        out.append(torch.logsumexp(logc, 0) / math.log(10))
    return torch.stack(out)


# --- criterion 4: one worldline across the CPT mirror ---------------------------------------------------------------


def crossings(t: torch.Tensor, levels: torch.Tensor, closed: bool):
    """(B, L) forward (particle) and backward (antiparticle) crossings of every slice t = u (segments s summed)."""
    a = t
    b = torch.roll(t, -1, 1) if closed else t[:, 1:]
    a = a if closed else t[:, :-1]
    a, b = a[..., None], b[..., None]  # (B, S, 1)
    up = ((a < levels) & (levels <= b)).sum(1)
    down = ((b < levels) & (levels <= a)).sum(1)
    return up, down


def charges_across_mirror(eta: torch.Tensor, ours: torch.Tensor, period: float):
    """Net charges on our slices eta = u (0 < u < period/2) and on their mirror images eta = -u, each read with its
    own arrow of time: forward in eta is a particle for us; in the CPT mirror the arrow points to -eta, so the sign
    flips. eta (W, S + 1) is unwrapped conformal time on a circle of length `period`; every copy u + j period counts."""
    j = torch.arange(math.floor(eta.min() / period) - 1, math.ceil(eta.max() / period) + 2, dtype=torch.float64)
    out = []
    for u in (ours, -ours):
        ext = (u[None, :] + period * j[:, None]).reshape(-1)  # (J L,)
        up, down = crossings(eta, ext, closed=False)
        out.append((up - down).view(len(eta), len(j), -1).sum(1))
    q_ours, q_mirror_global = out
    return q_ours, -q_mirror_global  # the mirror reads time backwards


def masses(m_light: torch.Tensor, ordering: str, c: dict):
    """(m1, m2, m3) in eV from the lightest mass and the NuFIT splittings."""
    if ordering == "normal":
        m1 = m_light
        return m1, torch.sqrt(m1**2 + c["dm21"]), torch.sqrt(m1**2 + c["dm31_no"])
    m3 = m_light
    m2 = torch.sqrt(m3**2 - c["dm32_io"])
    return torch.sqrt(m2**2 - c["dm21"]), m2, m3


def sum_mnu(m_light: torch.Tensor, ordering: str, c: dict) -> torch.Tensor:
    """Sum of the three neutrino masses (eV), what cosmology measures."""
    m1, m2, m3 = masses(m_light, ordering, c)
    return m1 + m2 + m3


def m_beta(m_light: torch.Tensor, ordering: str, c: dict) -> torch.Tensor:
    """Effective electron-neutrino mass KATRIN measures: m_beta^2 = sum_i |U_ei|^2 m_i^2 (PDG mixing angles)."""
    m1, m2, m3 = masses(m_light, ordering, c)
    s12, s13 = c["s12sq"], c["s13sq"]
    return torch.sqrt((1 - s12) * (1 - s13) * m1**2 + s12 * (1 - s13) * m2**2 + s13 * m3**2)


def allowed(x: torch.Tensor, upper_limit: float) -> torch.Tensor:
    """A point is allowed by an upper limit when it lies strictly below it."""
    return x < upper_limit
```

## 3. `tensors.py` — PyTorch tensors

```python
%%writefile problems/cosmology/projection_antimatter_universe/tensors.py
"""Concrete tensors: the cited numbers, worldlines in a space of `dims` dimensions, and the photon-energy grid.

A worldline is a path X(s) = (t, x_1, ..., x_{dims-1}); only its time coordinate decides how a slice reads it.
For the CPT mirror the time coordinate is conformal time eta on a circle: eta = 0 is the Big Bang, 0 < eta < 1 our
side, -1 < eta < 0 the mirror, and eta = +-1 the future boundaries (glued together: an extra assumption).
"""

import csv
import math
from pathlib import Path

import torch

DATA = Path(__file__).parent / "data"


def cited() -> dict:
    """data/antimatter_criteria.csv: quantity -> value (see the file header for every source)."""
    with open(DATA / "antimatter_criteria.csv", encoding="utf-8") as f:
        rows = list(csv.DictReader(r for r in f if not r.startswith("#")))
    return {r["quantity"]: float(r["value"]) for r in rows}


def energies(device=None) -> torch.Tensor:
    """(E,) photon energies today, 0.1 MeV to 1 TeV, log spaced."""
    return torch.logspace(-1, 6, 140, dtype=torch.float64, device=device)


def closed_loops(B: int, S: int, modes: int, dims: int, generator) -> torch.Tensor:
    """(B, S, dims) random closed Fourier worldlines X_bsi = sum_k (a_bik cos ks + c_bik sin ks) / k, each scaled so
    that its time coordinate stays inside (-0.95, 0.95)."""
    s = 2 * math.pi * torch.arange(S, dtype=torch.float64) / S
    k = torch.arange(1, modes + 1, dtype=torch.float64)
    coef = torch.randn(B, dims, modes, 2, generator=generator, dtype=torch.float64) / k[:, None]
    basis = torch.stack([torch.cos(k[:, None] * s), torch.sin(k[:, None] * s)], -1)  # (K, S, 2)
    X = torch.einsum("bikc,ksc->bsi", coef, basis)  # repeated k (harmonic) and c (cos/sin) => summed
    return 0.95 * X / X[..., :1].abs().amax(1, keepdim=True)


def winding_conformal_loop(w: torch.Tensor, S: int, wiggle: float, generator) -> torch.Tensor:
    """(W, S + 1) unwrapped conformal time of loops that wind w times round the eta circle of length 2:
    eta(s) = -1 + 2 w s / 2 pi + wiggle * (periodic zig-zag); the last sample is the first plus 2 w."""
    s = 2 * math.pi * torch.arange(S + 1, dtype=torch.float64) / S
    k = torch.arange(1, 6, dtype=torch.float64)
    coef = torch.randn(len(w), 5, 2, generator=generator, dtype=torch.float64) / k[:, None]
    basis = torch.stack([torch.cos(k[:, None] * s), torch.sin(k[:, None] * s)], -1)
    zig = torch.einsum("wkc,ksc->ws", coef, basis)  # periodic, so the loop closes on the circle
    return -1.0 + w.double()[:, None] * 2.0 * s / (2 * math.pi) + wiggle * zig
```

## 4. `solve.py` — solver

```python
%%writefile problems/cosmology/projection_antimatter_universe/solve.py
"""Solvers: evaluate each criterion against the cited data and collect the verdict table."""

import math

import torch

from . import equation, tensors

BOX_GPC = 30.0  # side of the domain picture, a bit more than the observable universe (diameter 28.6 Gpc)


def nearby(c: dict) -> dict:
    """Criterion 1: antinuclei in cosmic rays and gamma rays from clusters."""
    f_max = equation.cluster_antimatter_limit(c)
    return {"antihelium": c["antihelium_to_helium"], "f_cluster": f_max,
            "photons_100": equation.photons_above(100.0)}  # fmt: skip


def domain_sweep(c: dict, speed_factor: float, z_max: float, d_mpc: torch.Tensor) -> dict:
    """Criterion 2: predicted background for every domain size d, the observed background, and the minimum d."""
    E = tensors.energies()
    J = equation.annihilation_intensity(E, d_mpc, c, speed_factor, z_max)  # (D, E)
    res = equation.minimum_domain_size(c, speed_factor, z_max)
    ratio = res["d_min_mpc"] / d_mpc  # (D,) worst predicted / observed: exactly 1/d
    return {"E": E, "J": J, "comptel": equation.comptel_intensity(E, c), "fermi": equation.fermi_intensity(E, c),
            "d": d_mpc, "ratio": ratio, "d_min_mpc": res["d_min_mpc"], "E_worst_mev": res["E_worst_mev"]}  # fmt: skip


def domain_pictures(d_mpc: torch.Tensor, n_pix: int, generator) -> torch.Tensor:
    """(D, N, N) matter/antimatter maps over a BOX_GPC box for the domain sizes d (one noise, so domains grow)."""
    noise = torch.randn(n_pix, n_pix, generator=generator, dtype=torch.float64)
    d_pix = d_mpc / (1e3 * BOX_GPC / n_pix)
    return equation.domains(noise, d_pix)


def fine_tuning(c: dict, generator) -> dict:
    """Criterion 3: how unlikely a random single worldline makes our patch. N_B from eta and the horizon volume."""
    eta = 273.9e-10 * c["planck_omega_b_h2"]
    R = c["horizon_gpc"] * 1e3 * equation.MPC_CM
    N_gamma = c["n_gamma_cm3"] * 4 / 3 * math.pi * R**3
    N_B = eta * N_gamma
    t = lambda x: torch.tensor([x], dtype=torch.float64)  # noqa: E731
    today = equation.log10_tail(t(N_B), t(N_B)).item()  # all ~10^80 crossings the same sign (no antimatter seen)
    early = equation.log10_tail(t(N_B), t(N_gamma)).item()  # excess 10^80 among ~10^89 pairs (early universe)
    M = torch.tensor([10, 100, 1000])
    Q = equation.random_net_charge(M, 4000, generator)
    return {"N_B": N_B, "N_gamma": N_gamma, "log10_today": today, "log10_early": early, "M": M,
            "var": Q.double().var(1)}  # fmt: skip


def mirror(c: dict, dims: int, winding: int, generator) -> dict:
    """Criterion 4: one winding worldline across the CPT mirror, and the neutrino and r tests."""
    w = torch.tensor([winding])
    eta = tensors.winding_conformal_loop(w, 1200, wiggle=0.35, generator=generator)[0]  # (S + 1,)
    s = torch.linspace(0, 2 * math.pi, len(eta), dtype=torch.float64)
    k = torch.arange(1, 4, dtype=torch.float64)
    coef = torch.randn(dims - 1, 3, 2, generator=generator, dtype=torch.float64) / k[:, None]
    basis = torch.stack([torch.cos(k[:, None] * s), torch.sin(k[:, None] * s)], -1)
    space = torch.einsum("ikc,ksc->si", coef, basis)  # (S + 1, dims - 1): the other coordinates; only eta matters
    levels = torch.linspace(0.05, 0.95, 19, dtype=torch.float64)
    q_ours, q_mirror = equation.charges_across_mirror(eta[None], levels, period=2.0)
    zero = torch.zeros(1, dtype=torch.float64)
    m = torch.logspace(-4, 0, 200, dtype=torch.float64)
    nu = {
        "sum_no0": equation.sum_mnu(zero, "normal", c).item(),
        "sum_io0": equation.sum_mnu(zero, "inverted", c).item(),
        "mb_no0": equation.m_beta(zero, "normal", c).item(),
        "mb_io0": equation.m_beta(zero, "inverted", c).item(),
        "m": m, "sum_no": equation.sum_mnu(m, "normal", c), "sum_io": equation.sum_mnu(m, "inverted", c),
    }  # fmt: skip
    return {"eta": eta, "x": space[:, 0], "levels": levels, "q_ours": q_ours[0], "q_mirror": q_mirror[0],
            "w": winding, **nu}  # fmt: skip


def verdicts(c: dict, near: dict, dom: dict, ft: dict, mi: dict) -> list[dict]:
    """The verdict table: what each criterion requires, its status, and what would falsify it."""
    no_ok = mi["sum_no0"] < c["desi_dr2_sum_mnu"]
    io_ok = mi["sum_io0"] < c["desi_dr2_sum_mnu"]
    return [
        {"criterion": "1 nothing mixed in nearby", "status": "satisfied",
         "detail": f"antiHe/He < {near['antihelium']:.1e}; Coma antiproton fraction < {near['f_cluster']:.1e}",
         "falsify": "one antihelium nucleus (AMS-02) or annihilation gamma rays from a cluster"},
        {"criterion": "2 no domains inside the horizon", "status": "constrained",
         "detail": f"domains must be >= {dom['d_min_mpc'] / 1e3:.0f} Gpc (horizon radius {c['horizon_gpc']} Gpc)",
         "falsify": "a diffuse MeV excess or CMB distortion pointing to nearer walls; untestable beyond the horizon"},
        {"criterion": "3 not fine-tuned", "status": "fails without a mechanism",
         "detail": f"log10 P = {ft['log10_early']:.1e} (early) to {ft['log10_today']:.1e} (today)",
         "falsify": "not a test but a cost: only a mechanism (a CPT boundary at t = 0 plus winding) removes it"},
        {"criterion": "4a lightest neutrino massless", "status": "constrained (consistent)",
         "detail": f"sum m = {mi['sum_no0']:.4f} eV (NO, allowed: {no_ok}), {mi['sum_io0']:.4f} eV (IO, allowed: "
                   f"{io_ok}) vs DESI DR2 < {c['desi_dr2_sum_mnu']} eV",
         "falsify": "a measured lightest mass > 0 (e.g. sum m above the floor + m_light, or KATRIN/Project 8 signal)"},
        {"criterion": "4b dark matter = 4.8e8 GeV right-handed neutrino", "status": "untestable now",
         "detail": f"m_N = {c['bft_dm_mass_gev']:.1e} GeV, far beyond colliders; no direct-detection signal expected",
         "falsify": "dark matter identified as something else (e.g. a WIMP or axion signal)"},
        {"criterion": "4c no primordial gravitational waves", "status": "consistent",
         "detail": f"r = 0 vs BICEP/Keck r < {c['bicep_keck_r']}",
         "falsify": f"primordial B-modes with r >> {c['future_sigma_r']:.0e} (LiteBIRD, Simons Observatory)"},
    ]  # fmt: skip
```

## 5. `plot.py` — 2D/3D plot → GIF

```python
%%writefile problems/cosmology/projection_antimatter_universe/plot.py
"""Animate the criteria: growing matter/antimatter domains and their gamma rays, the worldline across the CPT mirror,
and the neutrino-mass and tensor-to-scalar axes with their bounds."""

from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import torch
from matplotlib.collections import LineCollection
from matplotlib.colors import ListedColormap
from PIL import Image

from problems._common.plotting import export_plot_data
from problems._common.progress import report

from . import equation, solve

BLUE, RED, AMBER, GREEN, GREY, INK = "#2563eb", "#dc2626", "#f59e0b", "#16a34a", "#6b7280", "#111827"
LIGHT_BLUE, LIGHT_RED = "#bfdbfe", "#fecaca"


def _wrapped(eta: np.ndarray) -> np.ndarray:
    """Conformal time folded back onto the circle [-1, 1)."""
    return (eta + 1) % 2 - 1


def _cuts(e: np.ndarray, x: np.ndarray, u: float):
    """x positions where the wrapped worldline crosses eta = u, split by direction in eta (wrap jumps skipped)."""
    a, b = e[:-1], e[1:]
    ok = np.abs(b - a) < 1
    up = ok & (a < u) & (u <= b)
    down = ok & (b < u) & (u <= a)
    frac = (u - a) / np.where(b != a, b - a, 1)
    xc = x[:-1] + frac * (x[1:] - x[:-1])
    return xc[up], xc[down]


def animate(c, dom, maps, mi, dims, path: Path) -> Path:
    n = len(dom["d"])
    fig = plt.figure(figsize=(11.5, 8.2))
    gs = fig.add_gridspec(2, 2, height_ratios=[1.05, 1], hspace=0.38, wspace=0.26)
    ax_m, ax_g, ax_w = fig.add_subplot(gs[0, 0]), fig.add_subplot(gs[0, 1]), fig.add_subplot(gs[1, 0])
    sub = gs[1, 1].subgridspec(2, 1, hspace=1.0)
    ax_s, ax_r = fig.add_subplot(sub[0]), fig.add_subplot(sub[1])

    # --- domains of matter and antimatter, glowing walls, our horizon ---
    L = solve.BOX_GPC
    cmap = ListedColormap([LIGHT_RED, AMBER, LIGHT_BLUE])  # -1 antimatter, 0 wall, +1 matter

    def picture(i: int) -> np.ndarray:
        s = maps[i].numpy()
        wall = np.zeros_like(s, dtype=bool)
        for sh, ax in ((1, 0), (1, 1)):  # one-sided: walls one pixel thick
            wall |= s != np.roll(s, sh, ax)
        return np.where(wall, 0, s)

    img = ax_m.imshow(picture(0), cmap=cmap, vmin=-1.5, vmax=1.5, extent=(-L / 2, L / 2, -L / 2, L / 2),
                      origin="lower", interpolation="nearest")  # fmt: skip
    th = np.linspace(0, 2 * np.pi, 200)
    R = c["horizon_gpc"]
    ax_m.plot(R * np.cos(th), R * np.sin(th), "--", color=INK, lw=1.2)
    ax_m.text(0, -R + 1.2, "our horizon (14.3 Gpc)", ha="center", fontsize=7, color=INK,
              bbox={"facecolor": "white", "edgecolor": "none", "pad": 1})  # fmt: skip
    ax_m.set_xlim(-L / 2, L / 2)
    ax_m.set_ylim(-L / 2, L / 2)
    ax_m.set_xlabel("comoving $x$ (Gpc)")
    ax_m.set_ylabel("comoving $y$ (Gpc)")
    for col, lab in ((LIGHT_BLUE, "matter domain"), (LIGHT_RED, "antimatter domain"), (AMBER, "annihilation wall")):
        ax_m.plot([], [], "s", color=col, ms=8, label=lab)
    ax_m.legend(fontsize=7, loc="upper right", framealpha=0.9)
    title_m = ax_m.set_title("", fontsize=9)

    # --- the diffuse gamma-ray background: predicted for this d against the data ---
    E = dom["E"].numpy()
    e2 = E**2
    ax_g.loglog(E, e2 * dom["comptel"].numpy(), color=INK, lw=3, label="COMPTEL fit (0.8-30 MeV)")
    ax_g.loglog(E, e2 * dom["fermi"].numpy(), color=GREY, lw=3, label="Fermi-LAT IGRB fit (0.1-820 GeV)")
    J = dom["J"].numpy()
    curve = ax_g.loglog(E, np.where(J[0] > 0, e2 * J[0], np.nan), color=AMBER, lw=2,
                        label="wall annihilation, $z$ = 20 to recombination")[0]  # fmt: skip
    ax_g.set_xlim(E[0], E[-1])
    ax_g.set_ylim(1e-6, 3.0)
    ax_g.set_xlabel("photon energy today $E$ (MeV)")
    ax_g.set_ylabel(r"$E^2\,dJ/dE$ (MeV cm$^{-2}$ s$^{-1}$ sr$^{-1}$)")
    ax_g.legend(fontsize=7, loc="upper right")
    title_g = ax_g.set_title("", fontsize=9)

    # --- one worldline winding through the Big Bang into the CPT mirror ---
    eta, x = mi["eta"].numpy(), mi["x"].numpy()
    e = _wrapped(eta)
    pts = np.stack([x, e], 1)
    seg = np.stack([pts[:-1], pts[1:]], 1)
    jump = np.abs(np.diff(e)) > 1
    fwd = np.diff(eta) > 0
    ours = (e[:-1] + e[1:]) / 2 > 0
    particle = fwd == ours  # forward in eta reads as matter on our side, as antimatter in the mirror
    colors = np.where(particle, BLUE, RED)
    lc = LineCollection(seg[~jump], colors=colors[~jump], linewidths=1.5)
    ax_w.axhspan(0, 1, color="#eff6ff", lw=0)
    ax_w.axhspan(-1, 0, color="#fef2f2", lw=0)
    ax_w.add_collection(lc)
    ax_w.axhline(0, color=INK, lw=1.5)
    pad = 0.1 * (x.max() - x.min())
    ax_w.set_xlim(x.min() - pad, x.max() + pad)
    ax_w.set_ylim(-1, 1)
    ax_w.text(x.min(), 0.88, "our universe: arrow of time up", fontsize=7, color="#1e40af")
    ax_w.text(x.min(), -0.95, "CPT mirror: arrow of time down", fontsize=7, color="#b91c1c")
    ax_w.text(x.max(), 0.03, "Big Bang $\\eta = 0$", fontsize=7, ha="right")
    ax_w.set_xlabel("space $x$, one of $D-1$ axes (a.u.)")
    ax_w.set_ylabel(r"conformal time $\eta$ (a.u.; $\pm1$ glued)")
    sl_o = ax_w.axhline(0.5, color=AMBER, lw=1.6)
    sl_m = ax_w.axhline(-0.5, color=AMBER, lw=1.6, ls="--")
    dots_b = ax_w.plot([], [], "o", color=BLUE, ms=6, mec="white")[0]
    dots_r = ax_w.plot([], [], "o", color=RED, ms=6, mec="white")[0]
    count = ax_w.text(0.99, 0.6, "", transform=ax_w.transAxes, ha="right", fontsize=7,
                      bbox={"facecolor": "white", "edgecolor": "#d1d5db"})  # fmt: skip
    ax_w.set_title(f"one worldline in $D = {dims}$ winding $w = {mi['w']}$ times: blue = matter, red = antimatter,\n"
                   "each side read with its own arrow of time (author's alignment)", fontsize=9)  # fmt: skip

    # --- the sum of neutrino masses ---
    ax_s.set_xscale("log")
    ax_s.set_xlim(0.03, 6.0)
    ax_s.set_ylim(0, 1)
    ax_s.set_yticks([])
    bounds = [(c["desi_dr2_sum_mnu"], "DESI DR2", 0.9), (c["desi_2024_sum_mnu"], "DESI 2024", 0.62),
              (c["planck_sum_mnu"], "Planck 2018", 0.9), (3 * c["katrin_2025"], "KATRIN 2025\n($3m_\\beta$)", 0.62),
              (3 * c["katrin_2022"], "KATRIN 2022\n($3m_\\beta$)", 0.9)]  # fmt: skip
    for b, lab, yy in bounds:
        ax_s.axvline(b, color=GREY, lw=1)
        ax_s.text(b * 1.03, yy, lab, fontsize=6, va="top", color=GREY)
    ax_s.axvspan(c["desi_dr2_sum_mnu"], 6.0, color="#f3f4f6", lw=0)
    ax_s.plot([mi["sum_no0"]], [0.22], "^", color=GREEN, ms=9)
    ax_s.plot([mi["sum_io0"]], [0.22], "v", color=AMBER, ms=9)
    nl = chr(10)
    ax_s.text(mi["sum_no0"], 0.02, f"normal{nl}{mi['sum_no0']:.3f} eV", fontsize=6, ha="right", color=GREEN)
    ax_s.text(mi["sum_io0"] * 1.04, 0.02, f"inverted{nl}{mi['sum_io0']:.3f} eV", fontsize=6, color="#b45309")
    ax_s.text(0.031, 0.62, f"massless{nl}lightest " + r"$\nu$:", fontsize=6, color=INK)
    toy_s = ax_s.plot([], [], "D", ms=7, mec=INK)[0]
    ax_s.set_xlabel(r"sum of neutrino masses $\Sigma m_\nu$ (eV)")
    title_s = ax_s.set_title("", fontsize=8)

    # --- the tensor-to-scalar ratio ---
    ax_r.set_xscale("log")
    ax_r.set_xlim(1e-4, 1.0)
    ax_r.set_ylim(0, 1)
    ax_r.set_yticks([])
    ax_r.axvspan(c["bicep_keck_r"], 1.0, color="#f3f4f6", lw=0)
    ax_r.axvline(c["bicep_keck_r"], color=GREY, lw=1)
    ax_r.text(c["bicep_keck_r"] * 1.1, 0.85, f"BICEP/Keck $r < {c['bicep_keck_r']}$", fontsize=6, va="top", color=GREY)
    ax_r.axvline(2 * c["future_sigma_r"], color=GREY, lw=1, ls="--")
    ax_r.text(2 * c["future_sigma_r"] * 1.1, 0.85, "future $2\\sigma(r)$", fontsize=6, va="top", color=GREY)
    ax_r.annotate("CPT universe: $r = 0$", xy=(1e-4, 0.3), xytext=(2.2e-4, 0.3), fontsize=7, va="center",
                  arrowprops={"arrowstyle": "->", "color": GREEN}, color=GREEN)  # fmt: skip
    toy_r = ax_r.plot([], [], "D", ms=7, mec=INK)[0]
    ax_r.set_xlabel("tensor-to-scalar ratio $r$ (dimensionless)")
    title_r = ax_r.set_title("", fontsize=8)

    levels = mi["levels"].numpy()
    toy_m = np.concatenate([[0.0], np.logspace(-3, np.log10(0.5), n - 1)])
    toy_rr = np.logspace(-4, np.log10(0.3), n)
    toy_sum = equation.sum_mnu(  # the toy points use the same functions as the tests
        torch.tensor(toy_m), "normal", c
    ).numpy()

    def update(i: int):
        d = dom["d"][i].item()
        img.set_data(picture(i))
        title_m.set_text(f"domain size $d$ = {d / 1e3:.2f} Gpc")
        curve.set_ydata(np.where(J[i] > 0, e2 * J[i], np.nan))
        ratio = dom["ratio"][i].item()
        verdict = "excluded" if ratio > 1 else "allowed"
        title_g.set_text(f"worst predicted / observed = {ratio:.2g}: {verdict}\n"
                         f"($d_{{min}}$ = {dom['d_min_mpc'] / 1e3:.0f} Gpc; flux $\\propto 1/d$)")  # fmt: skip
        k = (i * 3) % (2 * len(levels) - 2)
        u = levels[k if k < len(levels) else 2 * len(levels) - 2 - k]
        sl_o.set_ydata([u])
        sl_m.set_ydata([-u])
        bo, ro = _cuts(e, x, u)  # ours: forward = matter
        fm, bm = _cuts(e, x, -u)  # mirror: forward in eta = antimatter
        dots_b.set_data(np.concatenate([bo, bm]), np.concatenate([np.full_like(bo, u), np.full_like(bm, -u)]))
        dots_r.set_data(np.concatenate([ro, fm]), np.concatenate([np.full_like(ro, u), np.full_like(fm, -u)]))
        count.set_text(f"our slice: {len(bo)} matter, {len(ro)} anti, net {len(bo) - len(ro):+d}\n"
                       f"mirror slice: {len(bm)} matter, {len(fm)} anti, net {len(bm) - len(fm):+d}\n"
                       f"total {len(bo) - len(ro) + len(bm) - len(fm):+d}")  # fmt: skip
        s_ok = toy_sum[i] < c["desi_dr2_sum_mnu"]
        toy_s.set_data([toy_sum[i]], [0.55])
        toy_s.set_color(GREEN if s_ok else RED)
        title_s.set_text(f"toy point: lightest mass {toy_m[i]:.3f} eV (normal) -> "
                         f"{'allowed' if s_ok else 'excluded'} by DESI DR2")  # fmt: skip
        r_ok = toy_rr[i] < c["bicep_keck_r"]
        toy_r.set_data([toy_rr[i]], [0.3])
        toy_r.set_color(GREEN if r_ok else RED)
        title_r.set_text(f"toy point: $r$ = {toy_rr[i]:.1e} -> {'allowed' if r_ok else 'excluded'} by BICEP/Keck")

    fig.subplots_adjust(left=0.07, right=0.98, bottom=0.07, top=0.94)
    return save_gif_shared_palette(fig, update, list(range(n)), path, fps=6)


def save_gif_shared_palette(fig, update, frames: list[int], path: Path, fps: int = 8) -> Path:
    """Like plotting.save_gif, but every frame uses ONE palette built from a strip of all frames, so colours stay
    fixed through the animation (Pillow otherwise picks a new 256-colour palette per frame)."""
    fig.set_dpi(80)
    shots = []
    for i, f in enumerate(frames):
        update(f)
        fig.canvas.draw()
        shots.append(Image.fromarray(np.asarray(fig.canvas.buffer_rgba())[..., :3]))
        report(i + 1, len(frames), "frames")
    export_plot_data(fig, path)  # numbers behind the final frame (Results page)
    plt.close(fig)
    sample = shots[:: max(1, len(shots) // 12)]
    strip = Image.new("RGB", (sample[0].width, sample[0].height * len(sample)))
    for k, im in enumerate(sample):
        strip.paste(im, (0, k * im.height))
    palette = strip.quantize(colors=255, method=Image.Quantize.MEDIANCUT)
    frames_p = [im.quantize(palette=palette, dither=Image.Dither.NONE) for im in shots]
    tmp = path.with_name(f".{path.stem}.tmp{path.suffix}")
    frames_p[0].save(tmp, save_all=True, append_images=frames_p[1:], duration=int(1000 / fps), loop=0)
    return tmp.replace(path)
```
