# How many dimensions? The budget of the projection hypothesis

*cosmology · advanced* · [open in Tensorizer](https://jackyko1991.github.io/tensorizer/#/cosmology/projection_dimension_budget)

A speculative hypothesis, not established physics: how many hidden dimensions does the projection idea need, and how many do experiments allow? The quantum side, from step 1, needs one hidden complex dimension per outcome and $2^N$ for $N$ entangled particles, but these live in Hilbert space and need not be directions in space. On the gravity side, $n$ compact dimensions of radius $R$ change Newton's law below $R$. Torsion balances, the LHC and supernova and neutron-star bounds fill an $(n, R)$ map bound by bound. Every $n$ survives, but only if $R$ is small: below 38.6 μm for $n = 1$ and $2\times10^{-10}$ m for $n = 2$. Constrained, not ruled out.

**Vector form**

$$\nabla\cdot\mathbf g = -|S^{n+2}|\,G_D\,\rho\ \Rightarrow\ \oint_{S^{n+2}}\nabla\phi\cdot d\mathbf A = |S^{n+2}|\,G_D M,\qquad g(r) = \sum_{\mathbf k\in\mathbb Z^n}\frac{G_D M\,r}{(r^2 + |2\pi R\mathbf k|^2)^{(n+3)/2}} \to \begin{cases} G_D M/r^{n+2} & r\ll R\\ G_4 M/r^2 & r\gg R\end{cases},\qquad G_4 = \frac{|S^{n+2}|\,G_D}{4\pi(2\pi R)^n},\qquad M_{Pl}^2 = 8\pi R^n M_D^{n+2},\qquad V = -\frac{G M m}{r}\big(1 + \tfrac{8n}{3}e^{-r/R}\big),\qquad \text{allowed} \iff R < R_{max}(n) = \min_b R_b(n),\qquad D_Q \geq 2^N$$

**Index (tensor) form**

$$F_r = |S^{n+2}|\,r^{n+2}\,\langle \partial_i\phi\,u_{pi}\rangle_p,\qquad g_r = \sum_I \frac{r}{(r^2 + \rho_I^2)^{(n+3)/2}},\qquad \mathrm{excl}^{(b)}_{nR} = \big[R > R_b(n)\big],\qquad R_b(n) = \Big(\frac{M_{Pl}^2}{8\pi M_{b}^{n+2}}\Big)^{1/n},\qquad \mathrm{excl}_{nR} = \bigvee_b \mathrm{excl}^{(b)}_{nR},\qquad R_{max}(n) = \min_b R_b(n)$$

| Symbol | Code | Shape | Meaning |
|---|---|---|---|
| $F_r$ | `einsum('rpi,pi->rp', grad, u).mean(1) * area * r**(n+2)` | (R,) | repeated $i$ ⇒ summed: the outward pull at random points of $S^{n+2}$ (autograd); the flux is the same at every radius |
| $g_r$ | `image_force(r, n, K)` | (R,) | the mass and its images round the $n$ circles, summed over the image index $I$: $1/r^{n+2}$ inside, $G_4/r^2$ outside |
| $R_b(n)$ | `_limit(n, row)` | (N,) | largest radius one cited bound allows: $\lambda_{max}$ for torsion balances, the ADD radius at $M_D = M_b$ for collider and astrophysical limits |
| $\mathrm{excl}_{nR}$ | `exclusion_masks(n[:, None], R[None, :], rows)` | (N, R) | one boolean map per bound family, broadcast over the whole grid |
| $R_{max}(n)$ | `r_max(n, rows)` | (N,) | the dimension budget: compact dimensions are allowed only below this radius |
| $M_D$ | `add_md_tev(n, R)` | (N, R) | fundamental scale of gravity for $n$ dimensions of radius $R$ (Giudice–Rattazzi–Wells convention, as the PDG uses) |
| $D_Q$ | `hilbert_log2_dim(N)` | (N,) | Hilbert-space dimension $2^N$ that $N$ entangled qubits need (step 1) |

Status: speculative; constrained, not ruled out. Step 3 of the Projection Hypothesis (the author's own speculative idea, see the series description; not established physics), for postulate P5: quantum effects need some $d_Q$ and gravity some $d_G$ dimensions, estimated, never set by hand. Step 1 (cosmology/projection_postulate) found that the projection race needs exactly one hidden complex dimension per outcome; step 2 (cosmology/projection_one_particle) sliced one worldline; step 4 (cosmology/projection_singularity_fold) treats singularities as folds of the projection. No $n$ is assumed: the map scans $n = 0$ to $n_{max}$ and $R$ from the Planck length to 1 m. Known physics first. (a) Quantum side: the race reproduces the Born rule with $d + K$ dimensions for $K$ outcomes, and $N$ entangled two-level systems need a Hilbert space of dimension $2^N$; standard quantum mechanics uses the same space, and its dimensions are not directions in space. If they were read as spatial directions, $N$ qubits would need $2^{N+1} - 3$ extra real dimensions (5 for a Bell pair), a number that would change every time particles become entangled, which no fixed spatial manifold does. (b) Gravity side: in $3 + n$ space dimensions Gauss's law gives a pull $G_D M/r^{n+2}$ (checked by autograd fluxes through $S^{n+2}$ for $n = 0..6$). If the $n$ dimensions are circles of radius $R$, the image sum turns this into Newton's $G_4 M/r^2$ with $G_4 = |S^{n+2}|G_D/4\pi(2\pi R)^n$ beyond $R$ (checked for $n = 1, 2, 3$). Just outside, the first Kaluza–Klein graviton level gives a Yukawa term with $\alpha = 8n/3$ ($16/3$ for a 2-torus, as in the PDG review) and $\lambda = R$. The ADD relation of Arkani-Hamed, Dimopoulos and Dvali (1998), in the convention $M_{Pl}^2 = 8\pi R^n M_D^{n+2}$ used by the PDG and the LHC experiments, reproduces the PDG numbers: $R = 30$ μm, $n = 2$ gives $M_D = 4.0$ TeV, and $M_D = 1$ TeV needs $10^{12}$ m, 0.5 mm, ..., 0.1 MeV$^{-1}$ for $n = 1..6$. Bounds (data/gravity_bounds.csv, cited numbers; the torsion and LHC rows are those bundled in cosmology/hierarchy_extra_dimensions, cited again). Eöt-Wash torsion balances exclude $|\alpha| = 1$ above 56 μm (Kapner et al., PRL 98, 021101, 2007) and above 38.6 μm (Lee et al., PRL 124, 101101, 2020); since $\alpha\geq 1$ for every $n$, $R > 38.6$ μm is excluded for all $n\geq 1$ (conservative: the measured curve keeps falling at larger $\lambda$). For $n = 2$, Tan et al. (PRL 116, 131101, 2016) give $R < 30$ μm. The ATLAS monojet search (PRD 103, 112006, 2021, CC-BY 4.0) gives $M_D > 11.2, 8.5, 7.1, 6.4, 5.9$ TeV for $n = 2..6$. SN1987A cooling gives $M_D > 27$ (2.4) TeV for $n = 2$ (3) (Hanhart et al., PLB 509, 1, 2001), and neutron-star heating by trapped KK gravitons gives $M_D > 1700$ (76) TeV (Hannestad and Raffelt, PRD 67, 125008, 2003), both as quoted in the PDG 2024 review. Each bound becomes a boolean mask on the $(n, R)$ grid, and toy points are checked against hand-worked answers. Result: $R_{max} = 38.6$ μm ($n = 1$), $1.7\times10^{-10}$ m ($n = 2$, neutron stars), $2.6\times10^{-12}$ m ($n = 3$), and LHC radii for $n = 4..6$ ($5\times10^{-13}$ to $2.5\times10^{-15}$ m). For $n\geq 7$ only the torsion bound is bundled; the LHC limits weaken slowly with $n$ and, extrapolated (not done here), would push $R$ towards $1/M_D\sim 10^{-20}$ m. ADD with $M_D = 1$ TeV, the original solution of the hierarchy problem, is excluded for $n = 1..6$. The budget: minimum $n = 0$, because gravity needs no extra spatial dimension and quantum hidden dimensions need not be spatial; maximum none, because the data allow any $n$ if $R < R_{max}(n)$. Which reading the hypothesis takes: the Hilbert-space reading for P1 and P2 (hidden dimensions are state-space dimensions, so $D_Q\geq 2^N$ is no constraint on space), and the geometric reading for P4 and P5 (embedding dimensions; if gravity can move into them they must be compact and below $R_{max}(n)$, otherwise they are coordinates of the embedding, not places one can go). Verdict: constrained, not ruled out, and no new prediction: the hypothesis would only become testable if it fixed $n$ and $R$ near these bounds.

## 1. `main.py` — main loop

```python
%%writefile problems/cosmology/projection_dimension_budget/main.py
"""Main loop: (n, R) grid -> Gauss's law and ADD -> cited bounds -> allowed map -> dimension budget."""

from pathlib import Path

import torch

from problems._common.runtime import cli, device

from . import equation, plot, solve, tensors


def run(params: dict, out_dir: Path) -> Path:
    dev = device()
    gen = torch.Generator().manual_seed(1998)
    n_max, R_points = int(params["n_max"]), int(params["R_points"])
    M_star, qubits = float(params["M_star"]), int(params["qubits"])
    n, R = tensors.grid(n_max, R_points, dev)
    rows = tensors.bounds()

    # 1. gravity in 3 + n dimensions: Gauss's law and the return of Newton's law beyond R
    flux = solve.gauss_check(min(n_max, 6), gen)
    print(f"Gauss  flux / |S^(n+2)| for n = 0..{min(n_max, 6)} at r = 0.5, 1, 2: max deviation "
          f"{(flux - 1).abs().max():.1e}")  # fmt: skip
    fc = solve.force_curves(dev)
    for k, c in fc["curves"].items():
        print(f"       n = {k}: r^2 g / G_4 M = {c[0]:.3g} at r = 0.01 R (expect ~(R/r)^n x const), "
              f"{c[-1]:.6f} at r = 10 R (Newton: 1)")  # fmt: skip

    # 2. ADD relation, checked on the PDG numbers
    n2 = torch.tensor(2.0, dtype=torch.float64)
    print(f"ADD    M_Pl^2 = 8 pi R^n M_D^(n+2): R = 30 um, n = 2 -> M_D = {equation.add_md_tev(n2, 30e-6):.2f} TeV "
          f"(PDG: 4.0)")  # fmt: skip
    nn = n[1:]
    R_star = equation.add_radius_m(nn, torch.tensor(M_star))
    print(f"       M_D = {M_star:g} TeV needs R = " + ", ".join(f"{r:.2g} m (n={int(k)})" for k, r in zip(nn, R_star)))

    # 3. the bounds, one family at a time
    em = solve.exclusion_map(n, R, rows)
    rb = solve.radius_budget(n, rows)
    print("Bounds (cited): " + "; ".join(f"{r['source']} {r['year']}: {r['kind']} {r['value']:g}"
                                         + (f" (n={int(r['n'])})" if r["n"] else "") for r in rows))  # fmt: skip
    print("   n   R_max (m)    M_D at R_max (TeV)   set by")
    for i, k in enumerate(n.tolist()):
        rm = rb["rmax"][i].item()
        if k == 0:
            print("   0   -            -                    no extra dimension: nothing to bound")
            continue
        by = [r["bound"] for r in rows if abs(equation._limit(n[i : i + 1], r)[0].item() - rm) <= 1e-9 * rm]
        print(f"  {int(k):2d}   {rm:.3e}    {rb['md_at_rmax'][i]:.3g}" + " " * 14 + ", ".join(sorted(set(by))))
    tested = [(int(k), bool(r > rm)) for k, r, rm in zip(nn, R_star, rb["rmax"][1:]) if k <= 6]
    print(f"       ADD with M_D = {M_star:g} TeV, n = 1..6: excluded for n = {[k for k, x in tested if x] or 'none'}, "
          f"allowed for n = {[k for k, x in tested if not x] or 'none'}; n >= 7: only the torsion bound is bundled, "
          "so this map does not test it (collider limits would, if extended)")  # fmt: skip
    print(f"       allowed fraction of the grid (log R from the Planck length to 1 m): "
          f"{(~em['excluded'][1:]).double().mean():.3f} for n >= 1")  # fmt: skip

    # 4. the quantum side (step 1) and the verdict
    qb = solve.quantum_budget(3, 2, qubits, dev)
    print(f"Quantum  Born race: d + K = {qb['race']} (d = 3, K = 2); {qubits} entangled qubits: Hilbert D >= "
          f"{qb['hilbert_q']}; if spatial directions: n >= {qb['spatial_q']} extra real dimensions")  # fmt: skip
    print("Budget   minimum n = 0 (Hilbert-space reading: hidden dimensions need not be spatial); maximum n: none "
          "set by data, every n is allowed if R < R_max(n)")  # fmt: skip
    print("Verdict  constrained, not ruled out: compact extra dimensions must be small; no new prediction yet")

    return plot.animate(n, R, em, rb, fc, qb, R_star, M_star, out_dir / "projection_dimension_budget.gif")


if __name__ == "__main__":
    cli(run, __file__)
```

## 2. `equation.py` — physics → tensor translation

```python
%%writefile problems/cosmology/projection_dimension_budget/equation.py
"""Dimension budget -> tensors: Gauss's law in 3 + n dimensions, the ADD relation, the experimental exclusions.

Shapes in comments. Every bound becomes a boolean mask over the (n, R) grid, broadcast from n (N, 1) and R (1, R);
the summed index is the image (lattice) index k in the force of a mass on R^3 x T^n.
"""

import math

import torch

from . import tensors

# ---------- gravity in 3 + n space dimensions ----------


def sphere_area(k: int) -> float:
    """Area of the unit sphere S^k in R^(k+1): 2 pi^((k+1)/2) / Gamma((k+1)/2)."""
    return 2 * math.pi ** ((k + 1) / 2) / math.gamma((k + 1) / 2)


def potential(x: torch.Tensor, n: int) -> torch.Tensor:
    """phi = -G_D M / ((n+1) |x|^(n+1)) in 3 + n space dimensions (G_D M = 1): its gradient is r_hat / r^(n+2)."""
    return -1 / ((n + 1) * x.norm(dim=-1) ** (n + 1))


def gauss_flux(n: int, radii: torch.Tensor, points: int, generator) -> torch.Tensor:
    """(R,) flux of grad phi through spheres S^(n+2) of radius r: |S^(n+2)| r^(n+2) <grad phi . r_hat> (autograd)."""
    u = torch.randn(points, 3 + n, generator=generator, dtype=torch.float64)
    u = u / u.norm(dim=1, keepdim=True)  # (P, 3+n) random outward normals
    x = (radii[:, None, None] * u).requires_grad_(True)  # (R, P, 3+n)
    (grad,) = torch.autograd.grad(potential(x, n).sum(), x)
    radial = torch.einsum("rpi,pi->rp", grad, u)  # repeated i (coordinate) => summed: the outward component
    return sphere_area(n + 2) * radii ** (n + 2) * radial.mean(1)


def image_force(r: torch.Tensor, n: int, K: int) -> torch.Tensor:
    """(R,) pull at distance r along our 3 dimensions from a mass on R^3 x T^n (circles of radius R = 1):
    the mass and its images at 2 pi k, k in [-K, K]^n, each pulling with 1 / dist^(n+2) times r / dist."""
    axis = torch.arange(-K, K + 1, dtype=torch.float64, device=r.device)
    k = torch.cartesian_prod(*([axis] * n)).reshape(-1, n)  # ((2K+1)^n, n)
    rho2 = (2 * math.pi) ** 2 * (k**2).sum(1)  # (I,) squared distance of image I in the compact directions
    return (r[:, None] / (r[:, None] ** 2 + rho2) ** ((n + 3) / 2)).sum(1)  # summed over images


def g4_over_gd(n: int) -> float:
    """G_4 / G_D = |S^(n+2)| / (4 pi (2 pi R)^n): Gauss's law through S^2 x T^n far from the mass (R = 1)."""
    return sphere_area(n + 2) / (4 * math.pi * (2 * math.pi) ** n)


def torus_yukawa_alpha(n: torch.Tensor) -> torch.Tensor:
    """Strength of the first Kaluza-Klein graviton level on an n-torus: 2n modes x 4/3 = 8n/3 (PDG: 16/3 for n = 2)."""
    return 8 * n / 3


# ---------- ADD: M_Pl^2 = 8 pi R^n M_D^(n+2) ----------


def add_md_tev(n: torch.Tensor, R_m: torch.Tensor) -> torch.Tensor:
    """Fundamental scale M_D (TeV) for n dimensions of radius R (metres), GRW / PDG convention."""
    R = torch.as_tensor(R_m, dtype=torch.float64) / tensors.HBARC_M  # GeV^-1
    log_m = (2 * math.log(tensors.M_PL_GEV) - math.log(8 * math.pi) - n * torch.log(R)) / (n + 2)
    return torch.exp(log_m) / 1e3


def add_radius_m(n: torch.Tensor, M_tev: torch.Tensor) -> torch.Tensor:
    """Radius R (metres) that n dimensions need for M_Pl^2 = 8 pi R^n M_D^(n+2) (broadcast over n and M_D)."""
    M = torch.as_tensor(M_tev, dtype=torch.float64) * 1e3
    log_r = (2 * math.log(tensors.M_PL_GEV) - math.log(8 * math.pi) - (n + 2) * torch.log(M)) / n
    return torch.exp(log_r) * tensors.HBARC_M


# ---------- the experimental bounds as masks on the (n, R) grid ----------


def _limit(n: torch.Tensor, row: dict) -> torch.Tensor:
    """Largest allowed R (metres) that one cited bound sets for each n (inf where it says nothing)."""
    inf = torch.full_like(n, math.inf)
    if row["kind"] == "lambda_max":  # any n >= 1: alpha = 8n/3 >= 1, lambda = R
        return torch.where(n >= 1, torch.full_like(n, row["value"]), inf)
    on = n == row["n"]
    if row["kind"] == "R_max":
        return torch.where(on, torch.full_like(n, row["value"]), inf)
    R = add_radius_m(torch.where(on, n, torch.ones_like(n)), torch.tensor(row["value"]))  # M_D > M_lim <=> R < R(M_lim)
    return torch.where(on, R, inf)


def exclusion_masks(n: torch.Tensor, R: torch.Tensor, rows: list[dict]) -> dict:
    """{bound name: (N, R) mask of excluded points}; n (N, 1) and R (1, R) broadcast."""
    masks = {}
    for row in rows:  # a handful of cited numbers; each one is applied to the whole grid at once
        m = R > _limit(n.to(torch.float64), row)
        masks[row["bound"]] = masks.get(row["bound"], torch.zeros_like(m)) | m
    return masks


def excluded(n: torch.Tensor, R: torch.Tensor, rows: list[dict]) -> torch.Tensor:
    """(N, R) excluded by any bound."""
    m = exclusion_masks(n, R, rows)
    return torch.stack(torch.broadcast_tensors(*m.values())).any(0)


def r_max(n: torch.Tensor, rows: list[dict]) -> torch.Tensor:
    """(N,) largest radius compatible with every bound (inf when nothing applies, e.g. n = 0)."""
    return torch.stack([_limit(n.to(torch.float64), row) for row in rows]).amin(0)


# ---------- the quantum side (step 1) ----------


def race_dims(d: int, K: int) -> int:
    """Projection race with the Born rule: d observed dimensions plus one hidden complex dimension per outcome."""
    return d + K


def hilbert_log2_dim(N: torch.Tensor) -> torch.Tensor:
    """log2 of the dimension N entangled two-level systems need: D >= 2^N."""
    return N.clone()


def spatial_reading_extra_dims(N: torch.Tensor) -> torch.Tensor:
    """If every Hilbert dimension were a spatial direction: 2^N complex = 2^(N+1) real coordinates, minus our 3."""
    return 2 ** (N + 1) - 3
```

## 3. `tensors.py` — PyTorch tensors

```python
%%writefile problems/cosmology/projection_dimension_budget/tensors.py
"""Concrete tensors: constants, the (n, R) grid and the cited bounds (data/gravity_bounds.csv).

Particle physics in GeV, lengths in metres; 1 GeV^-1 = hbar c / GeV = 1.97e-16 m.
"""

import csv
import math
from pathlib import Path

import torch

DATA = Path(__file__).parent / "data"
M_PL_GEV = 1.220890e19  # Planck mass sqrt(hbar c / G) (CODATA 2018)
HBARC_M = 1.973269804e-16  # hbar c in GeV m
L_PLANCK_M = 1.616255e-35  # Planck length (CODATA 2018)


def grid(n_max: int, R_points: int, device) -> tuple[torch.Tensor, torch.Tensor]:
    """n = 0..n_max extra dimensions (N,) and radii R (R,) from the Planck length to 1 m, log spaced."""
    n = torch.arange(0, n_max + 1, dtype=torch.float64, device=device)
    R = torch.logspace(math.log10(L_PLANCK_M), 0, R_points, dtype=torch.float64, device=device)
    return n, R


def bounds() -> list[dict]:
    """Cited bounds: bound, source, year, n (None = any n >= 1), kind (lambda_max | R_max | M_D_min), value."""
    with open(DATA / "gravity_bounds.csv", encoding="utf-8") as f:
        rows = list(csv.DictReader(r for r in f if not r.startswith("#")))
    for r in rows:
        r["n"] = None if r["n"] == "any" else float(r["n"])
        r["value"] = float(r["value"])
    return rows


STAGES = ["torsion", "lhc", "sn1987a", "neutron_star"]  # the order in which the map fills in
LABELS = {"torsion": "torsion balances (Kapner 2007, Tan 2016, Lee 2020)", "lhc": "LHC monojet (ATLAS 2021)",
          "sn1987a": "SN1987A cooling (Hanhart 2001)",
          "neutron_star": "neutron-star heating (Hannestad & Raffelt 2003)"}  # fmt: skip
```

## 4. `solve.py` — solver

```python
%%writefile problems/cosmology/projection_dimension_budget/solve.py
"""Solvers: the (n, R) exclusion map bound by bound, the largest allowed radius, force laws, the quantum count."""

import math

import torch

from . import equation, tensors


def exclusion_map(n: torch.Tensor, R: torch.Tensor, rows: list[dict]) -> dict:
    """Masks per bound family (N, R), the first family that excludes each point (-1 = allowed), and M_D on the grid."""
    masks = equation.exclusion_masks(n[:, None], R[None, :], rows)
    first = torch.full((len(n), len(R)), -1, dtype=torch.long, device=R.device)
    for i, name in reversed(list(enumerate(tensors.STAGES))):  # earlier stages win where several exclude
        first = torch.where(masks[name], torch.full_like(first, i), first)
    md = equation.add_md_tev(n[1:, None].clamp_min(1), R[None, :])  # (N-1, R) M_D in TeV for n >= 1
    return {"masks": masks, "first": first, "md": md, "excluded": first >= 0}


def radius_budget(n: torch.Tensor, rows: list[dict]) -> dict:
    """R_max(n) after each stage of bounds (cumulative) and the M_D it corresponds to."""
    per_stage = []
    for k in range(len(tensors.STAGES)):
        used = [r for r in rows if r["bound"] in tensors.STAGES[: k + 1]]
        per_stage.append(equation.r_max(n, used))
    rmax = per_stage[-1]
    md = torch.where(n >= 1, equation.add_md_tev(n.clamp_min(1), rmax.clamp_max(1.0)), torch.nan)
    return {"per_stage": torch.stack(per_stage), "rmax": rmax, "md_at_rmax": md}


def force_curves(device) -> dict:
    """r^2 g(r) / (G_4 M) on R^3 x T^n, n = 1, 2, 3: (R/r)^n inside the compact space, 1 outside (r <= 10 R)."""
    r = torch.logspace(-2, 1, 50, dtype=torch.float64, device=device)
    out = {}
    for n, K in ((1, 2000), (2, 150), (3, 30)):
        out[n] = r**2 * equation.image_force(r, n, K) / equation.g4_over_gd(n)
    return {"r": r, "curves": out}


def gauss_check(n_max: int, generator) -> torch.Tensor:
    """Flux of the pull through spheres of radius 0.5, 1, 2 divided by |S^(n+2)|, for n = 0..n_max (should be 1)."""
    radii = torch.tensor([0.5, 1.0, 2.0], dtype=torch.float64)
    return torch.stack([equation.gauss_flux(n, radii, 2000, generator) / equation.sphere_area(n + 2)
                        for n in range(n_max + 1)])  # fmt: skip


def quantum_budget(d: int, outcomes: int, qubits: int, device) -> dict:
    """Step 1's count: d + K for the Born-rule race, 2^N for N entangled qubits, and the spatial reading."""
    N = torch.arange(1, 81, dtype=torch.float64, device=device)
    Nq = torch.tensor(float(qubits), dtype=torch.float64)
    return {"N": N, "log10_hilbert": equation.hilbert_log2_dim(N) * math.log10(2),
            "log10_spatial": torch.log10(equation.spatial_reading_extra_dims(N)),
            "race": equation.race_dims(d, outcomes),
            "qubits": qubits, "hilbert_q": int(2 ** qubits),
            "spatial_q": int(equation.spatial_reading_extra_dims(Nq).item())}  # fmt: skip
```

## 5. `plot.py` — 2D/3D plot → GIF

```python
%%writefile problems/cosmology/projection_dimension_budget/plot.py
"""Animate the (n, R) map filling in bound by bound, R_max(n), the force law and the quantum dimension count."""

import math
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import ListedColormap
from PIL import Image

from problems._common.plotting import ACCENT, export_plot_data, frame_indices
from problems._common.progress import report

from . import tensors

COLORS = ["#dcfce7", "#93c5fd", "#fca5a5", "#fcd34d", "#c4b5fd"]  # allowed, then one colour per bound family


def animate(n, R, em, rb, fc, qb, R_star, M_star, path: Path) -> Path:
    nv, lR = n.cpu().numpy(), np.log10(R.cpu().numpy())
    first = em["first"].cpu().numpy()
    per_stage = np.log10(np.clip(rb["per_stage"].cpu().numpy(), 1e-40, 1.0))  # (stages, N), inf -> 1 m
    n_st = len(tensors.STAGES)

    fig = plt.figure(figsize=(11, 7.2))
    ax_m, ax_b = fig.add_subplot(2, 2, 1), fig.add_subplot(2, 2, 2)
    ax_f, ax_q = fig.add_subplot(2, 2, 3), fig.add_subplot(2, 2, 4)

    # --- the map ---
    cmap = ListedColormap(COLORS)
    extent = (lR[0], lR[-1], nv[0] - 0.5, nv[-1] + 0.5)
    img = ax_m.imshow(np.zeros_like(first, dtype=float), origin="lower", aspect="auto", extent=extent, cmap=cmap,
                      vmin=-0.5, vmax=n_st + 0.5, interpolation="nearest")  # fmt: skip
    for k, c in enumerate(COLORS):
        label = "allowed" if k == 0 else tensors.LABELS[tensors.STAGES[k - 1]]
        ax_m.plot([], [], "s", color=c, ms=8, label=label)
    ax_m.legend(fontsize=6.5, loc="lower left", framealpha=0.9)
    ax_m.set_xlabel("radius of the extra dimensions $\\log_{10}(R / 1\\,\\mathrm{m})$ (dimensionless)")
    ax_m.set_ylabel("extra dimensions $n$ (dimensionless)")
    ax_m.set_yticks(nv)
    for x, txt in ((math.log10(tensors.L_PLANCK_M), "Planck\nlength"), (-15, "1 fm"), (-6, "1 μm")):
        ax_m.axvline(x, color="#6b7280", lw=0.6, ls=":")
        ax_m.text(x + 0.3, nv[-1] - 0.2, txt, fontsize=6, color="#374151", va="top")
    if nv[-1] >= 7:
        ax_m.axhspan(6.5, nv[-1] + 0.5, xmin=0.25, xmax=0.85, fill=False, hatch="//", ec="#9ca3af", lw=0)
        ax_m.text(-22, 6.75, r"$n \geq 7$: only torsion bounds bundled", fontsize=7, color="#374151")
    star = ax_m.plot([], [], "k--", lw=1.2, label=f"ADD with $M_D$ = {M_star:g} TeV")[0]
    rmax_dots = ax_m.plot([], [], "o", color="#111827", ms=4)[0]
    title_m = ax_m.set_title("", fontsize=9)

    # --- R_max(n) ---
    bars = ax_b.bar(nv, np.zeros_like(nv) - 36, bottom=0, color="#86efac", edgecolor="#166534")
    ax_b.set_xlim(nv[0] - 0.6, nv[-1] + 0.6)
    ax_b.set_ylim(-36, 0.5)
    ax_b.axhline(math.log10(tensors.L_PLANCK_M), color="#6b7280", lw=0.8, ls=":")
    ax_b.set_xticks(nv)
    ax_b.set_xlabel("extra dimensions $n$ (dimensionless)")
    ax_b.set_ylabel("largest allowed $\\log_{10}(R_{max} / 1\\,\\mathrm{m})$ (dimensionless)")
    ax_b.set_title("dimension budget: every $n$ survives, but only if small", fontsize=9)
    bar_txt = [ax_b.text(k, 0, "", fontsize=6, ha="center", va="bottom") for k in nv]

    # --- force law ---
    r = fc["r"].cpu().numpy()
    for j, (k, c) in enumerate(fc["curves"].items()):
        ax_f.loglog(r, c.cpu().numpy(), color=ACCENT[j], lw=1.8, label=f"$n = {k}$: $1/r^{{{k + 2}}}$ → $1/r^2$")
    ax_f.axhline(1, color="#111827", lw=0.8, ls="--")
    ax_f.axvline(1, color="#9ca3af", lw=0.8, ls=":")
    ax_f.text(1.1, 3e3, "$r = R$", fontsize=8, color="#374151")
    ax_f.set_xlabel("distance $r / R$ (dimensionless)")
    ax_f.set_ylabel("$r^2 g(r) / G_4 M$, Newton = 1 (dimensionless)")
    ax_f.legend(fontsize=7, loc="upper right")
    ax_f.set_title("gravity spreads into $n$ dimensions below $R$ (image sums);\n"
                   "torsion balances see Newton down to 52 μm, so $R$ < 38.6 μm", fontsize=9)  # fmt: skip

    # --- quantum side ---
    N = qb["N"].cpu().numpy()
    ax_q.plot(N, qb["log10_hilbert"].cpu().numpy(), color=ACCENT[3], lw=2, label="Hilbert space $D = 2^N$ (step 1)")
    ax_q.plot(N, qb["log10_spatial"].cpu().numpy(), color=ACCENT[4], lw=1.4, ls="--",
              label="if spatial: $2^{N+1} - 3$ extra real dims")  # fmt: skip
    ax_q.axhspan(0, math.log10(6), color="#fca5a5", alpha=0.5, lw=0, label="$n$ = 1..6 with collider bounds")
    ax_q.plot([qb["qubits"]], [math.log10(qb["hilbert_q"])], "o", color="#111827", ms=6)
    ax_q.set_xlabel("entangled two-level systems $N$ (dimensionless)")
    ax_q.set_ylabel("$\\log_{10}$ dimensions needed (dimensionless)")
    ax_q.legend(fontsize=7, loc="upper left")
    ax_q.set_title(f"quantum side: {qb['qubits']} qubits need $D \\geq$ {qb['hilbert_q']}; race $d + K$ = {qb['race']}"
                   "\nthese need not be spatial (Hilbert-space reading)", fontsize=9)  # fmt: skip

    pre, per, hold = 4, 14, 14
    frames = list(range(pre + n_st * per + hold))
    frames = [frames[i] for i in frame_indices(len(frames), 72)]
    R_star_np = np.log10(R_star.cpu().numpy())
    rmax = np.log10(np.clip(rb["rmax"].cpu().numpy(), 1e-40, 1.0))

    def update(f: int):
        stage = (f - pre) // per if f >= pre else -1
        stage = min(stage, n_st)
        shown = np.zeros_like(first, dtype=float)
        for k in range(min(stage, n_st)):  # finished stages
            shown[first == k] = k + 1
        if 0 <= stage < n_st:  # the current stage sweeps in from large R
            frac = ((f - pre) % per + 1) / per
            cols = np.nonzero((first == stage).any(0))[0]
            lo = lR[cols[0]] if len(cols) else lR[-1]
            cut = lR[-1] - frac * (lR[-1] - lo)
            shown[(first == stage) & (lR[None, :] >= cut)] = stage + 1
            title_m.set_text(f"adding: {tensors.LABELS[tensors.STAGES[stage]]}")
        img.set_data(shown)
        done = min(stage, n_st - 1)
        if stage >= n_st:
            star.set_data(np.r_[R_star_np], nv[1:])
            rmax_dots.set_data(rmax[1:], nv[1:])
            title_m.set_text("all bounds: excluded unless $R < R_{max}(n)$ (dots); dashed: $M_D$ = "
                             f"{M_star:g} TeV")  # fmt: skip
        top = per_stage[done] if stage >= 0 else np.zeros_like(nv)
        for k, (b, t) in enumerate(zip(bars, bar_txt)):
            h = top[k] if nv[k] > 0 else 0.0
            b.set_height(h + 36)
            b.set_y(-36)
            b.set_color("#e5e7eb" if nv[k] >= 7 else "#86efac")
            b.set_edgecolor("#166534")
            t.set_position((nv[k], h + 0.3))
            t.set_text("any" if nv[k] == 0 else f"{10**h:.0e}" if h < -0.01 else "")

    fig.subplots_adjust(left=0.07, right=0.98, bottom=0.08, top=0.9, wspace=0.28, hspace=0.48)
    return save_gif_shared_palette(fig, update, frames, path, fps=8)


def save_gif_shared_palette(fig, update, frames: list[int], path: Path, fps: int = 8) -> Path:
    """Like plotting.save_gif, but every frame uses ONE palette built from a strip of all frames, so colours stay
    fixed through the animation (Pillow otherwise picks a new 256-colour palette per frame)."""
    fig.set_dpi(80)
    shots = []
    for i, f in enumerate(frames):
        update(f)
        fig.canvas.draw()
        shots.append(Image.fromarray(np.asarray(fig.canvas.buffer_rgba())[..., :3]))
        report(i + 1, len(frames), "frames")
    export_plot_data(fig, path)  # numbers behind the final frame (Results page)
    plt.close(fig)
    sample = shots[:: max(1, len(shots) // 12)]
    strip = Image.new("RGB", (sample[0].width, sample[0].height * len(sample)))
    for k, im in enumerate(sample):
        strip.paste(im, (0, k * im.height))
    palette = strip.quantize(colors=255, method=Image.Quantize.MEDIANCUT)
    frames_p = [im.quantize(palette=palette, dither=Image.Dither.NONE) for im in shots]
    tmp = path.with_name(f".{path.stem}.tmp{path.suffix}")
    frames_p[0].save(tmp, save_all=True, append_images=frames_p[1:], duration=int(1000 / fps), loop=0)
    return tmp.replace(path)
```
