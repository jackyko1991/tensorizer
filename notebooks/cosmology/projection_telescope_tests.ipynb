# Telescope tests of the projection: gravitational waves, black-hole shadows and supernovae

*cosmology · advanced* · [open in Tensorizer](https://jackyko1991.github.io/tensorizer/#/cosmology/projection_telescope_tests)

A speculative hypothesis, not established physics, confronted with open astronomical data. If gravity sees more than $3+1$ dimensions, gravitational waves weaken faster than $1/d$; if curved spacetime is the shadow of a folded projection, a fold outside the photon sphere would enlarge a black hole's shadow. GW170817 with its host galaxy's distance gives $D = 4.00^{+0.07}_{-0.25}$, the EHT shadows of M87* and Sgr A* allow a fold only within about $2.3$ Schwarzschild radii, and Pantheon+ supernovae cannot tell a leaking (DGP) universe from $\Lambda$CDM. Everything agrees with general relativity in $3+1$: any extra dimensions must be ones gravity does not feel at these scales.

**Vector form**

$$h \propto d^{-(D-2)/2},\qquad d_{GW} = d_*\Big(\frac{d_{EM}}{d_*}\Big)^{\frac{D-2}{2}},\qquad d_{GW} = d_{EM}\Big[1 + \Big(\frac{d_{EM}}{R_c}\Big)^{\frac{n(D-4)}{2}}\Big]^{1/n},\qquad p(D) \propto \int \mathcal L\big(d_{GW}(d, D)\big)\,\mathcal N(d;\, d_{EM}, \sigma)\,dd,\qquad b_c = \min_r \frac{r}{\sqrt{f(r)}} = \sqrt{27}\,\frac{GM}{c^2},\qquad b_{edge} = \max\Big(b_c,\ \frac{r_f}{\sqrt{1 - r_s/r_f}}\Big),\qquad E^2 = \Omega_m(1+z)^3 + (1-\Omega_m)E^{\alpha}$$

**Index (tensor) form**

$$\mathcal L(x) = \frac{\sum_i w_i\,K_h(x - d_i)}{\sum_i w_i},\quad w_i = d_i^{-2},\qquad p_a = \frac{\sum_j \mathcal L\big(d_{GW}(d_j, D_a)\big)\,g_j\,\Delta d}{\sum_{a} (\cdots)\,\Delta D},\qquad \frac{d^2u_{fb}}{d\varphi^2} = -u_{fb} + \tfrac32\,u_{fb}^2,\qquad \delta = \frac{\theta_{sh}}{2\sqrt{27}\,\theta_g} - 1,\qquad \chi^2 = \sum_i w_i r_i^2 - \frac{(\sum_i w_i r_i)^2}{\sum_i w_i}$$

Known physics first. In general relativity in $3+1$ dimensions a gravitational wave's strain falls as $1/d$; in $D$ spacetime dimensions its energy spreads over a $(D-2)$-sphere and $h\propto d^{-(D-2)/2}$. A detector that assumes $1/d$ infers $d_{GW} = d_*(d_{EM}/d_*)^{(D-2)/2}$ ($d_* = 1$ Mpc, as in Pardo, Fishbach, Holz and Spergel 2018) or, if the leakage only starts beyond a screening scale $R_c$, $d_{GW} = d_{EM}[1 + (d_{EM}/R_c)^{n(D-4)/2}]^{1/n}$ (their screened form, meaningful for $D > 4$). The hypothesis' prediction: if gravity spreads into extra dimensions at large distances, $d_{GW} > d_{EM}$ and $D > 4$. The test (a): GW170817, a neutron-star merger with an optical counterpart in NGC 4993. Data: 8078 posterior samples of the GW luminosity distance from the GWTC-1 release (data/gw170817_distance.csv; LIGO Scientific and Virgo Collaborations, low-spin prior, sky position free; GWOSC, licence CC BY 4.0; Abbott et al. 2019, PRX 9, 031040). The analysis prior is $\propto d^2$, so each sample is weighted by $d^{-2}$ and a kernel density gives the likelihood of $d_{GW}$ (the published value is $40^{+8}_{-14}$ Mpc, Abbott et al. 2017). EM distances (data/em_distances.csv, cited numbers): surface brightness fluctuations $40.7\pm1.4\pm1.9$ Mpc (Cantiello et al. 2018, ApJL 854, L31, the default) and the Hubble-flow distances Pardo et al. used, $v_H = 3017\pm250$ km/s divided by $H_0 = 73.24\pm1.74$ (SH0ES) or $67.74\pm0.46$ km/s/Mpc (Planck). Result: $D = 4.00^{+0.07}_{-0.25}$ (68%, SBF; the autograd fit gives $4.00\pm0.11$), $3.98^{+0.09}_{-0.24}$ (SH0ES) and $3.94^{+0.08}_{-0.24}$ (Planck). Pardo et al. found $4.02^{+0.07}_{-0.10}$ (SH0ES) and $3.98^{+0.07}_{-0.09}$ (Planck); their lower error is smaller because they fixed the sky position to the counterpart and used the LVC $H_0$ posterior, while the samples bundled here have a free sky position, so the distance-inclination degeneracy leaves a longer tail to small distances. For screened leakage, $-2\ln[\mathcal L(R_c)/\mathcal L_{GR}] > 2.71$ excludes $R_c$ below a few hundred Mpc for $D = 5$, $n = 1$ and below about 50-80 Mpc for $n = 2$ to 5; gradual transitions ($n(D-4)/2$ small) are excluded up to the $10^4$ Mpc edge of the grid. Pardo et al. quote $R_c\gtrsim 20$ Mpc for any steepness, from a posterior rather than a likelihood ratio. Verdict (a): gravity behaves as $D = 4.0$ to about $\pm0.1$ out to 40 Mpc; any extra dimensions gravity sees are compact or screened beyond these distances. The test (b): black-hole shadows. In Schwarzschild the photon sphere is at $r = 3GM/c^2$ and the shadow radius is $b_c = \sqrt{27}\,GM/c^2$; here it is computed both as $\min_r r/\sqrt{f}$ and by ray tracing $u'' = -u + \tfrac32 u^2$ for a batch of impact parameters. Spin shrinks the shadow's area-equivalent radius by at most 7% (Bardeen's curve; $2(1+\sqrt2)\,GM/c^2$ for an extremal hole seen from the pole). The fold, a modelling choice of the author made precise here: P4 says a singularity is where the projection folds; suppose the projected geometry is Schwarzschild down to a fold at radius $r_f$ and light that reaches the fold never comes back. A fold inside the photon sphere ($r_f\leq 1.5\,r_s$) leaves the shadow unchanged; one outside makes the shadow edge the ray that grazes it, $b = r_f/\sqrt{1 - r_s/r_f}$, so $\delta = b/b_c - 1 > 0$. Data (data/eht_shadows.csv, cited numbers): M87* ring $42\pm3$ $\mu$as and $GM/(Dc^2) = 3.8\pm0.4$ $\mu$as (EHT 2019, Papers I and VI), against $3.64\pm0.22$ $\mu$as from stellar dynamics (Gebhardt et al. 2011, $6.6\pm0.4\times10^9\,M_\odot$ at 17.9 Mpc), so $\delta = +0.04\pm0.13$ (computed here, not a published value); Sgr A* ring $51.8\pm2.3$ $\mu$as (EHT 2022, Paper I) and $\delta = -0.04^{+0.09}_{-0.10}$ (Keck prior) and $-0.08\pm0.09$ (VLTI prior) from Paper VI, Table 2. Taking $\delta + 2\sigma$ and letting spin hide up to 7%, a fold may grow the shadow by at most 0.37 (M87*), 0.21 (Keck) and 0.17 (VLTI), so $r_f\leq 2.86$, $2.40$ and $2.28\,r_s$. Verdict (b): the shadows match Schwarzschild/Kerr; a fold outside about $2.3\,r_s$ is excluded, but a fold inside the photon sphere, where P4 would naturally put it (at the horizon or the central singularity), is invisible to shadow images. The test (c, optional): supernovae. In a DGP brane world gravity leaks into a fifth dimension beyond $r_c$ and the expansion obeys $H^2 - H/r_c\propto\rho$; Dvali and Turner's index $\alpha$ interpolates ($E^2 = \Omega_m(1+z)^3 + (1-\Omega_m)E^\alpha$, $\alpha = 0$ is $\Lambda$CDM, $\alpha = 1$ DGP).
```python
%%writefile problems/cosmology/projection_telescope_tests/main.py
"""Main loop: data -> fits -> verdicts -> plot. Step 6 of the Projection Hypothesis (speculative): telescope tests."""

import math
from pathlib import Path

import torch

from problems._common.runtime import cli

from . import plot, solve, tensors

F64 = torch.float64
PARDO = {"shoes": (4.02, 0.10, 0.07), "planck": (3.98, 0.09, 0.07)}  # MAP, -, + (68%), Pardo et al. 2018 abstract


def run(params: dict, out_dir: Path) -> Path:
    key = params["em_distance"]
    D_s, n_s, inc_max = float(params["D_screen"]), float(params["n_screen"]), float(params["inc_max_sgra"])

    # (a) GW170817: d_GW against d_EM, posterior of D
    gw = tensors.load_gw_samples()
    table = solve.likelihood_table(gw)
    print(f"(a) {gw['label']}: {len(gw['d'])} distance samples, d^2 prior divided out, kernel width "
          f"{float(gw['h']):.2f} Mpc")  # fmt: skip
    Ds = torch.linspace(3.0, 5.0, 801, dtype=F64)
    posts, fits = {}, {}
    for k in ("sbf", "shoes", "planck"):
        em = tensors.em_distance(k)
        posts[k] = solve.posterior_D(Ds, gw, em, table)
        fits[k] = solve.fit_D(gw, em, D0=4.5)
        lo, hi = posts[k]["interval"]
        mark = "  <- selected" if k == key else ""
        print(f"    d_EM = {em['d']:5.2f} +- {em['sigma']:.2f} Mpc ({em['label']}){mark}")
        print(f"      grid: D = {posts[k]['map']:.3f} (+{hi - posts[k]['map']:.3f} / -{posts[k]['map'] - lo:.3f}, "
              f"68% HDI);  autograd MAP {fits[k]['D']:.3f} +- {fits[k]['sigma']:.3f} (Laplace)")  # fmt: skip
    for k, (m, lo, hi) in PARDO.items():
        print(f"    Pardo et al. 2018 ({k} H0): D = {m:.2f} (+{hi:.2f} / -{lo:.2f})")
    sel = posts[key]
    lo, hi = sel["interval"]
    print(f"    verdict: gravity spreads like D = {sel['map']:.2f} (68% {lo:.2f} .. {hi:.2f}) out to 40 Mpc; "
          f"D = 4 is inside")  # fmt: skip

    em = tensors.em_distance(key)
    n_vals = torch.tensor(sorted({1.0, 2.0, 5.0, n_s}), dtype=F64)
    D_vals = torch.linspace(4.25, 8.0, 31, dtype=F64)
    scr = solve.screening_bound(torch.cat([D_vals, torch.tensor([D_s], dtype=F64)]), n_vals, gw, em, table)
    i_n = int((n_vals - n_s).abs().argmin())
    b = float(scr["bound"][-1, i_n])
    edge = b >= float(scr["Rc"][-1]) * 0.999
    print(f"    screened leakage (D = {D_s:g}, n = {n_s:g}): R_c {'>' if edge else '>='} {b:.0f} Mpc "
          f"(-2 ln L/L_GR > 2.71 below it)")  # fmt: skip

    # (b) EHT shadows and the fold
    sh = solve.shadow_tests(tensors.load_eht(), spin_inc_max=inc_max)
    edges = solve.traced_shadow_edge(torch.tensor([0.0, 2.0], dtype=F64))
    print(f"(b) Schwarzschild shadow radius: ray tracing {float(edges[0]):.4f} r_s, photon sphere "
          f"sqrt(27)/2 = {math.sqrt(27) / 2:.4f} r_s")  # fmt: skip
    for r in sh["rows"]:
        print(f"    {r['object']:6s} {r['prior']:30s} 2 sqrt27 theta_g = {r['shadow_schw']:5.1f} uas (ring "
              f"{r['ring_uas']:.1f} +- {r['ring_err']:.1f}); delta = {r['delta']:+.3f} (-{r['delta_lo']:.3f} "
              f"+{r['delta_hi']:.3f})")  # fmt: skip
        print(f"           delta_up (2 sigma) = {r['delta_up']:+.3f}, spin can shrink by {-r['kerr_min']:.3f} -> "
              f"fold may grow it by {r['allowed']:.3f}: r_f <= {r['r_bound']:.2f} r_s")  # fmt: skip
    print("    verdict: shadows match Schwarzschild/Kerr; a fold outside about 2.3-2.9 r_s is excluded, a fold "
          "inside the photon sphere (1.5 r_s) is invisible to the shadow")  # fmt: skip

    # (c) Pantheon+ supernovae: leakage index alpha
    sn = solve.supernova_fit(*tensors.load_supernovae())
    print(f"(c) Pantheon+ ({sn['n']} SNe, diagonal errors, offset marginalised): alpha = {sn['mean']:.2f} +- "
          f"{sn['sd']:.2f}; best fit alpha = {sn['a_fit']:.2f}, Om = {sn['om_fit']:.2f}")  # fmt: skip
    print(f"    chi2: LCDM (alpha 0) {sn['lcdm'][0]:.1f} at Om {sn['lcdm'][1]:.2f}; DGP (alpha 1) {sn['dgp'][0]:.1f} "
          f"at Om {sn['dgp'][1]:.2f}: supernovae alone cannot tell them apart")  # fmt: skip

    summary = plot.summary_rows(posts, sh, sn, PARDO)
    print("Summary (deviation from GR in 3+1, in standard deviations):")
    for row in summary:
        print(f"    {row['label']:34s} {row['pull']:+.2f}   {row['text']}")
    return plot.animate(gw, table, posts, key, scr, i_n, sh, sn, summary, out_dir / "projection_telescope_tests.gif")


if __name__ == "__main__":
    cli(run, __file__)
```

## 2. `equation.py` — physics → tensor translation

```python
%%writefile problems/cosmology/projection_telescope_tests/equation.py
r"""Physics -> tensor translation: three telescope tests of extra dimensions and of a fold in the projection.

(a) Gravitational waves. In D spacetime dimensions the wave's energy spreads over a (D-2)-sphere, so the strain
falls as h ~ d^{-(D-2)/2} instead of 1/d. A GW detector, assuming 1/d, then infers the distance

    d_GW = d_* (d_EM / d_*)^{(D-2)/2}                            (no screening, d_* = 1 Mpc, Pardo et al. 2018)
    d_GW = d_EM [1 + (d_EM / R_c)^{n (D-4)/2}]^{1/n}             (screened: leakage only beyond R_c, D > 4)

The GW likelihood of d_GW is a kernel density of the posterior samples with the d^2 volume prior divided out:

    L(x) = \sum_i w_i K_h(x - d_i) / \sum_i w_i,     w_i = d_i^{-2}       (i summed over samples)

and the posterior of D marginalises the electromagnetic distance:  p(D) \propto \int L(d_GW(d, D)) N(d; d_EM, σ) dd.

(b) Black-hole shadows. For ds^2 = -f dt^2 + dr^2/f + r^2 dΩ^2 a photon of impact parameter b obeys
(du/dφ)^2 = 1/b^2 - u^2 f(1/u), u = 1/r; with f = 1 - r_s/r, u'' = -u + (3/2) r_s u^2. The shadow edge is
b_c = min_r r/\sqrt{f(r)}: r_ph = 3GM/c^2, b_c = \sqrt{27} GM/c^2. The fold (P4, the author's modelling choice
here): the projected geometry is Schwarzschild down to a fold at r_f, and light that reaches the fold never returns.
Then b_edge = max(b_c, r_f/\sqrt{1 - r_s/r_f}) and δ = b_edge/b_c - 1. A Kerr hole's shadow (Bardeen 1973) is
slightly smaller; its area-equivalent radius gives the spin band.

(c) Supernovae. Dvali & Turner (2003): H^2 - H^α / r_c^{2-α} = 8πGρ/3, i.e. for a flat universe
E^2 = Ω_m (1+z)^3 + (1 - Ω_m) E^α. α = 0 is ΛCDM, α = 1 is the self-accelerating DGP brane world (gravity leaking
into a fifth dimension beyond r_c).

Distances in Mpc, radii in units of r_s (or GM/c^2 for Kerr, where stated).
"""

import math

import torch

C_KMS = 299792.458
B_SCHW = math.sqrt(27) / 2  # Schwarzschild shadow radius in units of r_s = 2GM/c^2


# ---------------------------------------------------------------- (a) gravitational-wave leakage
def gw_distance(d_em, D, R_c=None, n: float = 1.0, d_scale: float = 1.0):
    """d_GW the detector infers for a source at d_EM if gravity spreads into D dimensions.  broadcasts, Mpc."""
    if R_c is None:
        return d_scale * (d_em / d_scale) ** ((D - 2) / 2)
    return d_em * (1 + (d_em / R_c) ** (n * (D - 4) / 2)) ** (1 / n)


def strain_amplitude(d, D, R_c=None, n: float = 1.0):
    """h(d) in units of h at 1 Mpc for 4D gravity: the detector reads 1/d_GW."""
    return 1 / gw_distance(d, D, R_c, n)


def scott_bandwidth(d: torch.Tensor, w: torch.Tensor) -> torch.Tensor:
    """Weighted Scott's rule h = σ_w n_eff^{-1/5}, n_eff = (Σw)^2 / Σw^2."""
    p = w / w.sum()
    mean = (p * d).sum()
    sd = ((p * (d - mean) ** 2).sum()).sqrt()
    n_eff = 1 / (p**2).sum()
    return sd * n_eff ** (-0.2)


def gw_likelihood(x: torch.Tensor, d: torch.Tensor, w: torch.Tensor, h, chunk: int = 2048) -> torch.Tensor:
    r"""L(x) = Σ_i w_i N(x; d_i, h) / Σ_i w_i: einsum over samples i.  x: (...) -> (...); integrates to 1 over x.

    Chunked over x only to bound memory (x points × samples)."""
    flat = x.reshape(-1)
    p = w / w.sum()
    out = []
    for s in range(0, flat.numel(), chunk):
        z = (flat[s : s + chunk, None] - d[None, :]) / h  # (X, S)
        out.append(torch.einsum("xs,s->x", torch.exp(-0.5 * z**2), p) / (h * math.sqrt(2 * math.pi)))
    return torch.cat(out).reshape(x.shape)


def gw_log_likelihood(x: torch.Tensor, d: torch.Tensor, w: torch.Tensor, h) -> torch.Tensor:
    """log L(x) by logsumexp over samples: finite even far out in the tails (for gradient-based fits).  (...)"""
    z = (x[..., None] - d) / h  # (..., S)
    logp = torch.log(w / w.sum())
    return torch.logsumexp(-0.5 * z**2 + logp, -1) - torch.log(h * math.sqrt(2 * math.pi))


def interp(x: torch.Tensor, grid: torch.Tensor, values: torch.Tensor) -> torch.Tensor:
    """Linear interpolation on a uniform grid, zero outside it.  x: (...) -> (...)"""
    step = grid[1] - grid[0]
    t = (x - grid[0]) / step
    k = t.floor().long().clamp(0, len(grid) - 2)
    s = t - k
    v = values[k] * (1 - s) + values[k + 1] * s
    return torch.where((t >= 0) & (t <= len(grid) - 1), v, torch.zeros_like(v))


def gaussian(x, mu, sigma):
    return torch.exp(-0.5 * ((x - mu) / sigma) ** 2) / (sigma * math.sqrt(2 * math.pi))


# ---------------------------------------------------------------- (b) shadows and the fold
def schwarzschild_f(r: torch.Tensor, rs: torch.Tensor) -> torch.Tensor:
    return 1 - rs / r


def critical_impact_parameter(rs: torch.Tensor, f=schwarzschild_f, n: int = 20001):
    """b_c = min_r r / sqrt(f(r)) over r in (r_s, 4 r_s] and the photon-sphere radius (grid + parabola).  (B,)"""
    r = rs[:, None] * torch.linspace(1.001, 4.0, n, dtype=rs.dtype, device=rs.device)  # (B, n)
    b = r / f(r, rs[:, None]).sqrt()
    k = b.argmin(1).clamp(1, n - 2)
    idx = torch.arange(len(rs), device=rs.device)
    y0, y1, y2 = b[idx, k - 1], b[idx, k], b[idx, k + 1]
    step = r[:, 1] - r[:, 0]
    shift = 0.5 * (y0 - y2) / (y0 - 2 * y1 + y2)  # vertex of the parabola through 3 points
    return y1 - 0.25 * (y0 - y2) * shift, r[idx, k] + shift * step


def photon_rhs(y: torch.Tensor) -> torch.Tensor:
    """d/dφ (u, u') = (u', -u + 3/2 u^2) for Schwarzschild with r_s = 1.  y: (..., 2)"""
    u, v = y[..., 0], y[..., 1]
    return torch.stack([v, -u + 1.5 * u**2], -1)


def fold_shadow_radius(r_f: torch.Tensor) -> torch.Tensor:
    """Shadow radius (units r_s) with a fold at r_f (units r_s; r_f <= 1 means no fold outside the horizon)."""
    rf = r_f.clamp_min(1.0 + 1e-12)
    graze = rf / (1 - 1 / rf).sqrt()  # impact parameter of the ray that just touches r_f
    return torch.where(rf > 1.5, graze, torch.full_like(rf, B_SCHW))


def fold_delta(r_f: torch.Tensor) -> torch.Tensor:
    """Fractional shadow change δ = b_edge / b_c - 1 caused by the fold."""
    return fold_shadow_radius(r_f) / B_SCHW - 1


def fold_bound(allowed: torch.Tensor) -> torch.Tensor:
    """Largest r_f/r_s with δ(r_f) <= allowed: the largest root of r^3 - B^2 r + B^2 = 0, B = b_c (1 + allowed).

    The cubic is r^2 / (1 - 1/r) = B^2 rearranged; its largest root by the trigonometric formula. allowed <= 0
    leaves only r_f <= 1.5 (inside the photon sphere, invisible to the shadow)."""
    B2 = (B_SCHW * (1 + allowed.clamp_min(0))) ** 2
    p, q = -B2, B2
    arg = (3 * q / (2 * p) * (-3 / p).sqrt()).clamp(-1, 1)
    return 2 * (-p / 3).sqrt() * torch.cos(torch.arccos(arg) / 3)


def kerr_delta(a: torch.Tensor, inc_deg: torch.Tensor, n: int = 4000) -> torch.Tensor:
    r"""Fractional change of the Kerr shadow's area-equivalent radius from \sqrt{27} M.  a: (A,), inc: (I,) -> (A, I)

    Bardeen's curve (M = 1): photon orbits r in [r_1, r_2] have ξ = -(r^3 - 3r^2 + a^2 r + a^2)/(a(r - 1)) and
    η = r^3 (4a^2 - r(r - 3)^2)/(a^2 (r - 1)^2); on the sky α = -ξ/\sin θ, β^2 = η + a^2\cos^2θ - ξ^2\cot^2θ.
    Area = 2 \int β dα (trapezoid along r, β = 0 where the orbit is not visible)."""
    a = a[:, None, None]
    th = torch.deg2rad(inc_deg).clamp(1e-3, math.pi / 2)[None, :, None]
    r1 = 2 * (1 + torch.cos(2 / 3 * torch.arccos(-a)))
    r2 = 2 * (1 + torch.cos(2 / 3 * torch.arccos(a)))
    s = 0.5 - 0.5 * torch.cos(torch.linspace(0, math.pi, n, dtype=a.dtype, device=a.device))  # dense at the ends
    r = r1 + (r2 - r1) * s  # (A, 1, n)
    xi = -(r**3 - 3 * r**2 + a**2 * r + a**2) / (a * (r - 1))
    eta = r**3 * (4 * a**2 - r * (r - 3) ** 2) / (a**2 * (r - 1) ** 2)
    alpha = -xi / torch.sin(th)
    beta = (eta + a**2 * torch.cos(th) ** 2 - xi**2 / torch.tan(th) ** 2).clamp_min(0).sqrt()
    area = 2 * torch.trapezoid(beta, alpha, dim=-1).abs()
    return (area / math.pi).sqrt() / math.sqrt(27) - 1


# ---------------------------------------------------------------- (c) supernovae and leakage
def leaky_expansion(z: torch.Tensor, Om, alpha, n_iter: int = 40) -> torch.Tensor:
    """E(z) solving E^2 = Ω_m (1+z)^3 + (1 - Ω_m) E^α by Newton's method.  z: (Z,), Om, alpha: (...) -> (..., Z)"""
    Om, alpha = torch.as_tensor(Om, dtype=z.dtype)[..., None], torch.as_tensor(alpha, dtype=z.dtype)[..., None]
    m = Om * (1 + z) ** 3
    E = (m + 1 - Om).sqrt()  # the ΛCDM value as the start
    for _ in range(n_iter):
        g = E**2 - m - (1 - Om) * E**alpha
        dg = 2 * E - alpha * (1 - Om) * E ** (alpha - 1)
        E = E - g / dg
    return E


def distance_modulus(z: torch.Tensor, nodes: torch.Tensor, Om, alpha) -> torch.Tensor:
    """μ(z) - 5 log10(c/H0 / Mpc) - 25: the shape of the Hubble diagram (H0 and M are one offset).  -> (..., N)"""
    f = 1 / leaky_expansion(nodes, Om, alpha)  # (..., Z) dχ/dz
    h = nodes[1] - nodes[0]
    chi = torch.cat([torch.zeros_like(f[..., :1]), (0.5 * h * (f[..., 1:] + f[..., :-1])).cumsum(-1)], -1)
    t = z / h
    k = t.floor().long().clamp(0, len(nodes) - 2)
    s = t - k
    chi_z = chi[..., k] * (1 - s) + chi[..., k + 1] * s
    return 5 * torch.log10((1 + z) * chi_z)


def chi2_offset_marginal(mu_shape: torch.Tensor, mu: torch.Tensor, sigma: torch.Tensor) -> torch.Tensor:
    r"""min over a constant offset M of Σ_i ((μ_i - shape_i - M)/σ_i)^2 = Σ w r^2 - (Σ w r)^2 / Σ w (i summed)."""
    w = sigma**-2
    r = mu - mu_shape
    return torch.einsum("...i,i->...", r**2, w) - torch.einsum("...i,i->...", r, w) ** 2 / w.sum()
```

## 3. `tensors.py` — PyTorch tensors

```python
%%writefile problems/cosmology/projection_telescope_tests/tensors.py
"""Concrete tensors: the GW170817 distance samples, the electromagnetic distances, the EHT numbers, Pantheon+.

data/gw170817_distance.csv: 8078 posterior samples of the luminosity distance (GWTC-1, low-spin prior, GWOSC,
CC BY 4.0). The analysis used a d^2 (uniform in volume) prior, so each sample gets the weight 1/d^2 to turn the
posterior into the likelihood of d_GW. data/em_distances.csv and data/eht_shadows.csv hold cited published numbers;
data/pantheon_plus.csv is a copy of the supernova table of cosmology/cosmological_constant_fit.
"""

import csv
from pathlib import Path

import torch

from . import equation

DATA = Path(__file__).parent / "data"
F64 = torch.float64


def _rows(name: str) -> list[dict]:
    with open(DATA / name, encoding="utf-8") as f:
        return list(csv.DictReader(r for r in f if not r.startswith("#")))


def load_gw_samples(device=None) -> dict:
    """d (S,) posterior samples in Mpc, w (S,) = 1/d^2 (prior divided out), h the kernel width (Scott's rule)."""
    d = torch.tensor([float(r["d_L_Mpc"]) for r in _rows("gw170817_distance.csv")], dtype=F64, device=device)
    w = d**-2
    return {"d": d, "w": w, "h": equation.scott_bandwidth(d, w), "label": "GW170817 (GWTC-1)"}


def em_distance(key: str = "sbf") -> dict:
    """The electromagnetic distance to NGC 4993: sbf (Cantiello 2018), shoes or planck (Hubble flow, Pardo 2018)."""
    for r in _rows("em_distances.csv"):
        if r["key"] == key:
            return {"d": float(r["d_Mpc"]), "sigma": float(r["sigma_Mpc"]), "label": r["method"], "key": key}
    raise ValueError(f"unknown EM distance {key!r}")


def load_eht() -> list[dict]:
    """EHT rows (M87*, Sgr A* with Keck and VLTI priors) with numbers as floats (None where empty)."""
    out = []
    for r in _rows("eht_shadows.csv"):
        out.append({k: (float(v) if v and k not in ("key", "object", "prior") else v or None) for k, v in r.items()})
    return out


def load_supernovae(device=None):
    """z, μ, σ_μ (N,) from the bundled Pantheon+ table, sorted by redshift."""
    rows = _rows("pantheon_plus.csv")
    t = torch.tensor([[float(r["z"]), float(r["mu"]), float(r["sigma_mu"])] for r in rows], dtype=F64, device=device)
    t = t[t[:, 0].argsort()]
    return t[:, 0], t[:, 1], t[:, 2]


def synthetic_gw(d_em: float, D: torch.Tensor, frac_err: float, n: int, generator: torch.Generator) -> dict:
    """Fake GW 'posterior' (flat prior, weights 1) centred on d_GW(d_em, D) with a fractional error frac_err."""
    d_gw = float(equation.gw_distance(torch.tensor(d_em, dtype=F64), D))
    d = d_gw * (1 + frac_err * torch.randn(n, generator=generator, dtype=F64))
    w = torch.ones_like(d)
    return {"d": d, "w": w, "h": equation.scott_bandwidth(d, w), "label": "synthetic"}
```

## 4. `solve.py` — solver

```python
%%writefile problems/cosmology/projection_telescope_tests/solve.py
"""Solvers: the posterior of D and its autograd fit, the screening scale, ray-traced shadows, the fold bounds and the
supernova leakage fit."""

import math

import torch

from . import equation

F64 = torch.float64
D_TABLE = torch.linspace(0.05, 200.0, 4000, dtype=F64)  # d_GW grid (Mpc) for the tabulated GW likelihood


def likelihood_table(gw: dict) -> torch.Tensor:
    """L(d_GW) on D_TABLE: the kernel density is evaluated once, then interpolated."""
    return equation.gw_likelihood(D_TABLE.to(gw["d"].device), gw["d"], gw["w"], gw["h"])


def em_grid(em: dict, n: int = 401):
    """Grid over the electromagnetic distance (±6σ) and its Gaussian weights for the marginalisation."""
    d = torch.linspace(em["d"] - 6 * em["sigma"], em["d"] + 6 * em["sigma"], n, dtype=F64).clamp_min(0.5)
    return d, equation.gaussian(d, em["d"], em["sigma"])


def posterior_D(Ds: torch.Tensor, gw: dict, em: dict, table=None) -> dict:
    """p(D) on the grid Ds (uniform prior), its MAP and minimal 68% interval (highest density)."""
    table = likelihood_table(gw) if table is None else table
    d, g = em_grid(em)
    L = equation.interp(equation.gw_distance(d[None, :], Ds[:, None]), D_TABLE, table)  # (nD, nd)
    like = torch.trapezoid(L * g, d, dim=1)  # ∫ L(d_GW(d, D)) N(d) dd
    pdf = like / torch.trapezoid(like, Ds)
    order = pdf.argsort(descending=True)  # highest-density interval: add grid points from the top down
    mass = (pdf[order] * (Ds[1] - Ds[0])).cumsum(0)
    inside = order[: int((mass < 0.6827).sum()) + 1]
    return {"D": Ds, "pdf": pdf, "map": float(Ds[pdf.argmax()]),
            "interval": (float(Ds[inside].min()), float(Ds[inside].max())), "like": like}  # fmt: skip


def log_posterior(D: torch.Tensor, gw: dict, em: dict) -> torch.Tensor:
    """Smooth log p(D) (direct kernel sum, so autograd gives first and second derivatives)."""
    d, g = em_grid(em, 201)
    logL = equation.gw_log_likelihood(equation.gw_distance(d, D), gw["d"], gw["w"], gw["h"])  # (nd,)
    return torch.logsumexp(logL + torch.log(g * (d[1] - d[0])), 0)  # log ∫ L N dd (rectangle rule)


def derivatives(fn, x: torch.Tensor):
    """Value, gradient and Hessian of a scalar function by autograd (x: (P,))."""
    x = x.detach().clone().requires_grad_(True)
    y = fn(x)
    (g,) = torch.autograd.grad(y, x, create_graph=True)
    H = torch.stack([torch.autograd.grad(g[i], x, retain_graph=True)[0] for i in range(len(x))])
    return y.detach(), g.detach(), H.detach()


def newton_max(fn, x0: torch.Tensor, steps: int = 40, max_step: float = 0.2) -> torch.Tensor:
    """Maximise fn by damped Newton steps on autograd derivatives (a gradient step where fn is not concave)."""
    x = x0.clone()
    for _ in range(steps):
        _, g, H = derivatives(fn, x)
        concave = bool((torch.linalg.eigvalsh(H) < 0).all())
        step = -torch.linalg.solve(H, g) if concave else 0.05 * g / g.norm().clamp_min(1e-12)
        x = x + step.clamp(-max_step, max_step)
        if step.abs().max() < 1e-9:
            break
    return x


def fit_D(gw: dict, em: dict, D0: float = 4.5) -> dict:
    """MAP of D by Newton's method on autograd derivatives of log p(D), and its Laplace error (-d²log p/dD²)^-1/2."""
    fn = lambda x: log_posterior(x[0], gw, em)  # noqa: E731
    D = newton_max(fn, torch.tensor([D0], dtype=F64))
    _, g, H = derivatives(fn, D)
    return {"D": float(D[0]), "sigma": float((-H[0, 0]).clamp_min(1e-12).rsqrt()), "grad": float(g[0])}


def screening_bound(D_vals: torch.Tensor, n_vals: torch.Tensor, gw: dict, em: dict, table=None) -> dict:
    """Smallest allowed R_c for each (D, n): where -2 ln[L(R_c)/L(no leakage)] drops below 2.71 (95%, one-sided)."""
    table = likelihood_table(gw) if table is None else table
    d, g = em_grid(em, 201)
    Rc = torch.logspace(-1, 4, 400, dtype=F64)  # 0.1 ... 10^4 Mpc
    dgw = equation.gw_distance(d, D_vals[:, None, None, None], Rc[None, None, :, None], n_vals[None, :, None, None])
    like = torch.trapezoid(equation.interp(dgw, D_TABLE, table) * g, d, dim=-1)  # (nD, nn, nR)
    like0 = torch.trapezoid(equation.interp(d, D_TABLE, table) * g, d)
    dchi2 = -2 * torch.log(like.clamp_min(1e-300) / like0)
    excluded = dchi2 > 2.71
    last = torch.where(excluded, torch.arange(len(Rc)), torch.full_like(excluded, -1, dtype=torch.long)).amax(-1)
    bound = torch.where(last >= 0, Rc[(last + 1).clamp(max=len(Rc) - 1)], Rc[0])
    return {"Rc": Rc, "dchi2": dchi2, "bound": bound, "D": D_vals, "n": n_vals}


# ---------------------------------------------------------------- shadows
def trace(b: torch.Tensor, r_cap: torch.Tensor, phi_max: float = 6 * math.pi, dphi: float = 2e-3) -> torch.Tensor:
    """Ray-trace photons of impact parameter b (units r_s) inward from infinity; True where captured.

    b, r_cap broadcast to (F, B). Integrates (u, u') with RK4 in φ; a ray stops when it reaches u = 1/r_cap
    (captured) or comes back to u = 0 (escaped)."""
    b, r_cap = torch.broadcast_tensors(b, r_cap)
    y = torch.stack([torch.zeros_like(b), 1 / b], -1)  # (F, B, 2) start at infinity
    alive = torch.ones_like(b, dtype=torch.bool)
    captured = torch.zeros_like(alive)
    for _ in range(int(phi_max / dphi)):
        k1 = equation.photon_rhs(y)
        k2 = equation.photon_rhs(y + 0.5 * dphi * k1)
        k3 = equation.photon_rhs(y + 0.5 * dphi * k2)
        k4 = equation.photon_rhs(y + dphi * k3)
        y = torch.where(alive[..., None], y + dphi / 6 * (k1 + 2 * k2 + 2 * k3 + k4), y)
        hit = alive & (y[..., 0] >= 1 / r_cap)
        out = alive & (y[..., 0] < 0)
        captured |= hit
        alive &= ~(hit | out)
        if not alive.any():
            break
    return captured


def traced_shadow_edge(r_f: torch.Tensor, n: int = 400) -> torch.Tensor:
    """Shadow radius (units r_s) by ray tracing, for folds at r_f (<= 1: none, capture at the horizon).  (F,)

    Two passes: a coarse grid of b, then a fine one around the first escaping ray."""
    r_cap = r_f.clamp_min(1.0)[:, None]
    lo = torch.full_like(r_f, 2.3)
    hi = torch.maximum(torch.full_like(r_f, 3.0), 1.3 * r_cap[:, 0])
    for _ in range(2):
        s = torch.linspace(0, 1, n, dtype=F64)
        b = lo[:, None] + (hi - lo)[:, None] * s  # (F, n)
        cap = trace(b, r_cap)
        k = (~cap).long().argmax(1).clamp(1, n - 1)  # first escaping ray
        idx = torch.arange(len(r_f))
        lo, hi = b[idx, k - 1], b[idx, k]
    return 0.5 * (lo + hi)


def shadow_tests(eht: list[dict], spin_inc_max: float = 90.0, m87_inc: float = 17.0) -> dict:
    """δ for each EHT row, the spin band, the allowed fold growth and the bound on r_f/r_s.

    δ_up = δ + 2σ_+ (about 95%). A fold enlarges the shadow; spin can shrink it by up to |δ_Kerr,min|, so the fold
    may grow it by δ_up - δ_Kerr,min. M87*: δ = θ_g(EHT)/θ_g(dyn) - 1, inclination 17° (jet); Sgr A*: published δ,
    inclination up to spin_inc_max."""
    spins = torch.linspace(1e-3, 0.998, 40, dtype=F64)
    rows = []
    for r in eht:
        if r["delta"] is None:
            ratio = r["thetag_eht"] / r["thetag_dyn"]
            delta = ratio - 1
            sig = ratio * math.hypot(r["thetag_eht_err"] / r["thetag_eht"], r["thetag_dyn_err"] / r["thetag_dyn"])
            lo = hi = sig
            inc = torch.tensor([m87_inc], dtype=F64)
        else:
            delta, lo, hi = r["delta"], r["delta_lo"], r["delta_hi"]
            inc = torch.linspace(1.0, spin_inc_max, 30, dtype=F64)
        kerr_min = float(equation.kerr_delta(spins, inc).min())
        delta_up = delta + 2 * hi
        allowed = delta_up - kerr_min
        rows.append({**r, "delta": delta, "delta_lo": lo, "delta_hi": hi, "kerr_min": kerr_min, "delta_up": delta_up,
                     "allowed": allowed, "r_bound": float(equation.fold_bound(torch.tensor(allowed, dtype=F64))),
                     "shadow_schw": 2 * math.sqrt(27) * r["thetag_dyn"],
                     "pull": delta / (hi if delta < 0 else lo)})  # fmt: skip
    return {"rows": rows, "spins": spins}


# ---------------------------------------------------------------- supernovae
def supernova_fit(z, mu, sigma, n_a: int = 71, n_m: int = 56) -> dict:
    """Grid over (α, Ω_m) with the offset marginalised, the α posterior, ΛCDM vs DGP, and an autograd refinement."""
    nodes = torch.linspace(0, 1.01 * float(z.max()), 800, dtype=F64)
    alpha = torch.linspace(-1.0, 1.8, n_a, dtype=F64)
    Om = torch.linspace(0.05, 0.6, n_m, dtype=F64)
    chi2 = equation.chi2_offset_marginal(equation.distance_modulus(z, nodes, Om[None, :], alpha[:, None]), mu, sigma)
    post = torch.exp(-0.5 * (chi2 - chi2.min()))  # (A, M), uniform priors
    pa = torch.trapezoid(post, Om, dim=1)
    pa = pa / torch.trapezoid(pa, alpha)
    mean = torch.trapezoid(alpha * pa, alpha)
    sd = torch.trapezoid((alpha - mean) ** 2 * pa, alpha).sqrt()

    def best_at(a_val: float):
        c = equation.chi2_offset_marginal(equation.distance_modulus(z, nodes, Om, torch.tensor(a_val, dtype=F64)),
                                          mu, sigma)  # fmt: skip
        return float(c.min()), float(Om[c.argmin()])

    lcdm, dgp = best_at(0.0), best_at(1.0)
    i, j = divmod(int(chi2.argmin()), n_m)
    fn = lambda p: -0.5 * equation.chi2_offset_marginal(equation.distance_modulus(z, nodes, p[1], p[0]), mu, sigma)  # noqa: E731
    p = newton_max(fn, torch.tensor([float(alpha[i]), float(Om[j])], dtype=F64), steps=20, max_step=0.05)
    a_fit, om_fit = (float(x) for x in p)
    shape = {k: equation.distance_modulus(z, nodes, torch.tensor(o, dtype=F64), torch.tensor(a, dtype=F64)).detach()
             for k, a, o in (("lcdm", 0.0, lcdm[1]), ("dgp", 1.0, dgp[1]), ("fit", a_fit, om_fit))}  # fmt: skip
    return {"alpha": alpha, "Om": Om, "chi2": chi2, "pa": pa, "mean": float(mean), "sd": float(sd),
            "lcdm": lcdm, "dgp": dgp, "a_fit": a_fit, "om_fit": om_fit, "shape": shape,
            "z": z, "mu": mu, "sigma": sigma, "n": len(z)}  # fmt: skip
```

## 5. `plot.py` — 2D/3D plot → GIF

```python
%%writefile problems/cosmology/projection_telescope_tests/plot.py
"""Animate the telescope tests: GW amplitude against distance for a sweep of D, the posterior of D, the screening
bound, the supernova Hubble diagram, the shadow ring for a moving fold against the EHT bands, and the summary."""

import math
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import torch
from PIL import Image

from problems._common.plotting import ACCENT, export_plot_data, frame_indices
from problems._common.progress import report

from . import equation, solve

F64 = torch.float64
GREY, INK = "#6b7280", "#111827"
EM_COLOR = {"sbf": ACCENT[0], "shoes": ACCENT[4], "planck": ACCENT[3]}
EM_NAME = {"sbf": "SBF (Cantiello 2018)", "shoes": "Hubble flow, SH0ES $H_0$", "planck": "Hubble flow, Planck $H_0$"}


def summary_rows(posts, sh, sn, pardo) -> list[dict]:
    """One row per constraint: its deviation from GR in 3+1 in standard deviations, and the value as text."""
    rows = []
    for k in ("sbf", "shoes", "planck"):
        m, (lo, hi) = posts[k]["map"], posts[k]["interval"]
        s = (hi - m) if m < 4 else (m - lo)
        text = f"D = {m:.2f} (+{hi - m:.2f}/-{m - lo:.2f})"
        rows.append({"label": f"GW170817 D, {k}", "pull": (m - 4) / s, "text": text, "group": "gw"})
    for k, (m, lo, hi) in pardo.items():
        s = hi if m < 4 else lo
        rows.append({"label": f"Pardo 2018 D, {k}", "pull": (m - 4) / s, "text": f"D = {m:.2f} (+{hi:.2f}/-{lo:.2f})",
                     "group": "pub"})  # fmt: skip
    for r in sh["rows"]:
        name = "M87*" if r["key"] == "m87" else f"Sgr A* ({r['key'].split('_')[1]})"
        rows.append({"label": f"EHT shadow, {name}", "pull": r["pull"],
                     "text": f"delta = {r['delta']:+.2f}, r_f <= {r['r_bound']:.2f} r_s", "group": "eht"})  # fmt: skip
    rows.append({"label": "Pantheon+ leakage alpha", "pull": sn["mean"] / sn["sd"],
                 "text": f"alpha = {sn['mean']:.2f} +- {sn['sd']:.2f}", "group": "sn"})  # fmt: skip
    return rows


def _interval(x: np.ndarray, pdf: np.ndarray, q: float) -> float:
    cdf = np.concatenate([[0], np.cumsum(0.5 * (pdf[1:] + pdf[:-1]) * np.diff(x))])
    return float(np.interp(q * cdf[-1], cdf, x))


def animate(gw, table, posts, key, scr, i_n, sh, sn, summary, path: Path) -> Path:
    fig, axs = plt.subplots(2, 4, figsize=(14.5, 7.0))
    (ax_h, ax_p, ax_r, ax_s), (ax_i, ax_d, ax_a, ax_f) = axs

    # ---- (a1) amplitude against distance, relative to 4D gravity ----
    d = np.logspace(0, 3, 200)
    dt = torch.tensor(d, dtype=F64)

    def ratio(D):  # h / h_4D = d / d_GW
        return (equation.strain_amplitude(dt, torch.tensor(D, dtype=F64)) * dt).numpy()

    for D, ls in ((3.8, ":"), (4.2, "--"), (4.5, "-.")):
        ax_h.semilogx(d, ratio(D), color=GREY, lw=1, ls=ls)
        ax_h.text(d[-1] * 0.9, ratio(D)[-1] + 0.04, f"D = {D:g}", fontsize=7, color=GREY, ha="right")
    ax_h.axhline(1, color=GREY, lw=1)
    ax_h.text(1.2, 1.04, "D = 4 (GR)", fontsize=7, color=GREY)
    x, L = solve.D_TABLE.numpy(), table.numpy()
    g_med, g_lo, g_hi = (_interval(x, L, q) for q in (0.5, 0.16, 0.84))
    from . import tensors

    for j, k in enumerate(("sbf", "shoes", "planck")):
        em = tensors.em_distance(k)
        y = em["d"] / g_med
        ax_h.errorbar(em["d"] * (1 + 0.04 * (j - 1)), y, xerr=em["sigma"],
                      yerr=[[y - em["d"] / g_hi], [em["d"] / g_lo - y]],
                      fmt="o", color=EM_COLOR[k], ms=5, capsize=2, lw=1.2, label=EM_NAME[k])  # fmt: skip
    sweep = ax_h.semilogx(d, ratio(4.0), color=INK, lw=2)[0]
    ax_h.set_xlim(1, 1000)
    ax_h.set_ylim(0, 2.2)
    ax_h.set_xlabel("distance $d_{EM}$ (Mpc)")
    ax_h.set_ylabel("$h / h_{4D} = d_{EM}/d_{GW}$ (dimensionless)")
    ax_h.legend(fontsize=6.5, loc="upper left")
    th = ax_h.set_title("", fontsize=9)

    # ---- (a2) posterior of D ----
    Ds = posts[key]["D"].numpy()
    for k in ("sbf", "shoes", "planck"):
        ax_p.plot(Ds, posts[k]["pdf"].numpy(), color=EM_COLOR[k], lw=2.2 if k == key else 1.2, label=EM_NAME[k])
    top = max(float(p["pdf"].max()) for p in posts.values())
    for j, (k, (m, lo, hi)) in enumerate((("shoes", (4.02, 0.10, 0.07)), ("planck", (3.98, 0.09, 0.07)))):
        y = top * (1.08 + 0.07 * j)
        ax_p.errorbar(m, y, xerr=[[lo], [hi]], fmt="s", color=EM_COLOR[k], ms=4, capsize=2, mfc="white")
        ax_p.text(m - lo - 0.03, y, f"Pardo 2018 ({k})", fontsize=6.5, va="center", ha="right", color=INK)
    lo, hi = posts[key]["interval"]
    ax_p.axvspan(lo, hi, color="#dbeafe", lw=0, zorder=0)
    ax_p.axvline(4, color=INK, lw=0.8, ls=":")
    ax_p.set_xlim(3.3, 4.6)
    ax_p.set_ylim(0, top * 1.45)
    ax_p.set_xlabel("spacetime dimensions gravity sees, $D$ (dimensionless)")
    ax_p.set_ylabel("posterior density (per unit $D$)")
    ax_p.legend(fontsize=6.5, loc="center left")
    ax_p.set_title(f"$D$ = {posts[key]['map']:.2f} (+{hi - posts[key]['map']:.2f}/-{posts[key]['map'] - lo:.2f}), "
                   "68% shaded", fontsize=9)  # fmt: skip
    cur_p = ax_p.axvline(4, color=INK, lw=1.5)

    # ---- (a3) screening scale ----
    nD = len(scr["D"]) - 1
    Dv = scr["D"][:nD].numpy()
    for j, n in enumerate(scr["n"].tolist()):
        bnd = scr["bound"][:nD, j].numpy()
        ax_r.semilogy(Dv, bnd, color=ACCENT[j % len(ACCENT)], lw=2.0 if j == i_n else 1.2, label=f"n = {n:g}")
    ax_r.axhline(20, color=GREY, ls="--", lw=1)
    ax_r.text(7.9, 23, "Pardo 2018: $R_c \\gtrsim 20$ Mpc", fontsize=6.5, ha="right", color=GREY)
    ax_r.axhline(float(scr["Rc"][-1]), color=GREY, lw=0.6)
    ax_r.set_ylim(5, 2e4)
    ax_r.set_xlabel("$D$ beyond the screening scale (dimensionless)")
    ax_r.set_ylabel("smallest allowed $R_c$ (Mpc)")
    ax_r.legend(fontsize=6.5, loc="upper right", title="steepness", title_fontsize=6.5)
    ax_r.set_title("screened leakage: excluded below the curves\n(curves at $10^4$ Mpc: grid edge)", fontsize=9)

    # ---- (c1) supernova Hubble residuals ----
    z, mu, sig = (sn[k].numpy() for k in ("z", "mu", "sigma"))
    w = sig**-2

    def offset_model(shape):
        s = shape.numpy()
        return s + np.sum(w * (mu - s)) / np.sum(w)

    base = offset_model(sn["shape"]["lcdm"])
    bins = np.logspace(np.log10(z.min()), np.log10(z.max() * 1.001), 16)
    idx = np.digitize(z, bins) - 1
    zc = np.array([np.sum(w[idx == i] * z[idx == i]) / np.sum(w[idx == i]) for i in range(15)])
    rc = np.array([np.sum(w[idx == i] * (mu - base)[idx == i]) / np.sum(w[idx == i]) for i in range(15)])
    re = np.array([np.sum(w[idx == i]) ** -0.5 for i in range(15)])
    ax_s.scatter(z, mu - base, s=2, color="#d1d5db", zorder=1)
    ax_s.errorbar(zc, rc, yerr=re, fmt="o", color=INK, ms=4, capsize=2, zorder=3, label="Pantheon+ (binned)")
    ax_s.axhline(0, color=ACCENT[0], lw=1.6, label=f"$\\Lambda$CDM, $\\Omega_m$ = {sn['lcdm'][1]:.2f}")
    ax_s.plot(z, offset_model(sn["shape"]["dgp"]) - base, color=ACCENT[1], lw=1.6,
              label=f"DGP leakage, $\\Omega_m$ = {sn['dgp'][1]:.2f}")  # fmt: skip
    ax_s.plot(z, offset_model(sn["shape"]["fit"]) - base, color=ACCENT[2], lw=1.2, ls="--",
              label=f"best $\\alpha$ = {sn['a_fit']:.2f}")  # fmt: skip
    ax_s.set_xscale("log")
    ax_s.set_ylim(-0.3, 0.3)
    ax_s.set_xlabel("redshift $z$ (dimensionless)")
    ax_s.set_ylabel("$\\mu - \\mu_{\\Lambda CDM}$ (mag)")
    ax_s.legend(fontsize=6.5, loc="lower left")
    ax_s.set_title(f"supernovae: $\\chi^2$ $\\Lambda$CDM {sn['lcdm'][0]:.1f}, DGP {sn['dgp'][0]:.1f}", fontsize=9)

    # ---- (b1) shadow ring image for Sgr A* (VLTI theta_g) ----
    row = next(r for r in sh["rows"] if r["key"] == "sgra_vlti")
    tg = row["thetag_dyn"]  # uas
    ext = 60.0
    g = np.linspace(-ext, ext, 161)
    rho = np.hypot(*np.meshgrid(g, g))
    beam = 20.0 / 2.355  # EHT resolution ~ 20 uas FWHM (schematic blur)
    img = ax_i.imshow(np.zeros_like(rho), extent=(-ext, ext, -ext, ext), origin="lower", cmap="afmhot", vmin=0,
                      vmax=1.05)  # fmt: skip
    ax_i.grid(False)
    t = np.linspace(0, 2 * np.pi, 200)
    r_sch = math.sqrt(27) * tg
    bands = []  # drawn over the (moving) image every frame
    for dd, ls, lab in ((row["delta"] - row["delta_lo"], ":", None), (row["delta"] + row["delta_hi"], ":", "68%"),
                        (row["delta_up"], "--", "95% upper")):  # fmt: skip
        bands += ax_i.plot(
            (1 + dd) * r_sch * np.cos(t), (1 + dd) * r_sch * np.sin(t), color="white", lw=1.2, ls=ls, label=lab
        )
    edge_line = ax_i.plot([], [], color="#22d3ee", lw=1.6, label="shadow edge")[0]
    ax_i.set_xlabel("$\\Delta\\alpha$ ($\\mu$as)")
    ax_i.set_ylabel("$\\Delta\\delta$ ($\\mu$as)")
    leg_i = ax_i.legend(fontsize=6, loc="lower right", facecolor="black", labelcolor="white", framealpha=0.6)
    ti = ax_i.set_title("", fontsize=9)

    # ---- (b2) delta against the fold radius ----
    rf = np.linspace(1.0, 3.2, 300)
    dl = equation.fold_delta(torch.tensor(rf, dtype=F64)).numpy()
    ax_d.plot(rf, dl, color=INK, lw=2, label="fold model $\\delta(r_f)$")
    kmin = min(r["kerr_min"] for r in sh["rows"])
    ax_d.axhspan(kmin, 0, color="#e5e7eb", lw=0, label="Kerr spin band")
    cols = {"m87": ACCENT[5], "sgra_keck": ACCENT[4], "sgra_vlti": ACCENT[3]}
    for r in sh["rows"]:
        c = cols[r["key"]]
        name = "M87*" if r["key"] == "m87" else f"Sgr A* {r['key'].split('_')[1]}"
        ax_d.errorbar(1.15 + 0.1 * list(cols).index(r["key"]), r["delta"], yerr=[[r["delta_lo"]], [r["delta_hi"]]],
                      fmt="o", color=c, ms=4, capsize=2)  # fmt: skip
        ax_d.axhline(r["allowed"], color=c, lw=1, ls="--")
        ax_d.plot([r["r_bound"]], [r["allowed"]], "v", color=c, ms=7, label=f"{name}: $r_f \\leq$ {r['r_bound']:.2f}")
    ax_d.axvline(1.5, color=GREY, lw=0.8, ls=":")
    ax_d.text(1.52, 0.42, "photon\nsphere", fontsize=6.5, color=GREY)
    ax_d.set_xlim(1.0, 3.2)
    ax_d.set_ylim(-0.25, 0.5)
    ax_d.set_xlabel("fold radius $r_f / r_s$ (dimensionless)")
    ax_d.set_ylabel("shadow change $\\delta$ (dimensionless)")
    ax_d.legend(fontsize=6, loc="upper left")
    ax_d.set_title("dots: measured $\\delta$; dashed: growth a fold may add", fontsize=9)
    dot_d = ax_d.plot([], [], "o", color="#f59e0b", ms=8)[0]

    # ---- (c2) posterior of the leakage index ----
    a, pa = sn["alpha"].numpy(), sn["pa"].numpy()
    ax_a.fill_between(a, pa, color="#bbf7d0", lw=0)
    ax_a.plot(a, pa, color=ACCENT[2], lw=2)
    for v, lab, c in ((0.0, "$\\Lambda$CDM", ACCENT[0]), (1.0, "DGP", ACCENT[1])):
        ax_a.axvline(v, color=c, lw=1.2, ls="--")
        ax_a.text(v + 0.03, pa.max() * 1.02, lab, fontsize=7, color=c)
    ax_a.set_ylim(0, pa.max() * 1.15)
    ax_a.set_xlabel("leakage index $\\alpha$ (dimensionless)")
    ax_a.set_ylabel("posterior density (per unit $\\alpha$)")
    ax_a.set_title(f"Pantheon+: $\\alpha$ = {sn['mean']:.2f} $\\pm$ {sn['sd']:.2f} (both allowed)", fontsize=9)

    # ---- summary ----
    group_c = {"gw": ACCENT[0], "pub": GREY, "eht": ACCENT[3], "sn": ACCENT[2]}
    ys = np.arange(len(summary))[::-1]
    ax_f.axvspan(-2, 2, color="#f3f4f6", lw=0)
    ax_f.axvspan(-1, 1, color="#e5e7eb", lw=0)
    ax_f.axvline(0, color=INK, lw=1)
    for y, r in zip(ys, summary):
        ax_f.plot([r["pull"]], [y], "o", color=group_c[r["group"]], ms=7)
        ax_f.text(-3.9, y + 0.22, f"{r['label']}: {r['text']}", fontsize=6.2, color=INK)
    ax_f.set_xlim(-4, 4)
    ax_f.set_ylim(-0.8, len(summary) - 0.2)
    ax_f.set_yticks([])
    ax_f.set_xlabel("deviation from GR in 3+1 (standard deviations)")
    ax_f.set_title("every test: consistent with $D = 4$ (bands 1, 2 $\\sigma$)", fontsize=9)

    # ---- frames: sweep D, then sweep the fold ----
    n1, n2, hold = 36, 36, 6
    D_seq = np.concatenate([np.linspace(3.4, 4.6, n1 // 2), np.linspace(4.6, 4.0, n1 - n1 // 2)])
    r_seq = np.concatenate([np.linspace(1.0, 3.0, n2 * 3 // 4), np.linspace(3.0, 1.5, n2 - n2 * 3 // 4)])
    pdf_sel = posts[key]["pdf"].numpy()
    total = n1 + n2 + hold

    def update(f: int):
        D = D_seq[min(f, n1 - 1)]
        sweep.set_ydata(ratio(D))
        th.set_text(f"GW170817: $h\\propto d^{{-(D-2)/2}}$, D = {D:.2f}\n"
                    f"posterior density {np.interp(D, Ds, pdf_sel):.2f}")  # fmt: skip
        cur_p.set_xdata([D])
        r = r_seq[min(max(f - n1, 0), n2 - 1)]
        b = float(equation.fold_shadow_radius(torch.tensor([r], dtype=F64))[0])
        R = b / (math.sqrt(27) / 2) * r_sch  # shadow edge in uas
        ring = 1.1 * R  # emission ring just outside the edge (schematic)
        img.set_data(np.exp(-0.5 * ((rho - ring) / math.hypot(2.0, beam)) ** 2))
        edge_line.set_data(R * np.cos(t), R * np.sin(t))
        dlt = b / (math.sqrt(27) / 2) - 1
        ti.set_text(f"Sgr A* (VLTI $\\theta_g$), fold at $r_f$ = {r:.2f} $r_s$\n$\\delta$ = {dlt:+.3f}")
        dot_d.set_data([r], [dlt])

    fig.subplots_adjust(left=0.05, right=0.985, bottom=0.08, top=0.9, wspace=0.33, hspace=0.42)
    fig.suptitle("Projection Hypothesis step 6 (speculative): do telescopes see extra dimensions or a fold?",
                 fontsize=11)  # fmt: skip
    moving = [sweep, th, cur_p, img, *bands, edge_line, leg_i, ti, dot_d]
    return save_gif_shared_palette(fig, update, frame_indices(total, 78), path, fps=7, moving=moving)


def save_gif_shared_palette(fig, update, frames: list[int], path: Path, fps: int = 8, moving=()) -> Path:
    """Render every frame, then quantise all of them with ONE palette built from a strip of frames, so colours do
    not drift between frames (Pillow otherwise picks a new 256-colour palette per frame).

    The static parts are drawn once; each frame restores that background and redraws only the moving artists."""
    fig.set_dpi(80)
    for art in moving:
        art.set_animated(True)
    fig.canvas.draw()
    background = fig.canvas.copy_from_bbox(fig.bbox)
    shots = []
    for i, f in enumerate(frames):
        fig.canvas.restore_region(background)
        update(f)
        for art in moving:
            fig.draw_artist(art)
        shots.append(Image.fromarray(np.asarray(fig.canvas.buffer_rgba())[..., :3]))
        report(i + 1, len(frames), "frames")
    export_plot_data(fig, path)  # numbers behind the final frame (Results page)
    plt.close(fig)
    sample = shots[:: max(1, len(shots) // 12)]
    strip = Image.new("RGB", (sample[0].width, sample[0].height * len(sample)))
    for k, im in enumerate(sample):
        strip.paste(im, (0, k * im.height))
    palette = strip.quantize(colors=255, method=Image.Quantize.MEDIANCUT)
    frames_p = [im.quantize(palette=palette, dither=Image.Dither.NONE) for im in shots]
    path.parent.mkdir(parents=True, exist_ok=True)
    tmp = path.with_name(f".{path.stem}.tmp{path.suffix}")
    frames_p[0].save(tmp, save_all=True, append_images=frames_p[1:], duration=int(1000 / fps), loop=0)
    return tmp.replace(path)
```
