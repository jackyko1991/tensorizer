# The Riemann hypothesis in thousands of zeros

*maths · advanced* · [open in Tensorizer](https://jackyko1991.github.io/tensorizer/#/maths/riemann_hypothesis)

Locate thousands of nontrivial zeros of $\zeta$ on the critical line with the Riemann–Siegel formula in tensor form, count them against the Riemann–von Mangoldt law $N(T) = \theta(T)/\pi + 1 + S(T)$ and Gram's law, compare their spacings and pair correlation with random Hermitian (GUE) matrices, and rebuild the prime-power staircase $\psi(x)$ from them. The hypothesis itself, that every nontrivial zero has real part $\frac12$, is still open.

**Vector form**

$$Z(t) = 2\sum_{n\le N} \frac{\cos(\theta(t) - t\ln n)}{\sqrt n} + (-1)^{N-1}\Big(\frac{t}{2\pi}\Big)^{-\frac14}\sum_{j=0}^{2} C_j(p)\Big(\frac{t}{2\pi}\Big)^{-\frac j2},\qquad N(T) = \frac{\theta(T)}{\pi} + 1 + S(T),\qquad \psi(x) = x - \sum_\rho \frac{x^\rho}{\rho} - \ln 2\pi - \tfrac12\ln(1 - x^{-2})$$

**Index (tensor) form**

$$Z_t = 2\,C_{tn}\,w_n + (-1)^{N_t-1} a_t^{-1/2}\big(C_0 + C_1 a_t^{-1} + C_2 a_t^{-2}\big),\quad C_{tn} = [n \le N_t]\cos(\theta_t - t\ln n),\ w_n = n^{-1/2}\qquad R_2(u) = 1 - \Big(\frac{\sin \pi u}{\pi u}\Big)^2$$

| Symbol | Code | Shape | Meaning |
|---|---|---|---|
| $C_{tn}\,w_n$ | `einsum('tn,n->t', terms, n.rsqrt())` | (T, N) → (T,) | repeated $n$ ⇒ summed: the main sum for 250 000 heights at once, masked to $n \le N_t = \lfloor\sqrt{t/2\pi}\rfloor$ |
| $\Psi(p) = \frac{\cos 2\pi(p^2 - p - 1/16)}{\cos 2\pi p}$ | `chebyshev_basis(p, 26) @ psi_coefficients()` | (T,) | $\Psi$ is entire but the quotient is $0/0$ at $p = \frac14, \frac34$; a Chebyshev fit removes that |
| $C_1 = -\frac{\Psi^{(3)}}{96\pi^2},\ C_2 = \frac{\Psi^{(2)}}{64\pi^2} + \frac{\Psi^{(6)}}{18432\pi^4}$ | `einsum('pk,jk->jp', basis, coeffs)` | (7, T) | the Riemann–Siegel corrections: a Chebyshev series differentiates term by term, $c'_{k-1} = c'_{k+1} + 2k\,c_k$ |
| $\gamma_k$ | `find_zeros(t, z)` | (K,) | sign changes of $Z$ on a grid with step 0.02, then 32 bisection steps on all brackets at once |
| $S(t)$ | `searchsorted(gammas, t) - theta(t) / pi - 1` | (T,) | zero count minus its smooth part; stays small and averages to 0 |
| $g_n:\ \theta(g_n) = n\pi$ | `g - (theta(g) - n * pi) / theta_prime(g)` | (G,) | all Gram points by Newton at once; Gram's law says $(-1)^n Z(g_n) > 0$ |
| $x_{k+j} - x_k$ | `x[k[:, None] + j[None, :]] - x[k, None]` | (K, J) | differences of unfolded zeros $x = \theta(\gamma)/\pi + 1$ to the next $J = 10$ neighbours, for the pair correlation |
| $\sum_{\rho} x^\rho/\rho$ | `cumsum(2 * (exp(rho[:, None] * log(x)) / rho[:, None]).real, 0)` | (K, X) | every zero at every $x$, accumulated over the zeros: one row per height $T$ |

This page complements the problem "The Riemann zeta function and the primes" (maths/zeta_primes), which builds $\zeta$ from scratch with Euler–Maclaurin, finds the first zeros and rebuilds $\pi(x)$; read it first for the basics. Here the faster Riemann–Siegel formula, with its corrections $C_0, C_1, C_2$ (derivatives of $\Psi$ from a Chebyshev series), evaluates $Z(t)$ at a quarter of a million heights and finds every zero on the critical line below $T$; at $t \ge 100$ it agrees with a high-precision evaluation (mpmath) to better than $10^{-5}$. Below $t = 1000$ it finds 649 sign changes and below 5000 it finds 4520, exactly the total numbers of zeros in the strip up to those heights (computed independently with mpmath), so all of them lie on the line; the counting function follows $\theta(T)/\pi + 1$ with a small fluctuation $S(t)$ that averages to zero. Gram's law holds at most Gram points and first fails at $g_{126} \approx 282.45$. After unfolding to mean spacing 1, the zeros repel each other exactly like the eigenvalues of large random Hermitian matrices (GUE): Montgomery (1973) conjectured the pair correlation $1 - (\sin \pi u / \pi u)^2$, proved it, assuming the Riemann hypothesis, for test functions with Fourier support in $(-1, 1)$, and Odlyzko's computations near the $10^{20}$-th zero confirm it spectacularly; at the low heights here the agreement is already close. The explicit formula (von Mangoldt 1895, proven) turns the zeros into the primes: each pair $\frac12 \pm i\gamma$ adds a wave $x^{1/2}\cos(\gamma\ln x)$, and with thousands of zeros $\psi(x)$ becomes a sharp staircase. The Riemann hypothesis, that every nontrivial zero has real part $\frac12$, is one of the Clay Millennium Prize problems and is open. It has been verified for all zeros up to height $3\times10^{12}$ (Platt and Trudgian 2021). Proven results shown in the strip: no zeros on $\operatorname{Re} s = 1$ (Hadamard, de la Vallée Poussin 1896, equivalent to the prime number theorem) and the zero-free region $\sigma > 1 - 1/(5.558691\ln t)$ (Mossinghoff and Trudgian 2015), mirrored by the functional equation. Claimed, not established: in October 2026 OpenAI published (as reported; we have not checked the manuscripts), in its public openai/math repository of AI-generated manuscripts, a claimed zero-free half-plane $\operatorname{Re} s > 7/8$ for $\zeta$ and every Dirichlet $L$-function, with a companion argument for $\operatorname{Re} s > 11/12$. These preprints are unrefereed and have not been independently verified; if correct, a zero-free strip of fixed width would be a major advance, yet still far from the line $\frac12$. No computation can prove the hypothesis: a finite search can only find zeros on the line, or a counterexample.

## 1. `main.py` — main loop

```python
%%writefile problems/maths/riemann_hypothesis/main.py
"""Main loop: parameters → Z(t) on a long grid → zeros → counting, statistics, explicit formula → plot."""

import math
from pathlib import Path

import torch

from problems._common.runtime import cli, device, fast_mode

from . import equation, plot, solve, tensors


def run(params: dict, out_dir: Path) -> Path:
    dev = device()
    t_max = 600.0 if fast_mode() else params["t_max"]

    # 1. thousands of zeros of Z(t) on the critical line (Riemann–Siegel formula, sign changes, bisection)
    t = tensors.t_grid(t_max, 0.02, dev)  # (T,)
    z = solve.z_on_grid(t)
    gammas = solve.find_zeros(t, z)
    print(f"Z(t) at {len(t):,} heights 10 ≤ t ≤ {t_max:.0f}: {len(gammas):,} sign changes = zeros on the line")
    for k, known in tensors.KNOWN_ZEROS.items():
        if k <= len(gammas):
            print(f"  γ_{k:<5d} = {gammas[k - 1].item():.9f}   published {known:.9f}")
    s = solve.s_function(gammas, t[::10])
    print(f"N(T) = {len(gammas)}; smooth count θ(T)/π + 1 = {equation.smooth_count(torch.tensor(t_max)).item():.2f}")
    print(f"S(t) = N(t) − θ(t)/π − 1: mean {s.mean().item():+.3f}, range [{s.min().item():.2f}, {s.max().item():.2f}]")

    # 2. Gram points and Gram's law
    n_gram = int(equation.smooth_count(torch.tensor(t_max)).item()) - 2
    gram = tensors.gram_points(n_gram, dev)
    gram = gram[gram < t_max]
    bad = solve.gram_violations(equation.riemann_siegel_z(gram))
    print(
        f"Gram points below T: {len(gram)}; Gram's law (−1)^n Z(g_n) > 0 fails at {len(bad)} "
        f"({100 * len(bad) / len(gram):.1f}%), first at n = {bad[0].item() if len(bad) else '–'}"
    )

    # 3. statistics of the unfolded zeros versus GUE random matrices
    x = solve.unfold(gammas)
    spacing = x.diff()
    k = int(spacing.argmin())
    print(
        f"unfolded spacings: mean {spacing.mean().item():.4f}, closest pair γ = {gammas[k].item():.4f}, "
        f"{gammas[k + 1].item():.4f} (spacing {spacing[k].item():.3f} of the mean)"
    )
    edges = torch.linspace(0, 3, 31, dtype=torch.float64, device=dev)
    r2 = solve.pair_correlation(x, edges)
    mid = 0.5 * (edges[1:] + edges[:-1])
    dev_r2 = (r2 - equation.montgomery_r2(mid)).abs().mean().item()
    print(f"pair correlation: mean |R₂ − (1 − (sin πu/πu)²)| = {dev_r2:.3f} on 0 ≤ u ≤ 3 (Montgomery–Odlyzko)")

    # 4. explicit formula: ψ(x) rebuilt from the zeros below each height
    x_max = params["x_max"]
    xs = torch.linspace(1.5, x_max, 1200, dtype=torch.float64, device=dev)
    psi_exact = tensors.chebyshev_psi(int(x_max), dev)  # (x_max + 1,)
    exact = psi_exact[xs.floor().long()]
    psi_k = solve.psi_from_zeros(xs, gammas)  # (K + 1, X)
    half = torch.arange(2.5, x_max, 1.0, dtype=torch.float64, device=dev)
    err = (solve.psi_from_zeros(half, gammas)[-1] - psi_exact[half.floor().long()]).abs()
    print(f"explicit formula with all {len(gammas)} zeros: max |ψ_K(x) − ψ(x)| = {err.max():.3f} at half-integers")

    heights = torch.logspace(math.log10(25), math.log10(t_max), 48).tolist()
    path = out_dir / "riemann_hypothesis.gif"
    return plot.animate(t, z, gammas, gram, bad, s, t[::10], spacing, x, edges, xs, exact, psi_k, heights, path)


if __name__ == "__main__":
    cli(run, __file__)
```

## 2. `equation.py` — physics → tensor translation

```python
%%writefile problems/maths/riemann_hypothesis/equation.py
r"""Mathematics → tensor translation.

1. Riemann–Siegel formula for the real function Z(t) = e^{iθ(t)} ζ(½ + it), vectorised over many heights t:

       Z(t) = 2 Σ_{n ≤ N} n^{−1/2} cos(θ(t) − t ln n) + (−1)^{N−1} a^{−1/2} [C₀(p) + C₁(p) a^{−1} + C₂(p) a^{−2}]

   with a = √(t/2π), N = ⌊a⌋, p = a − N and, from Ψ(p) = cos 2π(p² − p − 1/16) / cos 2πp,
   C₀ = Ψ,  C₁ = −Ψ⁽³⁾/(96π²),  C₂ = Ψ⁽²⁾/(64π²) + Ψ⁽⁶⁾/(18432π⁴).
   Ψ is entire; it and its derivatives come from one Chebyshev series. The sum over n is an einsum over the
   masked (T, N) matrix.

2. Zero counting. N(t) = #{zeros 0 < γ ≤ t} = θ(t)/π + 1 + S(t) (Riemann–von Mangoldt), with S(t) small.
   Gram points g_n solve θ(g_n) = nπ; "Gram's law" (−1)^n Z(g_n) > 0 usually holds.

3. Unfolding x = θ(γ)/π + 1 makes the mean spacing 1; Montgomery's pair correlation conjecture says the density of
   differences x_j − x_k is R₂(u) = 1 − (sin πu / πu)², the same as for eigenvalues of random Hermitian (GUE) matrices.

4. Explicit formula (von Mangoldt): ψ(x) = Σ_{p^k ≤ x} ln p = x − Σ_ρ x^ρ/ρ − ln 2π − ½ ln(1 − x^{−2}),
   the zeros taken in conjugate pairs: Σ_ρ x^ρ/ρ = Σ_k 2 Re(x^{ρ_k}/ρ_k).
"""

import functools
import math

import torch


def theta(t: torch.Tensor) -> torch.Tensor:
    r"""θ(t) = t/2 ln(t/2π) − t/2 − π/8 + 1/(48t) + 7/(5760t³) + 31/(80640t⁵)   (asymptotic, fine for t > 10)."""
    return (
        t / 2 * torch.log(t / (2 * math.pi))
        - t / 2
        - math.pi / 8
        + 1 / (48 * t)
        + 7 / (5760 * t**3)
        + 31 / (80640 * t**5)
    )


def psi(p: torch.Tensor) -> torch.Tensor:
    r"""Ψ(p) = cos 2π(p² − p − 1/16) / cos 2πp. Every zero of the denominator is also a zero of the numerator, so
    Ψ is entire, but the quotient loses precision near p = ¼, ¾, and its derivatives even more."""
    return torch.cos(2 * math.pi * (p * p - p - 1 / 16)) / torch.cos(2 * math.pi * p)


def chebyshev_basis(p: torch.Tensor, degree: int) -> torch.Tensor:
    r"""T_k(2p − 1) for k = 0 … degree by the recurrence T_{k+1} = 2z T_k − T_{k−1}.   (P,) → (P, degree + 1)"""
    z = 2 * p - 1
    basis = [torch.ones_like(z), z]
    for _ in range(degree - 1):
        basis.append(2 * z * basis[-1] - basis[-2])
    return torch.stack(basis, dim=-1)


@functools.cache
def psi_coefficients(degree: int = 26, device: torch.device | None = None) -> torch.Tensor:
    r"""Least-squares Chebyshev coefficients c_k of Ψ on [0, 1], fitted at 200 Chebyshev nodes (none of which lies
    within 1e-4 of ¼ or ¾, where the quotient is still accurate to ~1e-12). Computed once.   → (degree + 1,)"""
    j = torch.arange(200, dtype=torch.float64, device=device)
    nodes = 0.5 * (1 + torch.cos(math.pi * (j + 0.5) / 200))
    basis = chebyshev_basis(nodes, degree)  # (200, degree + 1)
    return torch.linalg.lstsq(basis, psi(nodes)[:, None]).solution[:, 0]


@functools.cache
def psi_derivative_coefficients(max_order: int = 6, device: torch.device | None = None) -> torch.Tensor:
    r"""Chebyshev coefficients of Ψ, Ψ′, …, Ψ⁽⁶⁾: the derivative of Σ c_k T_k(2p − 1) is again a Chebyshev series,
    with c′_{k−1} = c′_{k+1} + 2k c_k (times 2 from dz/dp = 2; c′_0 is halved).   → (max_order + 1, D + 1)"""
    rows = [psi_coefficients(device=device)]
    for _ in range(max_order):
        c, d = rows[-1], torch.zeros_like(rows[-1])
        for k in range(len(c) - 1, 0, -1):
            d[k - 1] = (d[k + 1] if k + 1 < len(c) else 0.0) + 2 * k * c[k]
        d[0] /= 2
        rows.append(2 * d)
    return torch.stack(rows)


def psi_derivatives(p: torch.Tensor) -> torch.Tensor:
    r"""[Ψ, Ψ′, …, Ψ⁽⁶⁾] at every p, from the differentiated Chebyshev series (smooth everywhere, no 0/0).
    (P,) → (7, P):  Ψ⁽ʲ⁾_p = B_{pk} c⁽ʲ⁾_k, k summed"""
    coeffs = psi_derivative_coefficients(device=p.device)  # (7, D + 1)
    basis = chebyshev_basis(p, coeffs.shape[1] - 1)  # (P, D + 1)
    return torch.einsum("pk,jk->jp", basis, coeffs)


def riemann_siegel_z(t: torch.Tensor) -> torch.Tensor:
    r"""Z(t) by the Riemann–Siegel formula with the corrections C₀, C₁, C₂.   t: (T,) float64 → (T,)"""
    a = torch.sqrt(t / (2 * math.pi))
    big_n = a.floor()
    n = torch.arange(1, int(big_n.max()) + 1, dtype=t.dtype, device=t.device)  # (N_max,)
    keep = n[None, :] <= big_n[:, None]  # (T, N_max): only n ≤ N(t) contributes
    phase = theta(t)[:, None] - t[:, None] * torch.log(n)[None, :]  # (T, N_max)
    terms = torch.where(keep, torch.cos(phase), 0.0)
    main = 2 * torch.einsum("tn,n->t", terms, n.rsqrt())  # repeated n ⇒ summed
    d = psi_derivatives(a - big_n)
    c0 = d[0]
    c1 = -d[3] / (96 * math.pi**2)
    c2 = d[2] / (64 * math.pi**2) + d[6] / (18432 * math.pi**4)
    sign = 1 - 2 * ((big_n - 1) % 2)  # (−1)^{N−1}
    return main + sign * a.rsqrt() * (c0 + c1 / a + c2 / a**2)


def smooth_count(t: torch.Tensor) -> torch.Tensor:
    r"""N̄(t) = θ(t)/π + 1: the smooth part of the zero count (Riemann–von Mangoldt)."""
    return theta(t) / math.pi + 1


def montgomery_r2(u: torch.Tensor) -> torch.Tensor:
    r"""Montgomery's pair correlation R₂(u) = 1 − (sin πu / πu)² (GUE)."""
    return 1 - torch.sinc(u) ** 2  # torch.sinc(u) = sin(πu)/(πu)


def wigner_gue(s: torch.Tensor) -> torch.Tensor:
    r"""Wigner surmise for GUE spacings p(s) = (32/π²) s² e^{−4s²/π} (within ~1% of the exact Gaudin law)."""
    return 32 / math.pi**2 * s**2 * torch.exp(-4 * s**2 / math.pi)


def zero_waves(x: torch.Tensor, gammas: torch.Tensor) -> torch.Tensor:
    r"""2 Re(x^ρ/ρ) for every zero ρ_k = ½ + iγ_k and every x.   (X,), (K,) → (K, X)"""
    rho = 0.5 + 1j * gammas.to(torch.float64)
    return 2 * (torch.exp(rho[:, None] * torch.log(x)[None, :]) / rho[:, None]).real


def psi_smooth(x: torch.Tensor) -> torch.Tensor:
    r"""The zero-free part of the explicit formula: x − ln 2π − ½ ln(1 − x^{−2})."""
    return x - math.log(2 * math.pi) - 0.5 * torch.log(1 - x**-2)
```

## 3. `tensors.py` — PyTorch tensors

```python
%%writefile problems/maths/riemann_hypothesis/tensors.py
"""Build the concrete tensors: the height grid, Gram points and the prime-power staircase ψ(x)."""

import math

import torch

from .equation import theta

# Published zeros (Odlyzko's tables, reproduced with mpmath.zetazero) used only to check the computation.
KNOWN_ZEROS = {1: 14.134725142, 2: 21.022039639, 3: 25.010857580, 100: 236.524229666, 1000: 1419.422480946}


def t_grid(t_max: float, step: float, device: torch.device) -> torch.Tensor:
    """Heights on the critical line from 10 (below the first zero) to t_max: (T,)."""
    return torch.arange(10.0, t_max, step, dtype=torch.float64, device=device)


def gram_points(n_max: int, device: torch.device) -> torch.Tensor:
    """Gram points g_n, θ(g_n) = nπ for n = 0 … n_max, by Newton's method on all of them at once: (n_max + 1,).
    θ′(t) ≈ ½ ln(t/2π), and θ is increasing and convex for t > 7, so Newton converges from any start there."""
    n = torch.arange(n_max + 1, dtype=torch.float64, device=device)
    g = 20 + 2 * math.pi * n / torch.log(n + 3)  # rough start
    for _ in range(40):
        g = g - (theta(g) - n * math.pi) / (0.5 * torch.log(g / (2 * math.pi)))
    return g


def chebyshev_psi(x_max: int, device: torch.device) -> torch.Tensor:
    """ψ(n) = Σ_{p^k ≤ n} ln p for n = 0 … x_max: von Mangoldt's Λ placed on the prime powers, then a cumulative
    sum.   (x_max + 1,)"""
    is_prime = torch.ones(x_max + 1, dtype=torch.bool, device=device)
    is_prime[:2] = False
    for p in range(2, math.isqrt(x_max) + 1):
        if is_prime[p]:
            is_prime[p * p :: p] = False
    primes = torch.nonzero(is_prime).flatten()  # (P,)
    k = torch.arange(1, int(math.log2(max(x_max, 2))) + 1, device=device)  # exponents 1 … log₂ x
    powers = primes[:, None].double() ** k[None, :]  # (P, K) every prime power p^k
    keep = powers <= x_max
    lam = torch.zeros(x_max + 1, dtype=torch.float64, device=device)
    lam[powers[keep].long()] = torch.log(primes[:, None].double().expand_as(powers))[keep]  # Λ(p^k) = ln p
    return torch.cumsum(lam, dim=0)
```

## 4. `solve.py` — solver

```python
%%writefile problems/maths/riemann_hypothesis/solve.py
"""Find the zeros of Z(t), count them, unfold them and feed them into the explicit formula."""

import torch

from problems._common.progress import steps

from .equation import psi_smooth, riemann_siegel_z, smooth_count, zero_waves


def z_on_grid(t: torch.Tensor, chunk: int = 20000) -> torch.Tensor:
    """Z(t) on a long grid, in chunks so the (T, N) term matrix stays small."""
    return torch.cat([riemann_siegel_z(t[i : i + chunk]) for i in range(0, len(t), chunk)])


def find_zeros(t: torch.Tensor, z: torch.Tensor, n_bisect: int = 32) -> torch.Tensor:
    """Sign changes of Z bracket the zeros on the critical line; all brackets are bisected at once.   → (K,)"""
    change = (z[:-1] * z[1:]) < 0
    lo, hi, z_lo = t[:-1][change], t[1:][change], z[:-1][change]
    for _ in steps(n_bisect, "bisection"):
        mid = 0.5 * (lo + hi)
        z_mid = riemann_siegel_z(mid)
        same = torch.sign(z_mid) == torch.sign(z_lo)
        lo, z_lo, hi = torch.where(same, mid, lo), torch.where(same, z_mid, z_lo), torch.where(same, hi, mid)
    return 0.5 * (lo + hi)


def count_below(gammas: torch.Tensor, t: torch.Tensor) -> torch.Tensor:
    """N(t) = number of zeros with 0 < γ ≤ t (the first zero is at 14.13, so the grid from 10 misses none)."""
    return torch.searchsorted(gammas, t.contiguous(), right=True)


def s_function(gammas: torch.Tensor, t: torch.Tensor) -> torch.Tensor:
    """S(t) = N(t) − θ(t)/π − 1: the fluctuating part of the zero count.   (T,)"""
    return count_below(gammas, t) - smooth_count(t)


def gram_violations(z_at_gram: torch.Tensor) -> torch.Tensor:
    """Indices n where Gram's law (−1)^n Z(g_n) > 0 fails ("bad" Gram points)."""
    n = torch.arange(len(z_at_gram), device=z_at_gram.device)
    sign = 1 - 2 * (n % 2)
    return torch.nonzero(sign * z_at_gram <= 0).flatten()


def unfold(gammas: torch.Tensor) -> torch.Tensor:
    """x_k = θ(γ_k)/π + 1: zeros rescaled to mean spacing 1 (the local density is ln(t/2π)/2π)."""
    return smooth_count(gammas)


def pair_differences(x: torch.Tensor, neighbours: int = 10) -> torch.Tensor:
    """x_{k+j} − x_k for j = 1 … J and every k: the pair correlation needs only nearby pairs.   (K − J, J)"""
    k = len(x) - neighbours
    j = torch.arange(1, neighbours + 1, device=x.device)
    idx = torch.arange(k, device=x.device)[:, None] + j[None, :]  # (K − J, J)
    return x[idx] - x[:k, None]


def pair_correlation(x: torch.Tensor, edges: torch.Tensor, neighbours: int = 10) -> torch.Tensor:
    """Density of differences u = x_j − x_k (j > k) per zero and per unit u, on the bins `edges`: R₂(u)."""
    d = pair_differences(x, neighbours).flatten()
    counts = torch.histc(d, bins=len(edges) - 1, min=edges[0].item(), max=edges[-1].item())
    return counts / ((len(x) - neighbours) * (edges[1] - edges[0]))


def psi_from_zeros(x: torch.Tensor, gammas: torch.Tensor) -> torch.Tensor:
    """ψ_K(x) = x − Σ_{k ≤ K} 2 Re(x^{ρ_k}/ρ_k) − ln 2π − ½ ln(1 − x^{−2}) for every K = 0 … len(gammas).
    Returns (K + 1, X); row K uses the first K zeros."""
    smooth = psi_smooth(x)  # (X,)
    waves = torch.cumsum(zero_waves(x, gammas), dim=0)  # (K, X)
    return torch.cat([smooth[None], smooth[None] - waves])
```

## 5. `plot.py` — 2D/3D plot → GIF

```python
%%writefile problems/maths/riemann_hypothesis/plot.py
"""Six panels that grow with the height T: Z(t) near T, S(t), spacings, pair correlation, ψ(x) and the strip."""

from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

from problems._common.plotting import ACCENT, frame_indices, save_gif

DVP = 5.558691  # zero-free region σ > 1 − 1/(5.558691 ln t), proven (Mossinghoff–Trudgian 2015)


def _pair_density(x: np.ndarray, edges: np.ndarray, neighbours: int = 10) -> np.ndarray:
    d = np.concatenate([x[j:] - x[:-j] for j in range(1, neighbours + 1)])
    counts, _ = np.histogram(d, bins=edges)
    return counts / (max(len(x) - neighbours, 1) * (edges[1] - edges[0]))


def animate(t, z, gammas, gram, bad, s, t_s, spacing, x, edges, xs, exact, psi_k, heights, path: Path) -> Path:
    t, z, g = t.cpu().numpy(), z.cpu().numpy(), gammas.cpu().numpy()
    gram, bad = gram.cpu().numpy(), bad.cpu().numpy()
    s, t_s, spacing, x = s.cpu().numpy(), t_s.cpu().numpy(), spacing.cpu().numpy(), x.cpu().numpy()
    edges, xs, exact, psi_k = edges.cpu().numpy(), xs.cpu().numpy(), exact.cpu().numpy(), psi_k.cpu().numpy()
    mid = 0.5 * (edges[1:] + edges[:-1])
    good = np.ones(len(gram), bool)
    good[bad] = False

    fig, axes = plt.subplots(2, 3, figsize=(12, 6.6))
    (ax_z, ax_s, ax_h), (ax_r, ax_p, ax_c) = axes

    (z_line,) = ax_z.plot([], [], color="#374151", lw=1)
    ax_z.axhline(0, color="#9ca3af", lw=0.8)
    zeros_dot = ax_z.scatter([], [], s=18, color="#2563eb", zorder=3, label="zeros")
    gram_ok = ax_z.scatter([], [], s=22, marker="|", color="#16a34a", label="Gram point, law holds")
    gram_bad = ax_z.scatter([], [], s=40, marker="x", color="#dc2626", label="Gram's law fails")
    ax_z.set_ylim(-6, 6)
    ax_z.set_xlabel("$t$")
    ax_z.legend(fontsize=7, loc="lower left", ncol=1)
    title_z = ax_z.set_title("", fontsize=9)

    (s_line,) = ax_s.plot([], [], color=ACCENT[0], lw=0.6)
    for level in (-1, 1):
        ax_s.axhline(level, color="#9ca3af", ls="--", lw=0.8)
    ax_s.set_xscale("log")
    ax_s.set_xlim(10, t[-1])
    ax_s.set_ylim(-2.2, 2.2)
    ax_s.set_xlabel("$t$")
    ax_s.set_title(r"$S(t) = N(t) - \theta(t)/\pi - 1$ (Riemann–von Mangoldt)", fontsize=9)

    grid = np.linspace(0, 3, 300)
    bars_h = ax_h.stairs(np.zeros_like(mid), edges, fill=True, color="#93c5fd", label="zeta zeros")
    ax_h.plot(grid, 32 / np.pi**2 * grid**2 * np.exp(-4 * grid**2 / np.pi), color=ACCENT[1], lw=1.5, label="GUE")
    ax_h.plot(grid, np.exp(-grid), color="#6b7280", lw=1.2, ls="--", label="Poisson")
    ax_h.set_xlim(0, 3)
    ax_h.set_ylim(0, 1.15)
    ax_h.set_xlabel("unfolded spacing $s$")
    ax_h.legend(fontsize=7, loc="upper right")
    title_h = ax_h.set_title("", fontsize=9)

    bars_r = ax_r.stairs(np.zeros_like(mid), edges, fill=True, color="#93c5fd", label="zeta zeros")
    ax_r.plot(grid, 1 - np.sinc(grid) ** 2, color=ACCENT[1], lw=1.5, label=r"$1 - (\sin \pi u / \pi u)^2$")
    ax_r.axhline(1, color="#6b7280", lw=1.2, ls="--", label="no correlation")
    ax_r.set_xlim(0, 3)
    ax_r.set_ylim(0, 1.4)
    ax_r.set_xlabel("difference $u$ of unfolded zeros")
    ax_r.legend(fontsize=7, loc="lower right")
    ax_r.set_title("pair correlation (Montgomery's conjecture)", fontsize=9)

    ax_p.step(xs, exact - xs, where="post", color="#374151", lw=1.2, label=r"$\psi(x) - x$ (sieve)")
    (p_line,) = ax_p.plot(xs, psi_k[0] - xs, color=ACCENT[0], lw=1.1, label="explicit formula")
    ax_p.set_ylim((exact - xs).min() - 1.5, (exact - xs).max() + 1.5)
    ax_p.set_xlabel("$x$")
    ax_p.legend(fontsize=7, loc="upper left")
    title_p = ax_p.set_title("", fontsize=9)

    t_top = t[-1]
    tt = np.geomspace(10, t_top, 400)
    ax_c.fill_betweenx(tt, 1 - 1 / (DVP * np.log(tt)), 1.05, color="#bbf7d0", label="zero-free (proven)")
    ax_c.fill_betweenx(tt, -0.05, 1 / (DVP * np.log(tt)), color="#bbf7d0")
    ax_c.axvspan(7 / 8, 1, color="#fed7aa", alpha=0.6, lw=0, label=r"claimed 2026, unverified: $\sigma > 7/8$")
    ax_c.axvspan(0, 1 / 8, color="#fed7aa", alpha=0.6, lw=0)
    ax_c.axvline(0.5, color="#9ca3af", ls="--", lw=1)
    zeros_c = ax_c.scatter([], [], s=3, color="#2563eb", label="zeros found on the line")
    ax_c.set_xlim(-0.05, 1.05)
    ax_c.set_yscale("log")
    ax_c.set_ylim(10, t_top)
    ax_c.set_xlabel(r"$\sigma = \mathrm{Re}\, s$")
    ax_c.set_ylabel(r"$t = \mathrm{Im}\, s$")
    ax_c.legend(fontsize=6.5, loc="upper center", framealpha=0.95)
    ax_c.set_title("critical strip: RH says every zero has σ = 1/2", fontsize=9)

    def update(f: int):
        top = heights[f]
        n = int(np.searchsorted(g, top))
        window = (t > top - 8) & (t <= top)
        z_line.set_data(t[window], z[window])
        ax_z.set_xlim(top - 8, top)
        in_w = (g > top - 8) & (g <= top)
        zeros_dot.set_offsets(np.c_[g[in_w], np.zeros(in_w.sum())])
        gw = (gram > top - 8) & (gram <= top)
        gram_ok.set_offsets(np.c_[gram[gw & good], np.zeros((gw & good).sum())])
        gram_bad.set_offsets(np.c_[gram[gw & ~good], np.zeros((gw & ~good).sum())])
        title_z.set_text(f"Z(t) up to T = {top:.0f}: {n} zeros")
        keep = t_s <= top
        s_line.set_data(t_s[keep], s[keep])
        if n > 12:
            bars_h.set_data(np.histogram(spacing[: n - 1], bins=edges, density=True)[0])
            bars_r.set_data(_pair_density(x[:n], edges))
        title_h.set_text(f"nearest-neighbour spacings ({max(n - 1, 0)} gaps)")
        p_line.set_ydata(psi_k[n] - xs)
        title_p.set_text(f"ψ(x) − x rebuilt from the {n} zeros below T")
        zeros_c.set_offsets(np.c_[np.full(n, 0.5), g[:n]])

    fig.tight_layout()
    return save_gif(fig, update, frame_indices(len(heights), len(heights)), path, fps=6)
```
