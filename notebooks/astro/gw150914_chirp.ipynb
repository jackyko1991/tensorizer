# GW150914: the first gravitational-wave chirp

*astro · advanced* · [open in Tensorizer](https://jackyko1991.github.io/tensorizer/#/astro/gw150914_chirp)

Sixteen seconds of real LIGO strain from Hanford and Livingston, 14 September 2015. A Welch PSD, whitening and a band-pass bring out a chirp buried under noise a hundred times louder. A Q-transform map shows its frequency sweeping up to about 150 Hz, and a template bank of post-Newtonian inspirals, filtered in one batch, peaks at a chirp mass near $30\,M_\odot$. The ridge is fitted by autograd with $\dot f \propto M_c^{5/3} f^{11/3}$, and the Livingston signal leads Hanford's by about 7 ms.

**Vector form**

$$\frac{df}{dt} = \frac{96}{5}\,\pi^{8/3}\Big(\frac{G\mathcal M_c}{c^3}\Big)^{5/3} f^{11/3},\qquad \rho(t) = \Big|\,4\int_0^\infty \frac{\tilde d(f)\,\tilde h^*(f)}{S_n(f)}\,e^{2\pi i f t}\,df\,\Big|,\qquad \langle h|h\rangle = 1$$

**Index (tensor) form**

$$S_{dk} = \frac{2}{f_s\,w_nw_n}\,\big\langle |X_{dsk}|^2\big\rangle_s,\qquad \tilde h_{mk} = f_k^{-7/6}\,e^{-i(\Psi_{mk} - \pi/4)},\qquad z_{dmt} = 4\,\Delta f\sum_k \frac{\tilde d_{dk}\,\tilde h^*_{mk}}{S_{dk}}\,e^{2\pi i f_k t},\qquad \rho_{dm} = \max_t |z_{dmt}|$$

| Symbol | Code | Shape | Meaning |
|---|---|---|---|
| $x_{dn}$ | `load_strain()['x']` | (2, N) | strain of detector $d \in \{H1, L1\}$ at sample $n$, 16 s at 2048 Hz |
| $S_{dk}$ | `welch_psd(x, fs, L)` | (2, F) | one-sided PSD: Hann-windowed segments $s$, FFT, $/X/^2$ averaged over $s$ |
| $\tilde x_{dk}/\sqrt{S_{dk} f_s/2}$ | `whiten(x, psd, fs)` | (2, N) | whitened strain: unit-variance white noise where the PSD is right |
| $E_{qt}$ | `qtransform(white, fs, freqs, q)` | (Q, N) | energy of Gaussian band-pass filters of width $f_q/q$, one batch of inverse FFTs |
| $\Psi_{mk}$ | `spa_phase(f, mc, pn_order=2)` | (M, F) | stationary-phase inspiral phase up to 2PN for chirp mass $\mathcal M_m$, $\eta = 1/4$ |
| $\tilde h_{mk}$ | `template_bank(f, mc, psd)` | (M, F) | the whole bank at once, cut at the merger peak frequency, normalised so $\langle h/h\rangle = 1$ |
| $\langle a\vert b\rangle$ | `inner(a, b, psd, df)` | (…,) | $4\,\mathrm{Re}\sum_k a_k b_k^*/S_k\,\Delta f$, $k$ summed |
| $z_{dmt}$ | `snr_series(d, h, psd, df, N)` | (2, M, N) | $k$ summed by an inverse FFT: complex SNR for every detector, template and time |
| $f(\tau)$ | `chirp_frequency(tau, mc)` | (T,) | Newtonian track $\frac1\pi(5/256\tau)^{3/8}\mathcal M_c^{-5/8}$, fitted to the ridge by Adam |

On 14 September 2015 at 09:50:45 UTC both Advanced LIGO detectors recorded the merger of two black holes about 410 Mpc away (Abbott et al., PRL 116, 061102, 2016), the first direct detection of gravitational waves. This page repeats a simplified version of the analysis on the open data. Raw strain is dominated by seismic and thermal noise below 30 Hz, so the PSD $S_n(f)$ is estimated with Welch's method (4 s Hann segments, 50 % overlap) and the data are divided by $\sqrt{S_n}$ (whitening); the page checks that the whitened PSD is flat, Parseval's theorem and that every template has $\langle h|h\rangle = 1$. The template bank uses the stationary-phase inspiral with a 2PN phase for equal masses, cut at $f_\mathrm{peak} = 0.36/(2\pi M)$, the frequency where an equal-mass merger peaks in numerical relativity (or at the ISCO frequency, which loses most of the SNR). All templates and both detectors are filtered in one batch of FFTs. With the defaults the best template has $\mathcal M_c \approx 32\,M_\odot$, SNR about 16 (H1) and 10 (L1), and the two SNR peaks and the H1–(−L1) cross-correlation put Hanford 7–8 ms after Livingston (published: 6.9 +0.5/−0.4 ms; the L1 arm orientation flips the sign). A Newtonian fit to the Q-map ridge gives a heavier $\mathcal M_c \approx 37\,M_\odot$, and so does the 0PN bank: post-Newtonian terms slow the chirp at a given chirp mass, so a Newtonian model reads it as heavier. HONESTY: this is a teaching analysis, not the published one. The published values come from full inspiral–merger–ringdown waveforms with spins, a PSD from much longer data and Bayesian parameter estimation: detector-frame $\mathcal M_c \approx 30\,M_\odot$ and network SNR 24 (Abbott et al. 2016), source-frame $\mathcal M_c = 28.6^{+1.7}_{-1.5}\,M_\odot$, masses $35.6$ and $30.6\,M_\odot$, distance 440 Mpc in GWTC-1. Our SNRs are lower because the templates have no merger–ringdown and no spins, and because the PSD comes from only 16 s that include the signal. Data: GWOSC (Gravitational Wave Open Science Center), GW150914 release GWTC-1-confident v3, files H-H1_GWOSC_4KHZ_R1-1126259447-32 and the L1 equivalent, from gwosc.org/eventapi/json/GWTC-1-confident/GW150914/v3/, CC BY 4.0; 16 s from GPS 1126259455, low-passed at 900 Hz and decimated to 2048 Hz, stored as strain × 10⁻²¹ in data/gw150914_strain.csv. Please cite GWOSC (Abbott et al., SoftwareX 13, 100658, 2021) and Abbott et al. 2016. Prizes: Shaw Prize in Astronomy 2016 (Drever, Thorne and Weiss, for conceiving and designing LIGO), Special Breakthrough Prize in Fundamental Physics 2016 (Drever, Thorne, Weiss and the LIGO contributors) and Nobel Prize in Physics 2017 (Weiss, Barish and Thorne).

```python
%%writefile problems/astro/gw150914_chirp/main.py
"""Main loop: LIGO strain → Welch PSD → whitening → Q-map and ridge fit → batched template bank → M_c, SNR, delay."""

from pathlib import Path

import torch

from problems._common.runtime import cli

from . import plot, solve, tensors
from .equation import inner


def run(params: dict, out_dir: Path) -> Path:
    # CPU, float64: strains are ~1e-21 and PSDs ~1e-46, and the template phases reach thousands of radians.
    strain = tensors.load_strain()
    x, fs = strain["x"], strain["fs"]
    n = x.shape[-1]
    print(f"GW150914: {n / fs:.0f} s of H1 and L1 strain at {fs:.0f} Hz from GPS {strain['gps_start']:.0f} (GWOSC)")

    res = solve.analyse(
        strain,
        pn_order=params["pn_order"],
        f_low=params["f_low"],
        cutoff=params["cutoff"],
        n_templates=params["n_templates"],
        mc_lo=params["mc_min"],
        mc_hi=params["mc_max"],
        q=params["q"],
        seg=params["welch_segment"],
    )

    # --- checks on the real data ---
    big_x = torch.fft.fft(x[0])
    print(f"Parseval on H1: Σ|x|² / (Σ|X|²/N) = {(x[0].pow(2).sum() / (big_x.abs().pow(2).sum() / n)).item():.12f}")
    f100 = int(100 * n / fs)
    asd = res["psd"][:, f100].sqrt()
    print(f"Welch ASD at 100 Hz: H1 {asd[0]:.2e}, L1 {asd[1]:.2e} strain/√Hz (design-era O1 ≈ 1e-23)")
    w = res["white_bp"]
    psd_w, fw = solve.welch_psd(solve.whiten(x, res["psd"], fs)[:, 1024:-1024], fs, int(fs))
    band = (fw > 30) & (fw < 400)
    flat = (psd_w[:, band].std(-1) / psd_w[:, band].mean(-1)).tolist()
    print(f"whitened PSD 30–400 Hz: relative scatter H1 {flat[0]:.2f}, L1 {flat[1]:.2f} (flat ⇒ only Welch noise)")
    f = res["f"]
    h = solve.template_bank(f, res["bank"], res["psd"][0], params["pn_order"], params["f_low"], params["cutoff"])
    norms = inner(h, h, torch.where(res["psd"][0] > 0, res["psd"][0], torch.inf), f[1] - f[0])
    print(f"template bank: {len(h)} templates, ⟨h|h⟩ between {norms.min():.12f} and {norms.max():.12f}")

    # --- the detection ---
    k = res["k"]
    print(
        f"\nbest template: M_c = {res['mc_bank']:.1f} M☉ (detector frame, PN order {params['pn_order']}, "
        f"cut at {params['cutoff']}), SNR H1 {res['snr'][0]:.1f}, L1 {res['snr'][1]:.1f}, "
        f"network {res['network'][k]:.1f}"
    )
    print(f"merger time: GPS {strain['gps_start'] + res['t_event']:.3f} (GWOSC: {tensors.GPS_EVENT})")
    print(
        f"H1 − L1 delay: {res['dt_ms']:.1f} ms from the SNR peaks, {res['dt_cc_ms']:.1f} ms from the "
        f"cross-correlation of H1 with −L1 (published 6.9 +0.5/−0.4 ms; light travel time ≤ 10 ms)"
    )
    print(
        f"ridge fit (Newtonian df/dt ∝ M_c^(5/3) f^(11/3), {int(res['kept'].sum())} Q-tiles): "
        f"M_c = {res['mc_track']:.1f} M☉"
    )
    print(
        "published: detector-frame M_c ≈ 30 M☉ and network SNR 24 (Abbott et al. 2016); GWTC-1: source-frame "
        "M_c = 28.6 (+1.7/−1.5) M☉ at z ≈ 0.09, i.e. (1+z)M_c ≈ 31 M☉"
    )
    print(f"whitened strain peak |h| ≈ {w[0].abs().max():.1f} σ in H1")
    return plot.animate(res, out_dir / "gw150914_chirp.gif")


if __name__ == "__main__":
    cli(run, __file__)
```

## 2. `equation.py` — physics → tensor translation

```python
%%writefile problems/astro/gw150914_chirp/equation.py
r"""Physics → tensor translation for a compact-binary chirp and its matched filter.

Leading (Newtonian, quadrupole) order: two masses of chirp mass M_c = (m_1 m_2)^{3/5}/(m_1 + m_2)^{1/5} spiral in,
and the gravitational-wave frequency f (twice the orbital frequency) rises as

    df/dt = (96/5) π^{8/3} M_c^{5/3} f^{11/3}         ⇒   f(τ) = (1/π) (5/256 τ)^{3/8} M_c^{−5/8}

with τ = t_c − t the time left to coalescence and M_c in seconds (G M_c / c³; 1 M☉ = 4.925 μs).

Stationary-phase (SPA) frequency-domain template, post-Newtonian phase up to 2PN (η = m_1 m_2/M², v = (π M f)^{1/3}):

    h̃(f) = f^{−7/6} exp(−i[2π f t_c − π/4 + Ψ(f)]),
    Ψ(f) = 3/(128 η v⁵) [1 + (3715/756 + 55η/9) v² − 16π v³ + (15293365/508032 + 27145η/504 + 3085η²/72) v⁴]

Noise-weighted inner product (one-sided PSD S_n(f)) and the matched-filter SNR time series:

    ⟨a|b⟩ = 4 Re Σ_k ã(f_k) b̃*(f_k) / S_n(f_k) Δf,        z(t) = 4 Σ_k d̃(f_k) h̃*(f_k) / S_n(f_k) e^{2π i f_k t} Δf

with ⟨h|h⟩ = 1, so ρ(t) = |z(t)| is the SNR maximised over the template's phase. Index k (frequency) is summed.
"""

import math

import torch

G = 6.67430e-11  # m³ kg⁻¹ s⁻²
C = 299_792_458.0  # m/s
M_SUN = 1.98847e30  # kg
T_SUN = G * M_SUN / C**3  # 4.925 μs: one solar mass expressed as a time


def chirp_frequency(tau: torch.Tensor, mc: torch.Tensor) -> torch.Tensor:
    r"""f(τ) = (1/π)(5/(256 τ))^{3/8} M_c^{−5/8}   τ (…,) seconds to merger, mc in seconds (broadcast) → Hz."""
    return (5.0 / (256.0 * tau)) ** 0.375 * mc ** (-0.625) / math.pi


def chirp_rate(f: torch.Tensor, mc: torch.Tensor) -> torch.Tensor:
    r"""df/dt = (96/5) π^{8/3} M_c^{5/3} f^{11/3}   (Hz/s)."""
    return 96.0 / 5.0 * math.pi ** (8 / 3) * mc ** (5 / 3) * f ** (11 / 3)


def time_to_coalescence(f: torch.Tensor, mc) -> torch.Tensor:
    r"""τ(f) = (5/256) (π f)^{−8/3} M_c^{−5/3}, the inverse of chirp_frequency."""
    return 5.0 / 256.0 * (math.pi * f) ** (-8 / 3) * mc ** (-5 / 3)


def spa_phase(f: torch.Tensor, mc: torch.Tensor, eta: float = 0.25, pn_order: int = 2) -> torch.Tensor:
    r"""Ψ(f) up to pn_order 0 (Newtonian), 1 (1PN) or 2 (1.5PN tail + 2PN): f (F,), mc (K, 1) seconds → (K, F)."""
    total = mc * eta ** (-0.6)  # M = M_c η^{−3/5}
    v = (math.pi * total * f) ** (1 / 3)
    terms = torch.ones_like(v)
    if pn_order >= 1:
        terms = terms + (3715 / 756 + 55 * eta / 9) * v**2
    if pn_order >= 2:
        terms = terms - 16 * math.pi * v**3 + (15293365 / 508032 + 27145 * eta / 504 + 3085 * eta**2 / 72) * v**4
    return 3.0 / (128.0 * eta * v**5) * terms


def peak_frequency(mc: torch.Tensor, eta: float = 0.25) -> torch.Tensor:
    r"""f_peak = 0.36 / (2π M): GW frequency where an equal-mass, non-spinning merger peaks (M ω₂₂ ≈ 0.36, NR)."""
    return 0.36 / (2 * math.pi * mc * eta ** (-0.6))


def isco_frequency(mc: torch.Tensor, eta: float = 0.25) -> torch.Tensor:
    r"""f_ISCO = 1 / (6^{3/2} π M): twice the orbital frequency at the innermost stable circular orbit."""
    return 1.0 / (6**1.5 * math.pi * mc * eta ** (-0.6))


def spa_template(f: torch.Tensor, mc: torch.Tensor, f_low: float, f_high: torch.Tensor, pn_order: int = 2):
    r"""h̃(f) = f^{−7/6} e^{−i(Ψ − π/4)} for f_low ≤ f ≤ f_high (t_c = 0).   f (F,), mc and f_high (K, 1) → (K, F)"""
    fs = f.clamp(min=1.0)  # avoid f = 0 (masked anyway)
    band = (f >= f_low) & (f <= f_high)
    phase = spa_phase(fs, mc, pn_order=pn_order) - math.pi / 4
    return torch.where(band, fs ** (-7 / 6) * torch.exp(-1j * phase), torch.zeros((), dtype=torch.complex128))


def inner(a: torch.Tensor, b: torch.Tensor, psd: torch.Tensor, df) -> torch.Tensor:
    r"""⟨a|b⟩ = 4 Re Σ_k a_k b*_k / S_k Δf — k summed.   (…, F) → (…,)"""
    return 4.0 * df * (a * b.conj() / psd).real.sum(-1)


def snr_series(data_f: torch.Tensor, h: torch.Tensor, psd: torch.Tensor, df, n: int) -> torch.Tensor:
    r"""z(t) = 4 Σ_k d̃_k h̃*_k / S_k e^{2πi f_k t} Δf for every template.   d̃, S (D, F), h̃ (K, F) → (D, K, n)

    Only positive frequencies enter, so the inverse FFT returns a complex series whose modulus is the SNR
    maximised over the template phase (the real and imaginary parts are the cos and sin templates).
    """
    integrand = 4.0 * df * data_f[:, None, :] * h.conj()[None] / psd[:, None, :]  # (D, K, F)
    return torch.fft.ifft(integrand, n=n, dim=-1) * n  # ifft has a 1/n; Σ_k e^{2πi k j/n}
```

## 3. `tensors.py` — PyTorch tensors

```python
%%writefile problems/astro/gw150914_chirp/tensors.py
"""Build the concrete tensors: the LIGO strain, frequency grids, the chirp-mass bank and synthetic coloured noise.

data/gw150914_strain.csv holds 16 s of strain around GW150914 from both LIGO detectors (H1 Hanford, L1
Livingston), from the GWOSC release (GWTC-1-confident, v3, 4 kHz files, CC BY 4.0), low-passed at 900 Hz and
decimated to 2048 samples per second; values are strain × 10⁻²¹ (header lines start with #). Everything is float64.
"""

from pathlib import Path

import torch

DATA = Path(__file__).parent / "data" / "gw150914_strain.csv"
DTYPE = torch.float64
DETECTORS = ("H1", "L1")
GPS_EVENT = 1126259462.4  # GWOSC event time of GW150914


def load_strain(path: Path = DATA) -> dict:
    """Strain of both detectors: {'x': (2, N) strain, 'fs': Hz, 'gps_start': s, 't': (N,) s from the file start}."""
    header, rows = {}, []
    for line in Path(path).read_text(encoding="utf-8").splitlines():
        if line.startswith("#"):
            if "=" in line:
                key, value = line[1:].split("=")
                header[key.strip()] = float(value)
        elif line and line[0] not in "hH":  # skip the column-name row
            rows.append([float(v) for v in line.split(",")])
    x = torch.tensor(rows, dtype=DTYPE).T * 1e-21  # (2, N)
    fs = header["sample_rate"]
    t = torch.arange(x.shape[1], dtype=DTYPE) / fs
    return {"x": x, "fs": fs, "gps_start": header["gps_start"], "t": t}


def chirp_mass_grid(lo: float, hi: float, n: int) -> torch.Tensor:
    """Template bank: n chirp masses from lo to hi (detector frame, M☉): (K,)."""
    return torch.linspace(lo, hi, n, dtype=DTYPE)


def tukey(n: int, alpha: float = 0.1) -> torch.Tensor:
    """Tukey window: flat in the middle, cosine tapers over a fraction alpha at both ends: (n,)."""
    w = torch.ones(n, dtype=DTYPE)
    r = max(1, int(alpha * n / 2))
    ramp = 0.5 * (1 - torch.cos(torch.pi * torch.arange(r, dtype=DTYPE) / r))
    w[:r], w[-r:] = ramp, ramp.flip(0)
    return w


def model_psd(f: torch.Tensor) -> torch.Tensor:
    """Advanced-LIGO design PSD (zero-detuned high power, Ajith 2011 fit), x = f/215 Hz, frozen below 10 Hz: (F,).

    Only used for synthetic tests; the real analysis estimates the PSD from the data with Welch's method.
    """
    x = f.clamp(min=10.0) / 215.0
    return 1e-49 * (x ** (-4.14) - 5 / x**2 + 111 * (1 - x**2 + x**4 / 2) / (1 + x**2 / 2))


def coloured_noise(psd: torch.Tensor, n: int, fs: float, count: int, gen: torch.Generator) -> torch.Tensor:
    """Gaussian noise with one-sided PSD psd on the rfft grid of n samples: (count, n).

    A DFT coefficient of such noise has E|X_k|² = n fs S(f_k)/2, so each complex coefficient gets independent real
    and imaginary parts of variance n fs S/4 (DC and Nyquist are real).
    """
    sigma = (n * fs * psd / 4).sqrt()
    re = torch.randn(count, len(psd), generator=gen, dtype=DTYPE)
    im = torch.randn(count, len(psd), generator=gen, dtype=DTYPE)
    im[:, 0] = 0.0
    im[:, -1] = 0.0
    spec = sigma * torch.complex(re, im)
    spec[:, 0] *= 2**0.5
    spec[:, -1] *= 2**0.5
    return torch.fft.irfft(spec, n=n)
```

## 4. `solve.py` — solver

```python
%%writefile problems/astro/gw150914_chirp/solve.py
"""The search: Welch PSD, whitening, band-pass, Q-transform map, ridge fit with autograd, batched matched filter."""

import cmath
import math

import torch

from problems._common.progress import steps

from . import tensors
from .equation import (
    T_SUN,
    chirp_frequency,
    inner,
    isco_frequency,
    peak_frequency,
    snr_series,
    spa_template,
)


def welch_psd(x: torch.Tensor, fs: float, seg_len: int) -> tuple[torch.Tensor, torch.Tensor]:
    """One-sided PSD by Welch's method: Hann-windowed segments with 50 % overlap, averaged.   (…, n) → (…, F), (F,)"""
    seg = x.unfold(-1, seg_len, seg_len // 2)  # (…, S, L) overlapping segments
    w = torch.hann_window(seg_len, periodic=True, dtype=x.dtype)
    spec = torch.fft.rfft((seg - seg.mean(-1, keepdim=True)) * w, dim=-1)  # (…, S, F)
    psd = spec.abs().pow(2).mean(-2) * 2.0 / (fs * w.pow(2).sum())  # 2/(fs Σw²): one-sided, density per Hz
    psd[..., 0] /= 2  # DC and Nyquist are not doubled
    psd[..., -1] /= 2
    return psd, torch.fft.rfftfreq(seg_len, 1 / fs).to(x.dtype)


def interpolate(f_new: torch.Tensor, f: torch.Tensor, y: torch.Tensor) -> torch.Tensor:
    """Linear interpolation of y(f) (…, F) onto f_new (F2,): (…, F2)."""
    i = torch.searchsorted(f, f_new).clamp(1, len(f) - 1)
    w = ((f_new - f[i - 1]) / (f[i] - f[i - 1])).clamp(0, 1)
    return y[..., i - 1] + w * (y[..., i] - y[..., i - 1])


def whiten(x: torch.Tensor, psd: torch.Tensor, fs: float, alpha: float = 0.1) -> torch.Tensor:
    """x̃(f)/√(S(f) fs/2): unit-variance white noise where the PSD model is right.   x (…, n), psd (…, n//2+1)."""
    n = x.shape[-1]
    spec = torch.fft.rfft(x * tensors.tukey(n, alpha), dim=-1)
    return torch.fft.irfft(spec / (psd * fs / 2).sqrt(), n=n, dim=-1)


def bandpass(x: torch.Tensor, fs: float, lo: float, hi: float, taper: float = 5.0) -> torch.Tensor:
    """Zero-phase band-pass in the frequency domain with raised-cosine edges `taper` Hz wide.   (…, n) → (…, n)"""
    n = x.shape[-1]
    f = torch.fft.rfftfreq(n, 1 / fs).to(x.dtype)
    rise = ((f - lo + taper) / taper).clamp(0, 1)
    fall = ((hi + taper - f) / taper).clamp(0, 1)
    mask = 0.5 * (1 - torch.cos(torch.pi * rise)) * 0.5 * (1 - torch.cos(torch.pi * fall))
    return torch.fft.irfft(torch.fft.rfft(x, dim=-1) * mask, n=n, dim=-1)


def template_bank(f, mc_msun, psd, pn_order: int = 2, f_low: float = 20.0, cutoff: str = "peak"):
    """Normalised SPA templates for every chirp mass, all at once: f (F,), mc (K,) M☉, psd (F,) → (K, F), ⟨h|h⟩ = 1.

    cutoff 'peak': stop at the frequency where an equal-mass merger peaks (M ω₂₂ ≈ 0.36); 'isco': at f_ISCO.
    """
    mc = mc_msun[:, None] * T_SUN  # (K, 1) seconds
    f_high = peak_frequency(mc) if cutoff == "peak" else isco_frequency(mc)
    h = spa_template(f, mc, f_low, f_high, pn_order)  # (K, F)
    psd_safe = torch.where(psd > 0, psd, torch.full_like(psd, float("inf")))
    norm = inner(h, h, psd_safe, f[1] - f[0]).sqrt()  # (K,)
    return h / norm[:, None]


def injection(f, mc_msun: float, t_c: float, snr: float, psd, n: int, fs: float, pn_order: int = 2) -> torch.Tensor:
    """Time series of a template with ⟨s|s⟩^½ = snr that merges at t_c: (n,)."""
    h = template_bank(f, torch.tensor([mc_msun], dtype=f.dtype), psd, pn_order)[0]
    h = snr * h * torch.exp(-2j * math.pi * f * t_c)  # time shift by t_c
    return torch.fft.irfft(h * fs, n=n)  # continuous-FT convention: x̃ = Δt·DFT(x)


def matched_filter(x, mc_msun, psd, fs, pn_order: int = 2, f_low: float = 20.0, cutoff: str = "peak", edge=1.0):
    """Filter every detector against every template in one batch.   x (D, n), psd (D, n//2+1) on the rfft grid.

    Returns |z| (D, K, n), the peak SNR (D, K) and its time (D, K) (refined by a parabola through the
    three highest samples), searched away from the first and last `edge` seconds.
    """
    n = x.shape[-1]
    f = torch.fft.rfftfreq(n, 1 / fs).to(x.dtype)
    df = f[1] - f[0]
    data_f = torch.fft.rfft(x * tensors.tukey(n, 0.1), dim=-1) / fs  # x̃(f) ≈ Δt·DFT
    banks = [template_bank(f, mc_msun, psd[d], pn_order, f_low, cutoff) for d in range(x.shape[0])]
    h = torch.stack(banks)  # (D, K, F): normalised with each detector's own PSD
    psd_safe = torch.where(psd > 0, psd, torch.full_like(psd, float("inf")))
    rho = torch.stack([snr_series(data_f[d : d + 1], h[d], psd_safe[d : d + 1], df, n)[0] for d in range(len(h))])
    rho = rho.abs()  # (D, K, n)
    lo, hi = int(edge * fs), n - int(edge * fs)
    peak, j = rho[..., lo:hi].max(-1)
    j = j + lo
    y0, y1, y2 = (rho.gather(-1, (j + s)[..., None])[..., 0] for s in (-1, 0, 1))
    shift = 0.5 * (y0 - y2) / (y0 - 2 * y1 + y2)  # vertex of the parabola through 3 samples
    return {"rho": rho, "snr_peak": peak, "t_peak": (j + shift) / fs}


def qtransform(white: torch.Tensor, fs: float, freqs: torch.Tensor, q: float = 8.0) -> torch.Tensor:
    """Normalised energy of Gaussian band-pass filters centred on each frequency (a Q-transform): (Fq, n).

    The filter at f₀ has width σ_f = f₀/q, i.e. a time resolution σ_t = q/(2π f₀); every row is divided by its median.
    """
    n = white.shape[-1]
    f = torch.fft.rfftfreq(n, 1 / fs).to(white.dtype)
    spec = torch.fft.rfft(white)  # (F,)
    window = torch.exp(-0.5 * ((f[None] - freqs[:, None]) / (freqs[:, None] / q)) ** 2)  # (Fq, F)
    tiles = torch.fft.ifft(spec[None] * window, n=n, dim=-1)  # (Fq, n) analytic signal of each band
    energy = tiles.abs().pow(2)
    return energy / energy.median(-1, keepdim=True).values


def ridge(energy, t, freqs, t_lo: float, t_hi: float, threshold: float = 6.0):
    """Frequency of the loudest tile in every time column of [t_lo, t_hi] above threshold: (t_k, f_k) points."""
    cols = (t >= t_lo) & (t <= t_hi)
    e = energy[:, cols]  # (Fq, T')
    best, i = e.max(0)
    keep = best > threshold
    return t[cols][keep], freqs[i[keep]]


def fit_track(t_pts, f_pts, tc0: float, mc0: float = 25.0, n_steps: int = 400, lr: float = 0.02, clip: float = 0.3,
              free_tc: bool = True):  # fmt: skip
    """Fit f(t) = f_Newton(t_c − t; M_c) to ridge points with autograd (Adam on ln M_c, and t_c if free_tc).

    With free_tc=False, t_c is held at the matched-filter time (M_c and t_c are degenerate on a short ridge: a
    later t_c with a heavier M_c fits almost as well).

    Minimises the mean of r_k² with r_k = ln f_model − ln f_k, in two rounds: after the first, points with
    |r_k| > clip (stray noise tiles) are dropped. Returns (M_c in M☉, t_c, loss history, mask of points kept).
    """
    log_mc = torch.tensor(math.log(mc0), dtype=t_pts.dtype, requires_grad=True)
    tc = torch.tensor(tc0, dtype=t_pts.dtype, requires_grad=free_tc)
    keep = torch.ones_like(t_pts, dtype=torch.bool)
    history = []
    for _ in range(2):
        groups = [{"params": [log_mc], "lr": lr}] + ([{"params": [tc], "lr": lr * 0.01}] if free_tc else [])
        opt = torch.optim.Adam(groups)
        for _ in steps(n_steps, "ridge fit"):
            opt.zero_grad()
            tau = (tc - t_pts[keep]).clamp(min=1e-4)
            r = chirp_frequency(tau, log_mc.exp() * T_SUN).log() - f_pts[keep].log()
            loss = r.pow(2).mean()
            loss.backward()
            opt.step()
            history.append(loss.item())
        with torch.no_grad():
            tau = (tc - t_pts).clamp(min=1e-4)
            keep = (chirp_frequency(tau, log_mc.exp() * T_SUN).log() - f_pts.log()).abs() < clip
    return log_mc.exp().item(), tc.item(), history, keep


def time_offset(white_bp: torch.Tensor, fs: float, t0: float, half: float = 0.15, max_lag: float = 0.012):
    """H1–L1 delay (s) from the cross-correlation of H1 with −L1 around t0 (L1 is rotated ≈ 180° relative to H1)."""
    a = int((t0 - half) * fs)
    b = int((t0 + half) * fs)
    h, ell = white_bp[0, a:b], -white_bp[1]
    lags = torch.arange(-int(max_lag * fs), int(max_lag * fs) + 1)
    idx = torch.arange(a, b)[None] - lags[:, None]  # (Lags, T): L1 sample that lines up with each H1 sample
    cc = (h[None] * ell[idx]).sum(-1) / (h.norm() * ell[a:b].norm())
    k = int(cc.argmax().clamp(1, len(cc) - 2))
    y0, y1, y2 = cc[k - 1], cc[k], cc[k + 1]
    shift = (0.5 * (y0 - y2) / (y0 - 2 * y1 + y2)).item()
    return (lags[k].item() + shift) / fs, lags.double() / fs, cc


def analyse(strain: dict, pn_order: int = 2, f_low: float = 20.0, cutoff: str = "peak", n_templates: int = 81,
            mc_lo: float = 15.0, mc_hi: float = 55.0, q: float = 5.66, seg: float = 4.0) -> dict:  # fmt: skip
    """The whole pipeline on both detectors; returns every intermediate tensor the plot needs."""
    x, fs = strain["x"], strain["fs"]
    n = x.shape[-1]
    f = torch.fft.rfftfreq(n, 1 / fs).to(x.dtype)
    psd_w, fw = welch_psd(x, fs, int(seg * fs))  # (2, Fw)
    psd = interpolate(f, fw, psd_w)  # (2, F) on the full-length grid
    white = whiten(x, psd, fs)
    white_bp = bandpass(white, fs, 35.0, 350.0)
    bank = tensors.chirp_mass_grid(mc_lo, mc_hi, n_templates)
    mf = matched_filter(x, bank, psd, fs, pn_order, f_low, cutoff)
    network = mf["snr_peak"].pow(2).sum(0).sqrt()  # (K,) combined SNR (peak per detector, quadrature sum)
    k = int(network.argmax())
    t_h, t_l = mf["t_peak"][0, k].item(), mf["t_peak"][1, k].item()
    dt_cc, lags, cc = time_offset(white_bp, fs, t_h)
    freqs = torch.logspace(math.log10(30.0), math.log10(400.0), 60, dtype=x.dtype)
    qmaps = torch.stack([qtransform(white[0], fs, freqs, q), qtransform(-white[1], fs, freqs, q)])  # (2, Fq, n)
    both = qmaps[0] + torch.roll(qmaps[1], round(dt_cc * fs), dims=-1)  # L1 moved onto H1's clock
    t = strain["t"]
    t_pts, f_pts = ridge(both[:, ::8], t[::8], freqs, t_h - 0.15, t_h - 0.01, 8.0)  # one column per 4 ms
    mc_track, tc_track, hist, kept = fit_track(t_pts, f_pts, t_h)
    # the best template in the time domain, whitened like the data, for the overlay
    h_best = template_bank(f, bank[k : k + 1], psd[0], pn_order, f_low, cutoff)[0]
    h_best = h_best * torch.exp(-2j * math.pi * f * t_h)
    phase = _best_phase(x[0], h_best, psd[0], fs)
    h_t = torch.fft.irfft(mf["snr_peak"][0, k] * h_best * cmath.exp(1j * phase) * fs, n=n)
    template_white = bandpass(whiten(h_t[None], psd[:1], fs), fs, 35.0, 350.0)[0]
    return {
        "f": f, "psd": psd, "fw": fw, "psd_w": psd_w, "white_bp": white_bp, "t": t, "bank": bank,
        "snr_bank": mf["snr_peak"], "network": network, "k": k, "mc_bank": bank[k].item(),
        "snr": mf["snr_peak"][:, k].tolist(), "t_event": t_h, "dt_ms": (t_h - t_l) * 1e3, "dt_cc_ms": dt_cc * 1e3,
        "lags": lags, "cc": cc, "rho_best": mf["rho"][:, k], "freqs": freqs, "qmap": qmaps[0], "t_pts": t_pts,
        "kept": kept, "f_pts": f_pts, "mc_track": mc_track, "tc_track": tc_track, "fit_history": hist,
        "template_white": template_white,
    }  # fmt: skip


def _best_phase(x, h, psd, fs) -> float:
    """Phase that maximises Re⟨d|h e^{iφ}⟩ for a template already shifted to the peak time."""
    n = x.shape[-1]
    f = torch.fft.rfftfreq(n, 1 / fs).to(x.dtype)
    d = torch.fft.rfft(x * tensors.tukey(n, 0.1)) / fs
    psd_safe = torch.where(psd > 0, psd, torch.full_like(psd, float("inf")))
    z = 4 * (f[1] - f[0]) * (d * h.conj() / psd_safe).sum()
    return math.atan2(z.imag.item(), z.real.item())
```

## 5. `plot.py` — 2D/3D plot → GIF

```python
%%writefile problems/astro/gw150914_chirp/plot.py
"""Animate the detection: whitened strain revealing the chirp, the Q-map and its fit, the bank scan, the overlay."""

from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import torch
from matplotlib.ticker import NullFormatter

from problems._common.plotting import frame_indices, save_gif

from .equation import T_SUN, chirp_frequency

N_STEPS = 60


def animate(res: dict, path: Path) -> Path:
    t_e = res["t_event"]
    t = res["t"].numpy() - t_e  # seconds relative to the merger
    view = (t > -0.32) & (t < 0.06)
    tv = t[view]
    h1 = res["white_bp"][0].numpy()[view]
    fs = 1.0 / float(res["t"][1] - res["t"][0])
    shift = int(round(res["dt_cc_ms"] * 1e-3 * fs))
    l1 = -np.roll(res["white_bp"][1].numpy(), shift)[view]  # L1 moved onto H1's clock and sign-flipped
    tmpl = res["template_white"].numpy()[view]
    freqs = res["freqs"].numpy()
    cols = np.nonzero(view)[0][::4]
    qmap = res["qmap"].numpy()[:, cols]  # (Fq, T') H1 normalised energy
    tq = t[cols]
    bank, snr = res["bank"].numpy(), res["snr_bank"].numpy()
    net = res["network"].numpy()
    t_pts = res["t_pts"].numpy() - t_e
    f_pts = res["f_pts"].numpy()
    kept = res["kept"].numpy()
    tau = np.linspace(0.004, 0.32, 300)
    f_fit = chirp_frequency(torch.tensor(tau), res["mc_track"] * T_SUN).numpy()
    t_fit = res["tc_track"] - t_e - tau

    fig, axes = plt.subplots(2, 2, figsize=(10, 5.6))
    (ax_s, ax_q), (ax_b, ax_o) = axes
    lim = 1.1 * max(np.abs(h1).max(), np.abs(l1).max())

    (line_h,) = ax_s.plot([], [], color="#2563eb", lw=1.1, label="H1 Hanford")
    (line_l,) = ax_s.plot(
        [], [], color="#dc2626", lw=1.1, alpha=0.8, label=f"L1, shifted {res['dt_cc_ms']:.1f} ms, inverted"
    )
    ax_s.set_xlim(tv[0], tv[-1])
    ax_s.set_ylim(-lim, lim)
    ax_s.set_ylabel("whitened strain (σ)")
    ax_s.legend(loc="upper left", fontsize=7)
    ax_s.set_title("LIGO strain, whitened and band-passed 35–350 Hz", fontsize=9)

    masked = np.ma.masked_all_like(qmap)
    mesh = ax_q.pcolormesh(tq, freqs, masked, cmap="viridis", vmin=0, vmax=min(40, qmap.max()), shading="auto")
    ax_q.set_yscale("log")
    ax_q.set_xlim(tv[0], tv[-1])
    ax_q.set_ylim(freqs[0], freqs[-1])
    ax_q.set_yticks([30, 50, 100, 200, 400])
    ax_q.set_yticklabels(["30", "50", "100", "200", "400"])
    ax_q.yaxis.set_minor_formatter(NullFormatter())
    ax_q.set_ylabel("frequency (Hz)")
    ax_q.grid(False)
    pts = ax_q.scatter([], [], s=10, color="white", edgecolors="#111827", linewidths=0.5, zorder=3)
    (track,) = ax_q.plot([], [], color="#f59e0b", lw=2, zorder=4, label=f"Newtonian fit M_c = {res['mc_track']:.1f} M☉")
    ax_q.legend(loc="upper left", fontsize=7)
    ax_q.set_title("H1 Q-map (normalised energy) and ridge fit", fontsize=9)
    fig.colorbar(mesh, ax=ax_q, pad=0.01).ax.tick_params(labelsize=7)

    ax_b.set_xlim(bank[0], bank[-1])
    ax_b.set_ylim(0, 1.15 * net.max())
    (b_h,) = ax_b.plot([], [], color="#2563eb", lw=1.2, label="H1")
    (b_l,) = ax_b.plot([], [], color="#dc2626", lw=1.2, label="L1")
    (b_n,) = ax_b.plot([], [], color="#111827", lw=2, label="network √(ρ_H² + ρ_L²)")
    ax_b.axvspan(28.0, 32.0, color="#f59e0b", alpha=0.2, lw=0, label="published ≈ 30 M☉ (detector frame)")
    best_mark = ax_b.axvline(np.nan, color="#16a34a", ls="--", lw=1.2)
    ax_b.set_xlabel("template chirp mass M_c (M☉)")
    ax_b.set_ylabel("peak matched-filter SNR")
    ax_b.legend(loc="upper right", fontsize=7)
    title_b = ax_b.set_title("template bank, filtered in one batch", fontsize=9)

    ax_o.plot(tv, h1, color="#9ca3af", lw=1, label="H1 data (whitened)")
    (o_t,) = ax_o.plot([], [], color="#dc2626", lw=1.6, label=f"best template, M_c = {res['mc_bank']:.1f} M☉")
    ax_o.set_xlim(-0.2, 0.05)
    ax_o.set_ylim(-lim, lim)
    ax_o.set_xlabel("time from merger (s)")
    ax_o.set_ylabel("whitened strain (σ)")
    ax_o.legend(loc="upper left", fontsize=7)
    ax_o.set_title(f"template overlay: SNR {res['snr'][0]:.1f} (H1), {res['snr'][1]:.1f} (L1)", fontsize=9)
    for ax in (ax_s, ax_q):
        ax.set_xlabel("time from merger (s)")
    for ax in axes.flat:
        ax.tick_params(labelsize=7)

    def update(step: int):
        frac = step / (N_STEPS - 1)
        cursor = tv[0] + frac * (tv[-1] - tv[0])
        shown = tv <= cursor
        line_h.set_data(tv[shown], h1[shown])
        line_l.set_data(tv[shown], l1[shown])
        mesh.set_array(np.ma.masked_where(np.broadcast_to(tq > cursor, qmap.shape), qmap).ravel())
        seen = t_pts <= cursor
        pts.set_offsets(
            np.column_stack([t_pts[seen & kept], f_pts[seen & kept]]) if (seen & kept).any() else np.empty((0, 2))
        )
        on = t_fit <= cursor
        track.set_data(t_fit[on], f_fit[on])
        m = max(1, int(round(frac * len(bank))))
        b_h.set_data(bank[:m], snr[0, :m])
        b_l.set_data(bank[:m], snr[1, :m])
        b_n.set_data(bank[:m], net[:m])
        j = int(net[:m].argmax())
        best_mark.set_xdata([bank[j], bank[j]])
        title_b.set_text(f"template bank: best so far M_c = {bank[j]:.1f} M☉, network SNR {net[j]:.1f}")
        o_t.set_data(tv[shown], tmpl[shown])

    fig.tight_layout()
    return save_gif(fig, update, frame_indices(N_STEPS, N_STEPS), path, fps=10)
```
