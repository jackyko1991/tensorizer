# The hierarchy problem and gravity leaking into extra dimensions

*cosmology · advanced* · [open in Tensorizer](https://jackyko1991.github.io/tensorizer/#/cosmology/hierarchy_extra_dimensions)

Why is gravity about $10^{33}$ times weaker than the weak force ($G_F M_{Pl}^2$)? One speculative answer: gravity spreads into $n$ hidden compact dimensions. On $\mathbb R^3\times T^n$ the potential is summed two ways, over images of the mass and over the Kaluza–Klein tower, as one tensor over lattice shells. It falls like $1/r^{1+n}$ inside and $1/r$ outside, with a Yukawa tail $\alpha = 2n$, $\lambda = R$. ADD radii for $n = 1..6$ are set against real torsion-balance and LHC limits, and the Randall–Sundrum warp factor $e^{-\pi k r_c}$ turns $M_{Pl}$ into TeV. No extra dimension has been observed.

**Vector form**

$$\nabla^2_{3+n}\phi = (n+1)\,|S^{n+2}|\,G_D M\,\delta^{3+n},\qquad \phi = -\sum_{\mathbf k\in\mathbb Z^n}\frac{G_D M}{(r^2 + |2\pi R\,\mathbf k|^2)^{(n+1)/2}} = -\frac{G_4 M}{r}\sum_{\mathbf k} e^{-|\mathbf k| r/R},\qquad G_4 = \frac{(n+1)|S^{n+2}|\,G_D}{4\pi(2\pi R)^n},\qquad M_{Pl}^2 \sim M_*^{n+2}R^n,\qquad V = -\frac{GMm}{r}\big(1 + \alpha e^{-r/\lambda}\big),\qquad M_{TeV} = M_{Pl}\,e^{-\pi k r_c}$$

**Index (tensor) form**

$$r_n(m) = \big[\vartheta(q)^n\big]_m,\quad \vartheta = \textstyle\sum_j q^{j^2},\qquad \phi^{img}_{a} = -\,r_n(m)\,\big(r_a^2 + m L^2\big)^{-\frac{n+1}{2}} - \text{tail}_a,\qquad \phi^{KK}_{a} = -\frac{G_4}{r_a}\,r_n(m)\,e^{-\sqrt m\,r_a},\qquad \ln(\phi/\phi_N - 1)_a = \ln\alpha - r_a/\lambda,\qquad R_{ns} = \big(M_{Pl}^2/M_{*s}^{n+2}\big)^{1/n}$$

Status: speculative. The hierarchy problem asks why the electroweak scale ($v \approx 246$ GeV) sits so far below the Planck scale ($M_{Pl} \approx 1.22\times10^{19}$ GeV), with $G_F/G_N = G_F M_{Pl}^2 \approx 1.7\times10^{33}$. Arkani-Hamed, Dimopoulos and Dvali (ADD, 1998) proposed that gravity alone spreads into $n$ millimetre-to-femtometre compact dimensions, so the true scale of gravity $M_*$ could be a TeV. Randall and Sundrum (1999) proposed one warped dimension instead. No extra dimension has been found. Other proposed answers include supersymmetry, compositeness, the relaxion and anthropic selection. None is established. What is computed: a point mass on $\mathbb R^3\times T^n$ (circles of radius $R$, units $R = G_D M = 1$). The image sum and the Kaluza–Klein sum are equal by Poisson summation; both are grouped by shells $m = |\mathbf k|^2$ with counts $r_n(m)$. The plot uses the image sum inside ($r < R$, far images replaced by an integral) and the KK sum outside. Gauss's law in $3+n$ space dimensions fixes $G_4 = (n+1)|S^{n+2}|G_D/4\pi(2\pi R)^n$. The lightest KK level has $2n$ modes of mass $1/R$, so just outside the compact space gravity looks like a Yukawa correction with $\alpha = 2n$, $\lambda = R$ (torus; other shapes change $\alpha$). ADD radii use $M_{Pl}^2 = M_*^{n+2}R^n$; papers differ by factors like $(2\pi)^n$ and $8\pi$, so treat $R$ as an order of magnitude. Data, as cited numbers in data/*.csv: Eöt-Wash torsion balances exclude $|\alpha| = 1$ above $\lambda = 56\,\mu$m (Kapner et al., PRL 98, 021101, 2007, arXiv:hep-ph/0611184; one extra dimension $R \le 44\,\mu$m) and above $38.6\,\mu$m (Lee et al., PRL 124, 101101, 2020, arXiv:2002.11761). The exclusion curve itself is not bundled. Because it falls with $\lambda$, every point with $\alpha \ge 1$ and $\lambda > 38.6\,\mu$m is excluded, a conservative test. The ATLAS 13 TeV monojet search (139 fb$^{-1}$; Phys. Rev. D 103, 112006, 2021, arXiv:2102.10874, CC-BY 4.0) excludes $M_D$ below 11.2, 8.5, 7.1, 6.4 and 5.9 TeV for $n = 2..6$ (Giudice–Rattazzi–Wells convention, compared here directly with $M_*$). $n = 1$ would need a dimension the size of the Solar System and is excluded by planetary orbits. Randall–Sundrum: $k r_c \approx 11$–$12$ makes $\bar M_{Pl}e^{-\pi k r_c}$ a TeV; the LHC has not seen its Kaluza–Klein gravitons.

## 2. `equation.py` — physics → tensor translation

```python
%%writefile problems/cosmology/hierarchy_extra_dimensions/equation.py
"""Physics -> tensors: gravity on R^3 x T^n as an image sum and as a Kaluza-Klein sum, ADD and Randall-Sundrum.

Units: compact radius R = 1 (each circle has circumference L = 2 pi), G_D M = 1. A shell index m = |k|^2 groups the
mode vectors k of Z^n; r_n(m) counts them, so a sum over k becomes a sum over m (summed index, shapes in comments).
"""

import math

import torch

from . import tensors

L = 2 * math.pi  # circumference of each compact circle (R = 1)


def sphere_area(k: int) -> float:
    """Area of the unit sphere S^k in R^(k+1): 2 pi^((k+1)/2) / Gamma((k+1)/2)."""
    return 2 * math.pi ** ((k + 1) / 2) / math.gamma((k + 1) / 2)


def g4(n: int) -> float:
    """Effective 4D Newton constant from Gauss's law: G_4 = (n+1) |S^(n+2)| G_D / (4 pi (2 pi R)^n)."""
    return (n + 1) * sphere_area(n + 2) / (4 * math.pi * L**n)


def potential_images(r: torch.Tensor, n: int, counts: torch.Tensor, n_theta: int = 4001) -> torch.Tensor:
    """phi(r) = -sum_k G_D M / (r^2 + |k L|^2)^((n+1)/2): the mass and its images around the circles (r: (R,)).

    Shells m <= m_max are summed exactly; farther images are replaced by their continuum integral
    |S^(n-1)| / L^n * (1/r) * int_{theta_c}^{pi/2} sin^(n-1)(theta) d theta  (y = r tan theta).
    """
    m = torch.arange(len(counts), dtype=r.dtype, device=r.device)  # (M,)
    near = torch.einsum("m,rm->r", counts, (r[:, None] ** 2 + m * L**2) ** (-(n + 1) / 2))  # repeated m => summed
    ball = math.pi ** (n / 2) / math.gamma(n / 2 + 1)  # volume of the unit n-ball
    rho_c = L * (counts.sum().item() / ball) ** (1 / n)  # radius of the ball that holds as many lattice points
    t = torch.linspace(0, 1, n_theta, dtype=r.dtype, device=r.device)
    theta_c = torch.atan(rho_c / r)  # (R,)
    theta = theta_c[:, None] + (math.pi / 2 - theta_c[:, None]) * t  # (R, T)
    tail = torch.trapezoid(torch.sin(theta) ** (n - 1), theta, dim=1)
    return -(near + sphere_area(n - 1) / L**n * tail / r)


def kk_deviation(r: torch.Tensor, n: int, counts: torch.Tensor) -> torch.Tensor:
    """sum_{m >= 1} r_n(m) e^(-sqrt(m) r / R): the massive Kaluza-Klein modes, Yukawa tails of mass sqrt(m)/R (R,)."""
    m = torch.arange(1, len(counts), dtype=r.dtype, device=r.device)
    return torch.einsum("m,rm->r", counts[1:], torch.exp(-torch.sqrt(m) * r[:, None]))


def potential_kk(r: torch.Tensor, n: int, counts: torch.Tensor) -> torch.Tensor:
    """phi(r) = -(G_4 M / r) sum_k e^(-|k| r / R): the zero mode (Newton) plus the KK tower (R,)."""
    return -(g4(n) / r) * (1 + kk_deviation(r, n, counts))


def radial_field(phi, r: torch.Tensor) -> torch.Tensor:
    """d phi / dr by autograd (the pull towards the mass), for any potential function phi(r)."""
    x = r.detach().clone().requires_grad_(True)
    (grad,) = torch.autograd.grad(phi(x).sum(), x)
    return grad


def log_slope(phi, r: torch.Tensor) -> torch.Tensor:
    """Local power law d ln|phi| / d ln r (-(1+n) inside the compact space, -1 outside)."""
    r = r.to(torch.float64)
    return r * radial_field(phi, r) / phi(r)


def yukawa(r, alpha: float, lam: float, g: float = 1.0):
    """V = -(G M / r) (1 + alpha e^(-r / lambda)): the standard parametrisation of short-range deviations."""
    return -(g / r) * (1 + alpha * torch.exp(-r / lam))


def fit_yukawa(r: torch.Tensor, deviation: torch.Tensor) -> tuple[float, float]:
    """Least squares ln(V/V_Newton - 1) = ln alpha - r / lambda -> (alpha, lambda)."""
    design = torch.stack([torch.ones_like(r), -r], 1)  # (R, 2)
    coef = torch.linalg.lstsq(design, torch.log(deviation)[:, None]).solution[:, 0]
    return math.exp(coef[0].item()), 1 / coef[1].item()


def add_radius_m(n: torch.Tensor, m_star_tev: torch.Tensor) -> torch.Tensor:
    """ADD: M_Pl^2 = M*^(n+2) R^n  =>  R = (M_Pl^2 / M*^(n+2))^(1/n), in metres (broadcast over n and M*)."""
    log_r = (2 * math.log(tensors.M_PL_GEV) - (n + 2) * torch.log(m_star_tev * 1e3)) / n  # ln(R * GeV)
    return torch.exp(log_r) * tensors.HBARC_M


def rs_warp(krc: torch.Tensor) -> torch.Tensor:
    """Randall-Sundrum warp factor e^(-k pi r_c): mass M_Pl on one brane looks like
    M_Pl e^(-k pi r_c) on the other."""
    return torch.exp(-math.pi * krc)


def weak_over_gravity() -> float:
    """G_F / G_N = G_F M_Pl^2: how much stronger the weak interaction is than gravity at the same energy."""
    return tensors.G_FERMI * tensors.M_PL_GEV**2


def slice_potential_n1(x: torch.Tensor, y: torch.Tensor) -> torch.Tensor:
    """n = 1, closed form of the image sum -sum_k 1/(x^2 + (y - kL)^2) for L = 2 pi:
    -(sinh|x| / 2|x|) / (cosh x - cos y)."""
    ax = x.abs().clamp_min(1e-9)
    return -torch.sinh(ax) / (2 * ax) / (torch.cosh(ax) - torch.cos(y))


def excluded_by_torsion(alpha: torch.Tensor, lam_m: torch.Tensor, lam_limit_m: float) -> torch.Tensor:
    """Conservative: the measured exclusion curve falls with lambda, so alpha >= 1 and
    lambda > lambda(alpha=1) is out."""
    return (alpha >= 1) & (lam_m > lam_limit_m)
```

## 3. `tensors.py` — PyTorch tensors

```python
%%writefile problems/cosmology/hierarchy_extra_dimensions/tensors.py
"""Concrete tensors: constants, mode vectors of the torus and their shell counts, the measured limits.

Units in the potential: compact radius R = 1 (circumference 2 pi R), G_D M = 1. Particle physics in GeV.
"""

import csv
from pathlib import Path

import torch

DATA = Path(__file__).parent / "data"
M_PL_GEV = 1.220890e19  # Planck mass sqrt(hbar c / G) (CODATA 2018)
M_PL_REDUCED_GEV = 2.435e18  # reduced Planck mass M_Pl / sqrt(8 pi)
HBARC_M = 1.973269804e-16  # hbar c in GeV m: 1 GeV^-1 = 1.97e-16 m
G_FERMI = 1.1663788e-5  # Fermi constant G_F / (hbar c)^3 in GeV^-2 (CODATA 2018)
V_HIGGS = 246.22  # electroweak scale (Higgs vacuum value) in GeV


def mode_vectors(n: int, K: int, device) -> torch.Tensor:
    """(N, n) every Kaluza-Klein mode vector k in the cube [-K, K]^n (N = (2K+1)^n)."""
    axis = torch.arange(-K, K + 1, dtype=torch.float64, device=device)
    return torch.cartesian_prod(*([axis] * n)).reshape(-1, n)


def shell_counts(n: int, m_max: int, device) -> torch.Tensor:
    """(m_max+1,) r_n(m) = number of k in Z^n with |k|^2 = m: the n-th power of the theta series 1 + 2q + 2q^4 + ...

    Polynomials multiply by convolution, done here with an FFT: theta(q)^n without listing (2K+1)^n vectors.
    """
    j = torch.arange(0, int(m_max**0.5) + 1, device=device)
    theta = torch.zeros(m_max + 1, dtype=torch.float64, device=device)
    theta[j**2] = 2.0
    theta[0] = 1.0
    size = 1 << (n * (m_max + 1)).bit_length()
    power = torch.fft.irfft(torch.fft.rfft(theta, size) ** n, size)[: m_max + 1]
    return torch.round(power)


def torsion_limits() -> list[dict]:
    """Cited torsion-balance limits: |alpha| = 1 excluded above lambda (95% CL)."""
    with open(DATA / "torsion_balance_limits.csv", encoding="utf-8") as f:
        return [row for row in csv.DictReader(r for r in f if not r.startswith("#"))]


def lhc_limits() -> tuple[torch.Tensor, torch.Tensor]:
    """ATLAS 2021 monojet lower limits on M_D (TeV) for n = 2..6 extra dimensions."""
    with open(DATA / "lhc_add_limits.csv", encoding="utf-8") as f:
        rows = [row for row in csv.DictReader(r for r in f if not r.startswith("#"))]
    n = torch.tensor([float(r["n"]) for r in rows], dtype=torch.float64)
    m = torch.tensor([float(r["M_D_min_TeV"]) for r in rows], dtype=torch.float64)
    return n, m
```

## 4. `solve.py` — solver

```python
%%writefile problems/cosmology/hierarchy_extra_dimensions/solve.py
"""Solvers: the potential crossover for every n, Yukawa fits, the ADD table against the limits, the RS warp scan."""

import math

import torch

from . import equation, tensors


def crossover(ns: list[int], m_max: int, r: torch.Tensor) -> dict:
    """phi(r) r / (G_4 M) for each n on a log grid r (R,): images inside the compact space, KK modes outside."""
    counts = {n: tensors.shell_counts(n, m_max, r.device) for n in ns}
    inner, outer = r < 1.0, r >= 1.0
    ratio, slope, fits = [], [], []
    for n in ns:  # 6 values of n; each line is fully batched over r
        phi = torch.cat(
            [equation.potential_images(r[inner], n, counts[n]), equation.potential_kk(r[outer], n, counts[n])]
        )
        ratio.append(-phi * r / equation.g4(n))
        f_in = lambda x, n=n: equation.potential_images(x, n, counts[n])  # noqa: E731
        f_out = lambda x, n=n: equation.potential_kk(x, n, counts[n])  # noqa: E731
        slope.append(torch.cat([equation.log_slope(f_in, r[inner]), equation.log_slope(f_out, r[outer])]))
        r_fit = torch.linspace(16.0, 24.0, 25, dtype=r.dtype, device=r.device)
        fits.append(equation.fit_yukawa(r_fit, equation.kk_deviation(r_fit, n, counts[n])))
    agree = max(
        (equation.potential_images(r[(r > 0.6) & (r < 3)], n, counts[n])
         / equation.potential_kk(r[(r > 0.6) & (r < 3)], n, counts[n]) - 1).abs().max().item()
        for n in ns
    )  # fmt: skip
    return {"ratio": torch.stack(ratio), "slope": torch.stack(slope), "fits": fits, "poisson_gap": agree,
            "counts": counts}  # fmt: skip


def add_table(ns: torch.Tensor, m_star: torch.Tensor) -> dict:
    """ADD radius R(n, M*) (N, M) in metres; Yukawa strength alpha = 2n (torus); exclusions from the bundled limits."""
    R = equation.add_radius_m(ns[:, None], m_star[None, :])
    alpha = (2 * ns)[:, None].expand_as(R)
    lim = {row["source"]: float(row["lambda_max_um"]) * 1e-6 for row in tensors.torsion_limits()}
    lam_lee = lim["Lee et al."]
    torsion = equation.excluded_by_torsion(alpha, R, lam_lee)
    n_lhc, m_lhc = tensors.lhc_limits()
    lhc_min = torch.zeros_like(ns)
    for i, n in enumerate(ns.tolist()):
        hit = (n_lhc == n).nonzero()
        lhc_min[i] = m_lhc[hit[0, 0]] if len(hit) else float("nan")
    lhc = m_star[None, :] < lhc_min[:, None]  # NaN (n = 1) compares False
    return {"R": R, "alpha": alpha, "torsion": torsion, "lhc": lhc, "lam_limits": lim, "lhc_min": lhc_min}


def rs_scan(krc: torch.Tensor) -> torch.Tensor:
    """TeV-brane scale M_Pl e^(-k pi r_c) in GeV for every k r_c (K,)."""
    return tensors.M_PL_REDUCED_GEV * equation.rs_warp(krc)


def slice_field(half_width: float, n_grid: int, device) -> dict:
    """n = 1 slice (x, y): x in our 3D space, y around the compact circle (unrolled), window [-w, w]^2."""
    s = torch.linspace(-half_width, half_width, n_grid, dtype=torch.float64, device=device)
    y, x = torch.meshgrid(s, s, indexing="ij")
    xg = x.clone().requires_grad_(True)
    yg = y.clone().requires_grad_(True)
    phi = equation.slice_potential_n1(xg, yg)
    gx, gy = torch.autograd.grad(phi.sum(), (xg, yg))
    return {"x": s, "phi": phi.detach(), "Fx": -gx, "Fy": -gy, "w": half_width}


def window_scales(n_frames: int, w_max: float = 6 * math.pi, w_min: float = 0.25) -> torch.Tensor:
    """Zoom from far outside (w >> 2 pi R) to deep inside the compact space (w << R), log spaced."""
    return torch.logspace(math.log10(w_max), math.log10(w_min), n_frames, dtype=torch.float64)
```

## 5. `plot.py` — 2D/3D plot → GIF

```python
%%writefile problems/cosmology/hierarchy_extra_dimensions/plot.py
"""Animate gravity spreading into a compact dimension, the 1/r^(1+n) -> 1/r crossover and the alpha-lambda limits."""

import math
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

from problems._common.plotting import ACCENT, save_gif

from . import equation, solve


def animate(r, ns, cr, zoom, add, m_star, krc, tev, krc_sel, dev, path: Path) -> Path:
    rr = r.cpu().numpy()
    ratio = cr["ratio"].cpu().numpy()
    Z = len(zoom)
    sweep = m_star[1:].cpu().numpy()
    S = len(sweep)
    R_all = add["R"][:, 1:].cpu().numpy()  # (6, S)
    alpha = add["alpha"][:, 0].cpu().numpy()
    exc = (add["torsion"][:, 1:] | add["lhc"][:, 1:]).cpu().numpy()
    exc[0] = True  # n = 1: solar-system size, excluded by planetary orbits
    lam_lee = add["lam_limits"]["Lee et al."]
    lam_kap = add["lam_limits"]["Kapner et al."]

    fig, axs = plt.subplots(2, 2, figsize=(9.0, 7.0))
    ax_f, ax_p, ax_e, ax_r = axs[0, 0], axs[0, 1], axs[1, 0], axs[1, 1]

    # --- (b) crossover ---
    for i, n in enumerate(ns):
        ax_p.loglog(rr, ratio[i], color=ACCENT[i], lw=1.6, label=f"n = {n}")
        ax_p.loglog(rr, 1 / (equation.g4(n) * rr**n), color=ACCENT[i], lw=0.7, ls=":")  # pure 1/r^(1+n)
    ax_p.axhline(1, color="#111827", lw=0.8)
    ax_p.set_ylim(0.5, 1e6)
    ax_p.set_xlim(rr[0], rr[-1])
    ax_p.set_xlabel("distance $r/R$")
    ax_p.set_ylabel("$|\\phi|\\,r / G_4M$  (1 = Newton)")
    ax_p.legend(fontsize=7, ncol=2, loc="upper right")
    ax_p.set_title("$1/r^{1+n}$ inside (dotted) → $1/r$ outside", fontsize=9)
    marker = ax_p.axvline(1, color="#f59e0b", lw=1.5)

    # --- (c) alpha-lambda plane ---
    ax_e.fill_between([lam_lee, 1e15], 1, 1e3, color="#fecaca", alpha=0.7, lw=0)
    ax_e.text(2e-3, 30, "excluded: $\\alpha\\geq 1$, $\\lambda > 38.6\\,\\mu$m\n(Eöt-Wash, Lee et al. 2020)",
              fontsize=7,
              color="#991b1b")  # fmt: skip
    ax_e.plot([lam_lee], [1], "s", color="#991b1b", ms=5)
    ax_e.plot([lam_kap], [1], "D", color="#b45309", ms=4, label="Kapner 2007: 56 μm")
    ax_e.axvline(1.496e11, color="#9ca3af", ls="--", lw=0.8)
    ax_e.text(1.6e11, 0.15, "1 AU", fontsize=7, color="#6b7280")
    ax_e.set_xscale("log")
    ax_e.set_yscale("log")
    ax_e.set_xlim(1e-19, 1e15)
    ax_e.set_ylim(0.1, 1e3)
    ax_e.set_xlabel("Yukawa range $\\lambda = R$ (m)")
    ax_e.set_ylabel("strength $\\alpha = 2n$")
    pts = ax_e.scatter(R_all[:, 0], alpha, c=["#dc2626"] * 6, s=40, zorder=3, edgecolors="#111827")
    labels = [ax_e.text(R_all[i, 0], alpha[i] * 1.25, f"n={n}", fontsize=7, ha="center") for i, n in enumerate(ns)]
    ax_e.legend(fontsize=7, loc="lower right")
    e_title = ax_e.set_title("", fontsize=9)

    # --- (d) Randall-Sundrum ---
    k, t = krc.cpu().numpy(), tev.cpu().numpy()
    ax_r.semilogy(k, t, color=ACCENT[3], lw=2)
    ax_r.axhspan(1e3, 1e4, color="#bbf7d0", alpha=0.7, lw=0)
    ax_r.text(9.2, 2.2e3, "1–10 TeV", fontsize=8, color="#166534")
    w_sel = math.exp(-math.pi * krc_sel)
    ax_r.plot([krc_sel], [2.435e18 * w_sel], "o", color="#111827")
    ax_r.set_xlabel("$k r_c$")
    ax_r.set_ylabel("$\\bar M_{Pl}\\,e^{-\\pi k r_c}$ (GeV)")
    ax_r.set_title(f"Randall–Sundrum: $kr_c$ = {krc_sel:g} → warp {w_sel:.1e}", fontsize=9)

    def update(f: int):
        w = float(zoom[min(f, Z - 1)])
        if f < Z:  # zoom phase: recompute the field slice in the new window
            ax_f.clear()
            sl = solve.slice_field(w, 56, dev)
            x = sl["x"].cpu().numpy()
            phi = sl["phi"].cpu().numpy()
            Fx, Fy = sl["Fx"].cpu().numpy(), sl["Fy"].cpu().numpy()
            ax_f.contourf(x, x, np.log10(-phi), levels=12, cmap="Blues")
            ax_f.streamplot(x, x, Fx, Fy, color="#1e3a8a", density=0.8, linewidth=0.6, arrowsize=0.6)
            for j in range(-int(w / (2 * math.pi)) - 1, int(w / (2 * math.pi)) + 2):
                yc = 2 * math.pi * j
                ax_f.axhline(yc + math.pi, color="#dc2626", ls="--", lw=0.8)
                if abs(yc) <= w:
                    ax_f.plot([0], [yc], "o", color="#f59e0b" if j else "#111827", ms=4)
            ax_f.set_xlim(-w, w)
            ax_f.set_ylim(-w, w)
            ax_f.set_aspect("equal")
            ax_f.set_xlabel("$x$ (our space, units of $R$)")
            ax_f.set_ylabel("$y$ (compact circle, unrolled)")
            ax_f.set_title(f"n = 1: field lines, window {2 * w:.2g} R\n"
                           + ("far: flat (spreads in 3D, $1/r$)" if w > 6 else "near: radial in 4D ($1/r^2$)" if w < 1.5
                              else "crossover at $r \\sim R$"), fontsize=9)  # fmt: skip
        marker.set_xdata([w])
        s = f - Z if f >= Z else 0
        pts.set_offsets(np.c_[R_all[:, s], alpha])
        pts.set_color(np.where(exc[:, s], "#dc2626", "#16a34a"))
        pts.set_edgecolor("#111827")
        for i in range(6):
            labels[i].set_position((R_all[i, s], alpha[i] * 1.25))
        e_title.set_text(f"ADD points, M* = {sweep[s]:.2f} TeV (red: excluded by\ntorsion balances, the LHC or orbits)")

    fig.tight_layout()
    return save_gif(fig, update, list(range(Z + S)), path, fps=8)
```
