# The Hodge conjecture

*maths · advanced* · [open in Tensorizer](https://jackyko1991.github.io/tensorizer/#/maths/hodge_conjecture)

Compute Hodge diamonds by counting: for Fermat hypersurfaces $x_0^d + \dots + x_{n+1}^d = 0$ the Hodge numbers are coefficients of a polynomial power, and for complex tori and products they are outer products and convolutions. Shioda's character count then gives the rational Hodge classes, the classes the Hodge conjecture says should come from subvarieties. For surfaces this is the proven Lefschetz $(1,1)$ theorem, and the lines on the surface show it. In higher dimension the conjecture is open; this page only counts the classes.

**Vector form**

$$H^k(X,\mathbb C) = \bigoplus_{p+q=k} H^{p,q},\quad h^{p,q} = h^{q,p} = h^{n-p,n-q},\qquad \text{Hodge: } H^{2p}(X,\mathbb Q)\cap H^{p,p} = \operatorname{span}_{\mathbb Q}\{[Z] : Z \subset X,\ \operatorname{codim} Z = p\}\ ?$$

**Index (tensor) form**

$$h^{n-q,q}_{\text{prim}} = \big[t^{(q+1)d-(n+2)}\big]\Big(\sum_{b=0}^{d-2} t^b\Big)^{n+2},\qquad h^{T}_{pq} = \binom{g}{p}\binom{g}{q},\qquad h^{X\times Y}_{pq} = \sum_{ab} h^X_{ab}\,h^Y_{p-a,q-b},\qquad \rho = 1 + \#\Big\{a : \sum_i \Big\langle \frac{t a_i}{d}\Big\rangle = \frac n2 + 1\ \forall t\in(\mathbb Z/d)^\times\Big\}$$

| Symbol | Code | Shape | Meaning |
|---|---|---|---|
| $\big(\sum_b t^b\big)^{n+2}$ | `conv1d(poly, digit.flip(0), padding=d - 2)` | (L,) | multiplying polynomials is convolving coefficient vectors, repeated $n+2$ times: the monomial basis $x^b$, $0 \le b_i \le d-2$, of the Jacobian ring of the Fermat polynomial |
| $h^{n-q,q}_{\text{prim}}$ | `poly[(q + 1) * d - (n + 2)]` | (n + 1,) | Griffiths: the primitive middle Hodge numbers are the Jacobian ring in degree $(q+1)d - (n+2)$ |
| $h^{T}_{pq}$ | `einsum('p,q->pq', binom, binom)` | (g + 1, g + 1) | a complex torus: $\wedge^p$ of the $dz$ times $\wedge^q$ of the $d\bar z$ |
| $h^{X\times Y}_{pq}$ | `einsum('ab,cd->abcd', h1, h2) + index_put_(accumulate=True)` | (m + k + 1, m + k + 1) | Künneth: every pair of classes, summed into the cell $(a + c, b + d)$ |
| $a = (a_0, \dots, a_{n+1})$ | `characters(n, d)` | (A, n + 2) | Shioda's characters: $1 \le a_i \le d-1$, $\sum a_i \equiv 0 \bmod d$; each labels a line $V(a)$ of primitive cohomology |
| $\sum_i \langle t a_i/d\rangle$ | `((t[:, None, None] * chars) % d).sum(dim=2)` | (T, A) | repeated $i$ ⇒ summed: the Hodge type of every Galois conjugate $t\cdot a$ at once |
| $\forall t$ | `(conj_sum == d * (n // 2 + 1)).all(dim=0)` | (A,) bool | a rational Hodge class: type $(\frac n2, \frac n2)$ for every conjugate |
| $L_i \cdot L_j$ | `det(cat([lines_i, lines_j])).abs() < 1e-9` | (L, L) | two lines meet iff their four spanning vectors in $\mathbb C^4$ are dependent; $L^2 = 2-d$ by adjunction |
| $\operatorname{rank},\ (1, \rho - 1)$ | `eigvalsh(gram)` | (L,) | rank of the lattice spanned by the lines, and its signature (Hodge index theorem: exactly one positive eigenvalue) |

On a compact Kähler manifold every cohomology class splits into pieces of type $(p, q)$: $p$ holomorphic and $q$ antiholomorphic directions (Hodge, 1941). The numbers $h^{p,q}$ satisfy Hodge symmetry $h^{p,q} = h^{q,p}$ and Serre duality $h^{p,q} = h^{n-p,n-q}$, and every diamond on the left passes both checks. Their alternating sum is the Euler characteristic, which for a degree-$d$ hypersurface equals $((1-d)^{n+2} - 1)/d + n + 2$. The diamond reproduces it, along with $h^{1,1} = 20$ for the K3 quartic, $h^{2,1} = 101$ for the quintic threefold and $h^{3,1} = 1$ for the cubic fourfold. A subvariety $Z$ of codimension $p$ defines a rational class of type $(p, p)$. The Hodge conjecture (Hodge 1950; one of the Clay Millennium Prize problems) says that every rational $(p, p)$ class is a rational combination of such classes. What is proven: $p = 1$ by the Lefschetz $(1,1)$ theorem (1924), hence $p = n-1$ by hard Lefschetz. It also holds for abelian varieties of dimension at most 3, for uniruled fourfolds and cubic fourfolds (Conte–Murre, Zucker), and in many special cases, including Fermat varieties of many degrees (Shioda 1979, Ran 1980). The integral version is false (Atiyah–Hirzebruch, 1962). For Fermat varieties Shioda's characters make the rational Hodge classes countable, with one condition per Galois conjugate. For surfaces the count is the Picard number: $7, 20, 37, 86$ for $d = 3, 4, 5, 6$, and $3(d-1)(d-2) + 1$ when $\gcd(d, 6) = 1$. The $3d^2$ lines on the surface give explicit algebraic classes, and their intersection matrix (rank and Hodge-index signature $(1, \rho - 1)$, computed here) spans all of them over $\mathbb Q$ exactly when $d \le 4$ or $\gcd(d, 6) = 1$. For $d = 6, 8, 9, 10$ other curves are needed, which exist by Lefschetz. For Fermat fourfolds the program counts the Hodge classes, but whether each one is algebraic is the conjecture itself. Status (November 2026): open. The AI-generated manuscript collection published by OpenAI (github.com/openai/math, family 032) claims the rational Hodge conjecture for every CM abelian variety and reports a Lean formalisation of that statement. It is an unrefereed, machine-produced claim awaiting verification by the mathematical community, and even if correct it covers a special class, not the general conjecture.

## 2. `equation.py` — physics → tensor translation

```python
%%writefile problems/maths/hodge_conjecture/equation.py
r"""Mathematics → tensor translation.

1. Hodge numbers of a smooth hypersurface X of degree d in ℙ^{n+1} (Griffiths). The primitive middle cohomology is
   the Jacobian ring ℂ[x]/(∂F) in degree (q + 1)d − (n + 2); for the Fermat F = Σ x_i^d it has the monomial basis
   x^b with 0 ≤ b_i ≤ d − 2, so

        h^{n−q,q}_prim = [t^{(q+1)d − (n+2)}] (1 + t + … + t^{d−2})^{n+2},

   a coefficient of a polynomial power (repeated 1-D convolution). Off the middle row h^{p,p} = 1 (Lefschetz).

2. Complex tori of dimension g: h^{p,q} = C(g, p) C(g, q), an outer product. Products: Künneth, h(X × Y) = h(X) * h(Y)
   as a 2-D convolution. Euler characteristic χ = Σ (−1)^{p+q} h^{p,q}.

3. Hodge classes on Fermat varieties (Shioda, 1979). V(a) has type (n−q, q) with q + 1 = Σ_i ⟨a_i/d⟩, and it lies in
   H^n(X, ℚ) ∩ H^{n/2,n/2} for every Galois conjugate exactly when

        Σ_i ⟨t a_i / d⟩ = n/2 + 1   for every unit t mod d.

   So dim(Hodge classes in H^n) = 1 + #{a : condition}. For surfaces (n = 2) this is the Picard number ρ, and by the
   Lefschetz (1,1) theorem all of these classes are algebraic.

4. Lines on a Fermat surface: two lines meet iff their four spanning vectors are linearly dependent (det = 0);
   L² = 2 − d by adjunction. The rank of the Gram matrix is the rank of the lattice the lines span in NS(X).
"""

import math

import torch

from . import tensors


def jacobian_hodge(n: int, d: int) -> torch.Tensor:
    """h^{n−q,q}_prim for q = 0 … n: coefficients of (1 + t + … + t^{d−2})^{n+2}.   → (n + 1,) int64"""
    if d < 2:
        return torch.zeros(n + 1, dtype=torch.int64)
    digit = torch.ones(d - 1, dtype=torch.float64)  # 1 + t + … + t^{d−2}
    poly = torch.ones(1, dtype=torch.float64)
    for _ in range(n + 2):  # multiply polynomials = full convolution of coefficient vectors
        poly = torch.nn.functional.conv1d(poly[None, None], digit.flip(0)[None, None], padding=d - 2)[0, 0]
    degree = (torch.arange(n + 1) + 1) * d - (n + 2)
    inside = (degree >= 0) & (degree < len(poly))
    return torch.where(inside, poly[degree.clamp(0, len(poly) - 1)], 0.0).round().long()


def fermat_diamond(n: int, d: int) -> torch.Tensor:
    """Full Hodge diamond h[p, q] of the degree-d hypersurface in ℙ^{n+1} (n ≥ 1).   → (n + 1, n + 1) int64"""
    h = torch.eye(n + 1, dtype=torch.int64)  # h^{p,p} = 1: powers of the hyperplane class
    prim = jacobian_hodge(n, d)  # (n + 1,) indexed by q
    q = torch.arange(n + 1)
    h[n - q, q] += prim
    return h


def torus_diamond(g: int) -> torch.Tensor:
    """h^{p,q} = C(g,p) C(g,q) for a complex torus ℂ^g/Λ: an outer product.   → (g + 1, g + 1)"""
    binom = torch.tensor([math.comb(g, p) for p in range(g + 1)], dtype=torch.int64)
    return torch.einsum("p,q->pq", binom, binom)


def kunneth(h1: torch.Tensor, h2: torch.Tensor) -> torch.Tensor:
    """h(X × Y)[p, q] = Σ h1[a, b] h2[p − a, q − b]: a 2-D convolution.   (m+1, m+1), (k+1, k+1) → (m+k+1, m+k+1)"""
    m, k = h1.shape[0], h2.shape[0]
    outer = torch.einsum("ab,cd->abcd", h1, h2)  # every pair of classes
    p = (torch.arange(m)[:, None, None, None] + torch.arange(k)[None, None, :, None]).expand(m, m, k, k)
    q = (torch.arange(m)[None, :, None, None] + torch.arange(k)[None, None, None, :]).expand(m, m, k, k)
    out = torch.zeros(m + k - 1, m + k - 1, dtype=h1.dtype)
    out.index_put_((p.reshape(-1), q.reshape(-1)), outer.reshape(-1), accumulate=True)
    return out


def euler_characteristic(h: torch.Tensor) -> int:
    """χ = Σ_{p,q} (−1)^{p+q} h^{p,q}."""
    p = torch.arange(h.shape[0])
    sign = (-1) ** (p[:, None] + p[None, :])
    return int((sign * h).sum())


def euler_formula(n: int, d: int) -> int:
    """χ of a smooth degree-d hypersurface in ℙ^{n+1}: ((1 − d)^{n+2} − 1)/d + n + 2 (from c(T_X))."""
    return ((1 - d) ** (n + 2) - 1) // d + n + 2


def hodge_type(chars: torch.Tensor, d: int) -> torch.Tensor:
    """q(a) with q + 1 = Σ_i a_i / d (a_i in 1 … d−1), so V(a) ⊂ H^{n−q,q}.   (A, n+2) → (A,)"""
    return chars.sum(dim=1) // d - 1


def hodge_class_mask(chars: torch.Tensor, d: int) -> torch.Tensor:
    """Shioda's condition Σ_i ⟨t a_i/d⟩ = n/2 + 1 for every unit t, on all characters at once.   → (A,) bool"""
    t = tensors.units(d)  # (T,)
    n_half_plus_one = chars.shape[1] // 2  # (n + 2)/2
    conj = (t[:, None, None] * chars[None, :, :]) % d  # (T, A, n+2): the Galois conjugates t·a
    return (conj.sum(dim=2) == d * n_half_plus_one).all(dim=0)


def hodge_class_count(n: int, d: int) -> int:
    """dim of the rational Hodge classes in H^n of the Fermat variety (n even): 1 + #{a satisfying Shioda}."""
    if n % 2:
        return 0
    return 1 + int(hodge_class_mask(tensors.characters(n, d), d).sum())


def line_gram(d: int) -> torch.Tensor:
    """Intersection matrix of the 3d² lines on the Fermat surface of degree d.   → (L, L) int64"""
    lines = tensors.fermat_lines(d)  # (L, 2, 4)
    n_lines = len(lines)
    pair = torch.cat(
        [lines[:, None].expand(-1, n_lines, -1, -1), lines[None, :].expand(n_lines, -1, -1, -1)], dim=2
    )  # (L, L, 4, 4): the four spanning vectors of lines i and j
    meet = torch.linalg.det(pair).abs() < 1e-9  # dependent ⇔ the lines meet (or coincide)
    gram = meet.long()
    gram.fill_diagonal_(2 - d)  # self-intersection from adjunction
    return gram
```

## 3. `tensors.py` — PyTorch tensors

```python
%%writefile problems/maths/hodge_conjecture/tensors.py
"""Build the concrete tensors: Shioda's characters of Fermat varieties, the units mod d and the lines on a Fermat
surface."""

import math

import torch


def characters(n: int, d: int) -> torch.Tensor:
    """All a = (a_0, …, a_{n+1}) with 1 ≤ a_i ≤ d − 1 and Σ a_i ≡ 0 mod d.   → (A, n + 2) int64

    Each one labels a one-dimensional piece V(a) of the primitive middle cohomology of the Fermat variety
    x_0^d + … + x_{n+1}^d = 0 (the eigenspaces of the group μ_d^{n+2} acting by roots of unity)."""
    digits = torch.arange(1, d, dtype=torch.int64)
    if d < 2:
        return torch.zeros(0, n + 2, dtype=torch.int64)
    grid = torch.cartesian_prod(*([digits] * (n + 2))).reshape(-1, n + 2)
    return grid[grid.sum(dim=1) % d == 0]


def units(d: int) -> torch.Tensor:
    """t with 1 ≤ t < d and gcd(t, d) = 1: the Galois group of ℚ(e^{2πi/d}).   → (T,)"""
    return torch.tensor([t for t in range(1, d) if math.gcd(t, d) == 1] or [1], dtype=torch.int64)


def fermat_lines(d: int) -> torch.Tensor:
    """The 3d² lines on x^d + y^d + z^d + w^d = 0, each spanned by two vectors of ℂ⁴.   → (3d², 2, 4) complex128

    For a pairing of the coordinates, e.g. {x, y}, {z, w}: x = α y, z = β w with α^d = β^d = −1."""
    roots = torch.exp(1j * math.pi * (2 * torch.arange(d, dtype=torch.float64) + 1) / d)  # d-th roots of −1
    alpha, beta = torch.meshgrid(roots, roots, indexing="ij")
    alpha, beta = alpha.flatten(), beta.flatten()  # (d²,)
    one, zero = torch.ones_like(alpha), torch.zeros_like(alpha)
    lines = []
    for (i, j), (k, m) in (((0, 1), (2, 3)), ((0, 2), (1, 3)), ((0, 3), (1, 2))):
        v1 = torch.zeros(len(alpha), 4, dtype=torch.complex128)
        v2 = torch.zeros(len(alpha), 4, dtype=torch.complex128)
        v1[:, i], v1[:, j] = alpha, one  # x_i = α x_j
        v2[:, k], v2[:, m] = beta, one  # x_k = β x_m
        v1[:, k], v1[:, m], v2[:, i], v2[:, j] = zero, zero, zero, zero
        lines.append(torch.stack([v1, v2], dim=1))
    return torch.cat(lines)
```

## 4. `solve.py` — solver

```python
%%writefile problems/maths/hodge_conjecture/solve.py
"""Assemble the Hodge diamonds shown frame by frame, and the Picard-number / Hodge-class tables of Fermat varieties."""

import torch

from .equation import (
    euler_characteristic,
    euler_formula,
    fermat_diamond,
    hodge_class_count,
    kunneth,
    line_gram,
    torus_diamond,
)

NAMES = {
    (1, 3): "plane cubic (elliptic curve)",
    (2, 2): "quadric surface",
    (2, 3): "cubic surface",
    (2, 4): "quartic K3 surface",
    (3, 5): "quintic Calabi–Yau threefold",
    (4, 3): "cubic fourfold",
}


def cases(d_max: int, g_max: int) -> list[dict]:
    """Fermat hypersurfaces, complex tori and products: label, diamond, χ and (for even n) the Hodge-class count."""
    out = []
    for n, degrees in (
        (1, (3, 4, 5)),
        (2, range(2, min(d_max, 7) + 1)),
        (3, (3, 4, 5)),
        (4, range(3, min(d_max, 6) + 1)),
    ):
        for d in degrees:
            h = fermat_diamond(n, d)
            label = NAMES.get((n, d), f"Fermat hypersurface, n = {n}, d = {d}")
            out.append(
                {
                    "label": f"{label}: $x_0^{{{d}}} + \\dots + x_{{{n + 1}}}^{{{d}}} = 0$",
                    "h": h,
                    "chi": euler_characteristic(h),
                    "chi_formula": euler_formula(n, d),
                    "hodge": hodge_class_count(n, d) if n % 2 == 0 else None,
                    "fermat": (n, d),
                }
            )
    for g in range(1, g_max + 1):
        h = torus_diamond(g)
        out.append({"label": f"complex torus of dimension {g}", "h": h, "chi": euler_characteristic(h), "torus": g})
    k3, e, c3 = fermat_diamond(2, 4), torus_diamond(1), fermat_diamond(1, 4)
    for label, h in (
        ("K3 × elliptic curve (Künneth)", kunneth(k3, e)),
        ("genus-3 curve × genus-3 curve", kunneth(c3, c3)),
    ):
        out.append({"label": label, "h": h, "chi": euler_characteristic(h)})
    return out


def surface_table(d_max: int) -> dict:
    """For Fermat surfaces d = 3 … d_max: h^{1,1}, Picard number ρ (Shioda) and the rank of the lattice of lines."""
    ds = list(range(3, d_max + 1))
    h11 = [int(fermat_diamond(2, d)[1, 1]) for d in ds]
    rho = [hodge_class_count(2, d) for d in ds]
    ranks, signs = [], []
    for d in ds:
        ev = torch.linalg.eigvalsh(line_gram(d).double())
        tol = 1e-8 * ev.abs().max()
        ranks.append(int((ev.abs() > tol).sum()))
        signs.append((int((ev > tol).sum()), int((ev < -tol).sum())))
    return {"d": ds, "h11": h11, "rho": rho, "lines": ranks, "signature": signs}


def fourfold_table(d_max: int) -> dict:
    """For Fermat fourfolds d = 3 … d_max: h^{2,2} and the dimension of rational Hodge classes in H⁴."""
    ds = list(range(3, d_max + 1))
    return {
        "d": ds,
        "h22": [int(fermat_diamond(4, d)[2, 2]) for d in ds],
        "hodge": [hodge_class_count(4, d) for d in ds],
    }


def euler_check(n_max: int, d_max: int) -> torch.Tensor:
    """(n, d, χ from the diamond, χ from the formula) for every Fermat hypersurface in range.   → (K, 4)"""
    rows = [
        (n, d, euler_characteristic(fermat_diamond(n, d)), euler_formula(n, d))
        for n in range(1, n_max + 1)
        for d in range(1, d_max + 1)
    ]
    return torch.tensor(rows, dtype=torch.int64)
```

## 5. `plot.py` — 2D/3D plot → GIF

```python
%%writefile problems/maths/hodge_conjecture/plot.py
"""Hodge diamonds one by one, with Picard numbers of Fermat surfaces, Hodge classes of Fermat fourfolds and χ checks."""

from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

from problems._common.plotting import ACCENT, frame_indices, save_gif


def animate(shown: list, surf: dict, four: dict, check, path: Path) -> Path:
    fig = plt.figure(figsize=(12, 7))
    grid = fig.add_gridspec(2, 3, width_ratios=[1.5, 1, 1])
    ax_h = fig.add_subplot(grid[:, 0])
    ax_s = fig.add_subplot(grid[0, 1:])
    ax_f = fig.add_subplot(grid[1, 1])
    ax_e = fig.add_subplot(grid[1, 2])

    ds = np.array(surf["d"])
    w = 0.27
    ax_s.bar(ds - w, surf["h11"], w, color="#cbd5e1", label=r"$h^{1,1}$")
    ax_s.bar(ds, surf["rho"], w, color=ACCENT[0], label=r"Hodge classes $\rho$ (Shioda) = algebraic (Lefschetz)")
    ax_s.bar(ds + w, surf["lines"], w, color=ACCENT[4], label=r"rank spanned by the $3d^2$ lines")
    ax_s.set_yscale("log")
    ax_s.set_xlabel("degree $d$ of the Fermat surface $x^d + y^d + z^d + w^d = 0$")
    ax_s.legend(fontsize=7, loc="upper left")
    ax_s.set_title(r"surfaces: Hodge classes in $H^2$ are all algebraic (proved)", fontsize=9)
    mark_s = ax_s.axvspan(-11, -10, color="#fde68a", alpha=0.5, zorder=0)

    d4 = np.array(four["d"])
    ax_f.bar(d4 - 0.2, four["h22"], 0.4, color="#cbd5e1", label=r"$h^{2,2}$")
    ax_f.bar(d4 + 0.2, four["hodge"], 0.4, color=ACCENT[3], label="rational Hodge classes")
    ax_f.set_yscale("log")
    ax_f.set_xlabel("degree $d$ (Fermat fourfold)")
    ax_f.legend(fontsize=7, loc="upper left")
    ax_f.set_title("fourfolds: algebraic? (the conjecture)", fontsize=9)
    mark_f = ax_f.axvspan(-11, -10, color="#fde68a", alpha=0.5, zorder=0)

    chk = check.numpy()
    ax_e.scatter(np.abs(chk[:, 3]) + 1, np.abs(chk[:, 2]) + 1, s=14, c=chk[:, 0], cmap="viridis", zorder=3)
    top = np.abs(chk[:, 2:]).max() + 1
    ax_e.plot([1, top], [1, top], color=ACCENT[1], lw=1)
    ax_e.set_xscale("log")
    ax_e.set_yscale("log")
    ax_e.set_xlabel(r"$|\chi|+1$ from $((1-d)^{n+2}-1)/d + n + 2$")
    ax_e.set_ylabel(r"$|\chi|+1$ from the diamond")
    ax_e.set_title(f"Euler characteristic, {len(chk)} hypersurfaces", fontsize=9)
    (mark_e,) = ax_e.plot([], [], "o", ms=11, mfc="none", mec=ACCENT[1], mew=2)

    def draw_diamond(case: dict):
        ax_h.clear()
        h = case["h"].numpy()
        n = h.shape[0] - 1
        vmax = np.log1p(h.max())
        for p in range(n + 1):
            for q in range(n + 1):
                x, y = q - p, p + q  # h^{0,0} at the bottom, h^{n,n} at the top
                middle = p + q == n
                colour = plt.cm.Blues(0.15 + 0.6 * np.log1p(h[p, q]) / max(vmax, 1e-9))
                ax_h.scatter(
                    x,
                    y,
                    s=1300 if n <= 4 else 750,
                    marker="D",
                    color=colour,
                    edgecolor=ACCENT[1] if middle else "#9ca3af",
                    lw=1.6 if middle else 0.8,
                )
                size = 11 if h[p, q] < 1000 else 8
                ax_h.text(
                    x,
                    y,
                    str(h[p, q]),
                    ha="center",
                    va="center",
                    fontsize=size if n <= 4 else size - 2,
                    color="white" if np.log1p(h[p, q]) > 0.6 * vmax and h[p, q] > 1 else "#111827",
                )
        ax_h.set_xlim(-n - 1, n + 1)
        ax_h.set_ylim(-1.2, 2 * n + 1.5)
        ax_h.set_aspect("equal")
        ax_h.axis("off")
        lines = [case["label"], f"dimension n = {n},  χ = {case['chi']}"]
        if "chi_formula" in case:
            lines[-1] += f"  (formula: {case['chi_formula']})"
        if case.get("hodge") is not None:
            lines.append(
                rf"rational Hodge classes in $H^{{{n}}}$: {case['hodge']} "
                rf"of $h^{{{n // 2},{n // 2}}}$ = {h[n // 2, n // 2]}"
            )
        if "torus" in case:
            lines.append(r"$h^{p,q} = \binom{g}{p}\binom{g}{q}$")
        ax_h.set_title("\n".join(lines), fontsize=9)
        ax_h.text(
            0,
            -1.0,
            r"symmetric under $p \leftrightarrow q$ (Hodge) and $(p,q) \to (n-p, n-q)$ (Serre)",
            ha="center",
            fontsize=7.5,
            color="#6b7280",
        )

    def update(f: int):
        case = shown[f]
        draw_diamond(case)
        fermat = case.get("fermat")
        for patch, xs, on in ((mark_s, ds, fermat and fermat[0] == 2), (mark_f, d4, fermat and fermat[0] == 4)):
            x = fermat[1] if on and fermat[1] in xs else -10
            patch.set_x(x - 0.5)
            patch.set_width(1.0)
        if fermat:
            row = chk[(chk[:, 0] == fermat[0]) & (chk[:, 1] == fermat[1])]
            mark_e.set_data(np.abs(row[:, 3]) + 1, np.abs(row[:, 2]) + 1)
        else:
            mark_e.set_data([], [])

    ax_s.set_xlim(ds[0] - 0.6, ds[-1] + 0.6)
    ax_f.set_xlim(d4[0] - 0.6, d4[-1] + 0.6)
    fig.tight_layout()
    return save_gif(fig, update, frame_indices(len(shown), len(shown)), path, fps=1)
```
