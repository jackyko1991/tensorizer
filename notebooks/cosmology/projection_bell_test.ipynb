# Bell tests of the projection race

*cosmology · advanced* · [open in Tensorizer](https://jackyko1991.github.io/tensorizer/#/cosmology/projection_bell_test)

A speculative hypothesis, not established physics: two entangled particles are one shared state $\Psi$ in $D$ dimensions, and each measurement is a projection followed by the race of step 1. Run jointly on $\Psi$, as one event in the full space, the race gives exactly the quantum $E = -\cos\theta$ and CHSH $S = 2\sqrt2$. Run separately on each party's local hidden coordinates, it is a local hidden-variable model with $S\leq2$, which loophole-free Bell tests rule out. A race with more than one hidden dimension per outcome breaks Tsirelson's bound and allows faster-than-light signalling.

**Vector form**

$$w_{ij}(a,b) = \big\|(P^a_i\otimes P^b_j)\,\Psi\big\|^2,\quad P^n_\pm = \tfrac12(1\pm n\cdot\sigma),\qquad k^* = \arg\max_k \frac{w_k}{E_k},\qquad E(a,b) = \sum_{ij} s_is_j P_{ij} \overset{m=1}{=} -a\cdot b,\qquad S = \big|E(a,b) - E(a,b^\prime) + E(a^\prime,b) + E(a^\prime,b^\prime)\big| \leq 2\sqrt2,\qquad E_{\rm local}(a,b) = -\big\langle g_m(a\cdot n)\,g_m(b\cdot n)\big\rangle_n,\quad g_m(x) = 2I_{\frac{1+x}{2}}(m,m)-1,\quad S_{\rm local}\leq 2$$

**Index (tensor) form**

$$w_{bij} = \bar\psi_{xy}\,(P^a)_{bixu}\,(P^b)_{bjyv}\,\psi_{uv},\qquad E_b = s_i\,s_j\,P_{bij},\qquad P_{bk} = \int ds\; e^{\,m\ln(w_{bk}t) - w_{bk}t - \ln\Gamma(m) + \sum_l \ln Q(m, w_{bl}t) - \ln Q(m, w_{bk}t)},\qquad E^{\rm loc}_b = -\tfrac1N\, g_m(a_{bk}n_{\nu k})\,g_m(b_{bk}n_{\nu k}),\qquad S_b = \max_q\Big|\sum_p E_{pb} - 2E_{qb}\Big|$$

| Symbol | Code | Shape | Meaning |
|---|---|---|---|
| $P^n_{\pm}$ | `projectors(n)` | (B, 2, 2, 2) | spin projectors along unit vectors $n$, built with an einsum over the Pauli matrices |
| $w_{bij}$ | `einsum('xy,bixu,bjyv,uv->bij', psi.conj(), Pa, Pb, psi)` | (B, 2, 2) | weight of each joint outcome: the qubit indices $x, y, u, v$ are summed |
| $E_b$ | `einsum('i,j,bij->b', s, s, P)` | (B,) | correlation, outcome signs $s = (+1, -1)$ summed |
| $P_{bk}$ | `race_probabilities(w, m)` | (M, B, 4) | the joint race over the four outcomes; the sum over $l$ (minus $l = k$) is the product over the other outcomes; $m = 1$ gives $P = w$ |
| $E_{tk}$ | `hidden_block_energies(T, 4, m, D)` | (T, 4) | hidden-block energies read from samples of $\Psi$ uniform on the sphere of $\mathbb C^D$ |
| $n_{\nu k}$ | `shared_directions(N, D)` | (N, 3) | hidden direction set at the source, three real coordinates of $\Psi$: uniform on $S^2$ for any $D$ |
| $E^{\rm loc}_b$ | `-(g(a @ n.T) * g(b @ n.T)).mean(-1)` | (B,) | local races: each party races on its own weights $(1\pm a\cdot n)/2$; the shared $\nu$ is averaged |
| $S_b$ | `(Es.sum(0) - 2 * Es).abs().amax(0)` | (B,) | CHSH value, maximised over which of the four terms carries the minus sign |

Step 5 of the Projection Hypothesis (see the series description), the author's speculative hypothesis, not established physics. It tests postulates P1 (projection) and P2 (event) of step 1 (cosmology/projection_postulate) on entanglement. Step 4 (cosmology/projection_singularity_fold) is the geometry, and step 6 (cosmology/projection_telescope_tests) the observations. The dimension $D$ of the shared state stays open: it is a parameter, and no result depends on it. Known physics first. The singlet gives $E(a,b) = -a\cdot b = -\cos\theta$. Bell (1964) and Clauser, Horne, Shimony and Holt (1969) proved $S\leq2$ for any local hidden-variable model, while quantum mechanics reaches Tsirelson's bound $S = 2\sqrt2$. The hypothesis: there is one shared $\Psi$ (P0, P3); each party's measurement is a projection plus the race of step 1, where the outcome with the largest $\|P_k\Psi\|^2/E_k$ happens and $E_k$ is the energy in $m$ hidden complex coordinates of $\Psi$. (i) Joint model: the race runs once over the four joint outcomes, with weights $\|(P^a_i\otimes P^b_j)\Psi\|^2$. Nonlocal in 3D, it is a single event in $D$ dimensions. With $m = 1$ (the only value step 1 left alive) it reproduces quantum mechanics exactly: $E = -\cos\theta$ in closed form to $10^{-15}$, and a Monte Carlo from $\Psi$ in $\mathbb C^D$ within statistics. $S = 2\sqrt2$ at the optimal settings, the maximum over 2000 random settings stays below it, and Alice's marginals do not depend on Bob's setting (no signalling). (ii) Local model: each party races only on its own hidden energies and a hidden direction $n$ set at the source, i.e. on weights $(1\pm a\cdot n)/2$. This is a local hidden-variable model, so $S\leq2$ is guaranteed. $m = 1$ gives $E = -\cos\theta/3$ (max $S\approx0.93$), and $m\to\infty$ is Bell's own deterministic model $E = -(1-2\theta/\pi)$, which reaches exactly $S = 2$. Beyond Born: a joint race with $m\neq1$ is not quantum mechanics. For $m > 1$ it breaks Tsirelson's bound ($S = 3.85$ at $m = 3$, tending to the PR-box value 4). For a partly entangled state it lets Alice's statistics depend on Bob's setting, which is faster-than-light signalling ($dS/dm = 1.34$, signalling $\approx0.18\,|m-1|$). So $m = 1$ is required by relativistic causality as well as by the triple-slit data. Data (data/bell_tests.csv, cited numbers from the abstracts). Hensen et al. (Nature 526, 682, 2015): NV spins 1.3 km apart, $S = 2.42\pm0.20$, $p = 0.039$. Their second run (Sci. Rep. 6, 30289, 2016): $S = 2.35\pm0.18$. Giustina et al. and Shalm et al. (Phys. Rev. Lett. 115, 250401 and 250402, 2015): photons, CH–Eberhard inequality, $p\leq3.74\times10^{-31}$ and $2.3\times10^{-7}$. Rosenfeld et al. (Phys. Rev. Lett. 119, 010402, 2017): atoms 398 m apart, $S = 2.221\pm0.033$. Storz et al. (Nature 617, 265, 2023): superconducting qubits 30 m apart, $S = 2.0747\pm0.0033$, 22.6$\sigma$ above 2. Verdict: model (ii), and with it every local version of the race, is ruled out. Model (i) agrees with every test, but only because it is quantum mechanics rewritten; it makes no new prediction. The deviation still allowed by step 1 ($|m-1|\leq6\times10^{-3}$) would shift $S$ by at most $8\times10^{-3}$. That is more than Storz's $\sigma_S$, but measured $S$ values sit well below $2\sqrt2$ (visibility 0.73), and an unknown visibility absorbs such a shift. Only the shape of $E(\theta)$ (residual $\leq1.5\times10^{-3}$) or a direct no-signalling test ($\leq10^{-3}$) could see it. Wikipedia: Bell's theorem, CHSH inequality.

## 2. `equation.py` — physics → tensor translation

```python
%%writefile problems/cosmology/projection_bell_test/equation.py
"""Postulates -> tensors: joint Born weights of two qubits, the projection race, local races, CHSH, no-signalling.

Shapes in comments. Summed indices: the qubit indices x, y, u, v in w_ij = psi*_xy (P^a_i)_xu (P^b_j)_yv psi_uv, the
outcome signs in E = s_i s_j P_ij, the race product over the other outcomes (a sum of logs) and the quadrature
variable t, and the shared hidden direction n (averaged) in the local model.
"""

import math

import torch

PAULI = torch.tensor(
    [[[0, 1], [1, 0]], [[0, -1j], [1j, 0]], [[1, 0], [0, -1]]], dtype=torch.complex128
)  # (3, 2, 2) sigma_x, sigma_y, sigma_z
SIGNS = torch.tensor([1.0, -1.0], dtype=torch.float64)  # outcome +1, -1

# ---------- quantum mechanics of two qubits ----------


def singlet(device) -> torch.Tensor:
    """(2, 2) psi_xy of (|01> - |10>) / sqrt 2."""
    return partly_entangled(math.pi / 4, device)


def partly_entangled(eta: float, device) -> torch.Tensor:
    """(2, 2) psi_xy of cos(eta) |01> - sin(eta) |10>: the singlet at eta = pi/4."""
    psi = torch.zeros(2, 2, dtype=torch.complex128, device=device)
    psi[0, 1], psi[1, 0] = math.cos(eta), -math.sin(eta)
    return psi


def projectors(n: torch.Tensor) -> torch.Tensor:
    """(..., 2, 2, 2) P_i = (1 + s_i n.sigma) / 2 for the outcomes s = +1, -1 along unit vectors n (..., 3)."""
    ns = torch.einsum("...k,kxy->...xy", n.to(torch.complex128), PAULI.to(n.device))  # n.sigma, k summed
    eye = torch.eye(2, dtype=torch.complex128, device=n.device)
    s = SIGNS.to(n.device).to(torch.complex128)[:, None, None]
    return (eye + s * ns[..., None, :, :]) / 2


def joint_born(psi: torch.Tensor, a: torch.Tensor, b: torch.Tensor) -> torch.Tensor:
    """(..., 2, 2) w_ij = <psi| P^a_i (x) P^b_j |psi> = ||P_ij Psi||^2: the weight of each joint outcome."""
    w = torch.einsum("xy,...ixu,...jyv,uv->...ij", psi.conj(), projectors(a), projectors(b), psi)  # x y u v summed
    return w.real


def correlation(P: torch.Tensor) -> torch.Tensor:
    """(...,) E = sum_ij s_i s_j P_ij from joint outcome probabilities (..., 2, 2)."""
    s = SIGNS.to(P.device)
    return torch.einsum("i,j,...ij->...", s, s, P)


def chsh(E, a, a2, b, b2) -> torch.Tensor:
    """(B,) CHSH S = max over which term carries the minus sign of |E(a,b) + E(a,b') + E(a',b) + E(a',b') - 2 E_k|."""
    Es = torch.stack([E(a, b), E(a, b2), E(a2, b), E(a2, b2)])  # (4, B)
    return (Es.sum(0, keepdim=True) - 2 * Es).abs().amax(0)


# ---------- the projection race (step 1), run on joint outcomes ----------


def race_probabilities(w: torch.Tensor, a: torch.Tensor, n_grid: int = 4000) -> torch.Tensor:
    """P(k wins) for k* = argmax_k w_k / E_k, E_k ~ Gamma(a) i.i.d.; w (..., K), a broadcast to (...). a = 1: P = w.

    P_k = int_0^oo f_a(t; w_k) prod_{j != k} Q(a, w_j t) dt (Q: regularised upper incomplete gamma), on a log grid.
    """
    w = w.double()
    a = torch.as_tensor(a, dtype=torch.float64, device=w.device).expand(w.shape[:-1])  # (...)
    w_pos = torch.where(w > 1e-15 * w.amax(-1, keepdim=True), w, torch.zeros_like(w))
    w_min = torch.where(w_pos > 0, w_pos, torch.full_like(w, math.inf)).amin(-1)
    s_lo = math.log(1e-30) / a - torch.log(w.amax(-1))
    s_hi = torch.log((a + 60 + 12 * a.sqrt()) / w_min)
    u = torch.linspace(0, 1, n_grid, dtype=torch.float64, device=w.device)
    t = torch.exp(s_lo[..., None] + (s_hi - s_lo)[..., None] * u)  # (..., G)
    wt = w_pos[..., :, None] * t[..., None, :]  # (..., K, G)
    aa = a[..., None, None]
    log_q = torch.log(torch.special.gammaincc(aa.expand_as(wt), wt).clamp_min(1e-300))
    log_f_t = aa * torch.log(wt) - wt - torch.lgamma(aa)  # density of ln t
    integrand = torch.exp(log_f_t + log_q.sum(-2, keepdim=True) - log_q)  # product over j != k
    return torch.trapezoid(integrand, dim=-1) * ((s_hi - s_lo)[..., None] / (n_grid - 1))


def joint_race_correlation(w: torch.Tensor, E: torch.Tensor) -> torch.Tensor:
    """(B,) Monte Carlo E(a, b) of the race on the joint outcomes: w (B, 4) weights, E (T, 4) hidden-block energies."""
    win = (w[:, None, :] / E[None]).argmax(-1)  # (B, T) the one event, decided in the full space
    P = torch.nn.functional.one_hot(win, 4).double().mean(1)  # (B, 4)
    return correlation(P.view(-1, 2, 2))


def joint_race_chsh(psi, m: torch.Tensor, a, a2, b, b2) -> torch.Tensor:
    """(M,) CHSH S of the joint race with m hidden complex dimensions per joint outcome, for each m."""
    w = torch.stack([joint_born(psi, a, b), joint_born(psi, a, b2), joint_born(psi, a2, b), joint_born(psi, a2, b2)])
    w = w.flatten(-2)[:, 0]  # (4, 4) setting pair, joint outcome
    P = race_probabilities(w.expand(len(m), 4, 4), m[:, None])  # (M, 4, 4)
    Es = correlation(P.view(len(m), 4, 2, 2)).T  # (4, M)
    return (Es.sum(0, keepdim=True) - 2 * Es).abs().amax(0)


def max_signalling(psi, m: torch.Tensor, generator, device, n_bob: int = 64) -> torch.Tensor:
    """(M,) largest change of Alice's P(+ | a, b) as Bob turns his setting b: zero means no signalling."""
    a = torch.nn.functional.normalize(torch.randn(1, 3, generator=generator, dtype=torch.float64), dim=-1)
    b = torch.nn.functional.normalize(torch.randn(n_bob, 3, generator=generator, dtype=torch.float64), dim=-1)
    w = joint_born(psi, a.to(device).expand(n_bob, 3), b.to(device)).flatten(-2)  # (n_bob, 4)
    P = race_probabilities(w.expand(len(m), n_bob, 4), m[:, None])  # (M, n_bob, 4)
    pA = P[..., 0] + P[..., 1]  # Alice +, Bob either
    return pA.amax(-1) - pA.amin(-1)


# ---------- local races: each party sees only its own hidden coordinates ----------


def local_response(x: torch.Tensor, m: float) -> torch.Tensor:
    """<A> = P(+) - P(-) of a two-outcome race with weights (1 +- x)/2 and m hidden complex dimensions each.

    P(+) = I_w(m, m) = P(Binomial(2m - 1, w) >= m) for integer m; m = 1 gives x itself, m -> oo gives sign(x).
    """
    if math.isinf(m):
        return torch.sign(x)
    mi = int(m)
    if mi != m or mi < 1:
        raise ValueError("local_response needs an integer m >= 1 or inf")
    w = ((1 + x) / 2).clamp(0, 1)
    n = 2 * mi - 1
    j = torch.arange(mi, n + 1, dtype=torch.float64, device=x.device)
    log_c = math.lgamma(n + 1) - torch.lgamma(j + 1) - torch.lgamma(n - j + 1)
    terms = torch.exp(log_c + j * torch.log(w[..., None].clamp_min(1e-300)) + (n - j) * torch.log1p(-w[..., None]))
    return 2 * terms.sum(-1) - 1


def local_correlation(a: torch.Tensor, b: torch.Tensor, n: torch.Tensor, m: float) -> torch.Tensor:
    """(B,) E(a, b) = -< g_m(a.n) g_m(b.n) >_n: Alice races on (1 +- a.n)/2, Bob on (1 -+ b.n)/2, independently.

    n (N, 3) is the hidden direction both can read (set at the source); each party's hidden energies are its own.
    """
    x, y = a @ n.T, b @ n.T  # (B, N)
    return -(local_response(x, m) * local_response(y, m)).mean(-1)  # average over the shared hidden n
```

## 3. `tensors.py` — PyTorch tensors

```python
%%writefile problems/cosmology/projection_bell_test/tensors.py
"""Concrete tensors: measurement directions, CHSH settings, the shared state Psi in C^D, the Bell-test data.

Psi is uniform on the unit sphere of C^D (a complex Gaussian vector divided by its norm). Its hidden coordinates give
the race energies (blocks of m) and, in the local model, a hidden direction n = (Re z1, Im z1, Re z2) / norm that is
uniform on the sphere whatever D is (step 1).
"""

import csv
import math
from pathlib import Path

import torch

DATA = Path(__file__).parent / "data"


def unit(phi: torch.Tensor) -> torch.Tensor:
    """(..., 3) unit vector at angle phi from the z axis in the x-z plane."""
    return torch.stack([torch.sin(phi), torch.zeros_like(phi), torch.cos(phi)], -1)


def random_directions(B: int, generator, device) -> torch.Tensor:
    """(B, 3) uniform unit vectors."""
    v = torch.randn(B, 3, generator=generator, dtype=torch.float64)
    return torch.nn.functional.normalize(v, dim=-1).to(device)


def planar_pairs(theta: torch.Tensor, device):
    """a along z and b at angle theta from it: (B, 3) each."""
    theta = theta.to(device)
    return unit(torch.zeros_like(theta)), unit(theta)


def optimal_settings(device):
    """CHSH-optimal coplanar settings a = 0, a' = 90, b = 45, b' = 135 degrees, each (1, 3)."""
    deg = torch.tensor([0.0, 90.0, 45.0, 135.0], dtype=torch.float64, device=device) * math.pi / 180
    return tuple(unit(deg[i : i + 1]) for i in range(4))


def random_settings(B: int, generator, device):
    """Four independent batches (B, 3) of random directions a, a', b, b'."""
    return tuple(random_directions(B, generator, device) for _ in range(4))


def psi_hidden(T: int, D: int, generator, device) -> torch.Tensor:
    """(T, D) samples of the shared state Psi, uniform on the unit sphere of C^D."""
    x = torch.randn(T, D, 2, generator=generator, dtype=torch.float64)
    z = torch.complex(x[..., 0], x[..., 1])
    return (z / z.norm(dim=1, keepdim=True)).to(device)


def hidden_block_energies(T: int, K: int, m: int, D: int, generator, device) -> torch.Tensor:
    """(T, K) E_k = ||hidden block k||^2: Psi in C^D, 4 visible coordinates, then K blocks of m hidden ones.

    Dividing by ||Psi|| rescales every E_k by the same factor, so the race argmax_k w_k / E_k is that of
    E_k ~ Gamma(m) (step 1).
    """
    if D < 4 + K * m:
        raise ValueError("need D >= 4 + K m")
    h = psi_hidden(T, D, generator, device)[:, 4 : 4 + K * m]
    return (h.abs() ** 2).view(T, K, m).sum(-1)


def shared_directions(N: int, D: int, generator, device) -> torch.Tensor:
    """(N, 3) hidden direction n read from three real coordinates of Psi in C^D: uniform on S^2 for any D."""
    z = psi_hidden(N, D, generator, device)
    n = torch.stack([z[:, -1].real, z[:, -1].imag, z[:, -2].real], -1)
    return torch.nn.functional.normalize(n, dim=-1)


def bell_tests() -> list[dict]:
    """Cited loophole-free Bell tests: experiment, year, platform, inequality, S, sigma, p_value, separation, doi."""
    with open(DATA / "bell_tests.csv", encoding="utf-8") as f:
        return [row for row in csv.DictReader(r for r in f if not r.startswith("#"))]
```

## 4. `solve.py` — solver

```python
%%writefile problems/cosmology/projection_bell_test/solve.py
"""Solvers: correlation curves and CHSH values of the joint and local projection races, signalling, the Bell data."""

import math

import torch

from . import equation, tensors

TSIRELSON = 2 * math.sqrt(2)
# step 1 (projection_postulate): three-path interference allows |m - 1| <= 6e-3 (heralded single photons) and
# <= 1.1e-4 (classical light, if every detection is one race)
SORKIN_BANDS = {"single photons": 6e-3, "classical light": 1.1e-4}
LOCAL_M = [1.0, 2.0, math.inf]


def correlations(m_max: int, D: int, trials: int, generator, device) -> dict:
    """E(theta) for quantum mechanics, the joint race (closed form for m = 1..m_max, Monte Carlo from Psi in C^D
    for m = 1) and the local races (m = 1, 2, oo), plus S(theta) along the CHSH family
    a = 0, b = t, a' = 2t, b' = 3t."""
    psi = equation.singlet(device)
    theta = torch.linspace(0, math.pi, 181, dtype=torch.float64, device=device)
    a, b = tensors.planar_pairs(theta, device)
    w = equation.joint_born(psi, a, b).flatten(-2)  # (Th, 4)
    m = torch.arange(1, m_max + 1, dtype=torch.float64, device=device)
    E_joint = equation.correlation(
        equation.race_probabilities(w.expand(m_max, -1, -1), m[:, None]).view(m_max, -1, 2, 2)
    )
    th_mc = theta[::10]
    E_mc = equation.joint_race_correlation(w[::10], tensors.hidden_block_energies(trials, 4, 1, D, generator, device))
    n = tensors.shared_directions(20_000, D, generator, device)
    E_local = torch.stack([equation.local_correlation(a, b, n, mm) for mm in LOCAL_M])  # (L, Th)

    def S_of(E_curve):  # S(t) = |3 E(t) - E(3t)| from a curve sampled on [0, pi], E even and 2 pi periodic
        t3 = torch.remainder(3 * theta, 2 * math.pi)
        t3 = torch.where(t3 > math.pi, 2 * math.pi - t3, t3)
        idx = (t3 / math.pi * (len(theta) - 1)).round().long()
        return (3 * E_curve - E_curve[..., idx]).abs()

    return {"theta": theta, "E_qm": -torch.cos(theta), "E_joint": E_joint, "th_mc": th_mc, "E_mc": E_mc,
            "E_local": E_local, "S_qm": S_of(-torch.cos(theta)), "S_joint": S_of(E_joint), "S_local": S_of(E_local),
            "mc_err": (E_mc + torch.cos(th_mc)).abs().max().item()}  # fmt: skip


def random_chsh(B: int, D: int, generator, device) -> dict:
    """S over B random measurement settings for the joint Born race (= quantum mechanics) and the local races."""
    settings = tensors.random_settings(B, generator, device)
    psi = equation.singlet(device)
    S_joint = equation.chsh(lambda a, b: equation.correlation(equation.joint_born(psi, a, b)), *settings)
    n = tensors.shared_directions(4000, D, generator, device)
    S_local = torch.stack([equation.chsh(lambda a, b, mm=mm: equation.local_correlation(a, b, n, mm), *settings)
                           for mm in LOCAL_M])  # fmt: skip
    opt = tensors.optimal_settings(device)
    S_local_opt = [
        equation.chsh(lambda a, b, mm=mm: equation.local_correlation(a, b, n, mm), *opt).item() for mm in LOCAL_M
    ]
    return {"S_joint": S_joint, "S_local": S_local, "S_local_opt": S_local_opt}


def beyond_born(m_max: int, generator, device) -> dict:
    """Joint race with m != 1: CHSH at the optimal settings, signalling for a partly entangled state, and the
    deviations allowed by the triple-slit bound of step 1."""
    m = torch.cat([torch.linspace(0.5, 1, 11), torch.linspace(1, m_max, 40)[1:]]).double().to(device)
    opt = tensors.optimal_settings(device)
    S = equation.joint_race_chsh(equation.singlet(device), m, *opt)
    sig = equation.max_signalling(equation.partly_entangled(0.3, device), m, generator, device)
    h = torch.tensor([1 - 1e-3, 1 + 1e-3], dtype=torch.float64, device=device)
    dS = equation.joint_race_chsh(equation.singlet(device), h, *opt).diff().item() / 2e-3
    gs = torch.Generator().manual_seed(5)
    sig1 = equation.max_signalling(equation.partly_entangled(0.3, device), h[1:], gs, device).item() / 1e-3
    # shape test: does E(theta) of a race with m = 1 + delta differ from a damped cosine V cos(theta)?
    theta = torch.linspace(0, math.pi, 91, dtype=torch.float64, device=device)
    w = equation.joint_born(equation.singlet(device), *tensors.planar_pairs(theta, device)).flatten(-2)
    delta = torch.tensor(list(SORKIN_BANDS.values()), dtype=torch.float64, device=device)
    E = equation.correlation(equation.race_probabilities(w.expand(len(delta), -1, -1), 1 + delta[:, None]).view(
        len(delta), -1, 2, 2))  # fmt: skip
    c = -torch.cos(theta)
    V = (E * c).sum(-1) / (c * c).sum()  # least-squares visibility
    shape = (E - V[:, None] * c).abs().amax(-1)  # what no visibility can absorb
    bands = {k: (v, abs(dS) * v, sig1 * v, shape[i].item()) for i, (k, v) in enumerate(SORKIN_BANDS.items())}
    return {"m": m, "S": S, "signal": sig, "dS_dm": dS, "dsignal_dm": sig1, "bands": bands}


def data_table() -> list[dict]:
    """The cited Bell tests with their distance from the local bound 2 and the quantum bound 2 sqrt 2."""
    out = []
    for r in tensors.bell_tests():
        row = dict(r)
        if r["S"]:
            S, s = float(r["S"]), float(r["sigma"])
            row.update(S=S, sigma=s, above_local=(S - 2) / s, below_tsirelson=(TSIRELSON - S) / s,
                       visibility=S / TSIRELSON)  # fmt: skip
        out.append(row)
    return out
```

## 5. `plot.py` — 2D/3D plot → GIF

```python
%%writefile problems/cosmology/projection_bell_test/plot.py
"""Animate: measurement directions sweeping, E(theta) and S(theta) of the models, beyond-Born races, the data."""

import math
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
from PIL import Image

from problems._common.plotting import ACCENT, export_plot_data
from problems._common.progress import report

N_FRAMES = 40
T2 = 2 * math.sqrt(2)
LOCAL_COLORS = ["#ea580c", "#9333ea", "#0891b2"]


def animate(co, rc, bb, rows, D, path: Path) -> Path:
    th = co["theta"].cpu().numpy()
    deg = np.degrees(th)
    fig, axs = plt.subplots(2, 3, figsize=(11.5, 7.0))
    ax_d, ax_e, ax_s = axs[0]
    ax_h, ax_m, ax_x = axs[1]

    # --- measurement directions ---
    circ = np.linspace(0, 2 * np.pi, 200)
    ax_d.plot(np.cos(circ), np.sin(circ), color="#d1d5db", lw=1)
    ax_d.set_aspect("equal")
    ax_d.set_xlim(-1.35, 1.35)
    ax_d.set_ylim(-1.35, 1.35)
    ax_d.set_axis_off()
    arrows = {}
    for name, col in (("a", ACCENT[0]), ("a'", ACCENT[0]), ("b", ACCENT[1]), ("b'", ACCENT[1])):
        line = ax_d.plot([], [], color=col, lw=2.2 if "'" not in name else 1.4, ls="-" if "'" not in name else "--")[0]
        label = ax_d.text(0, 0, name, color=col, fontsize=10, ha="center", va="center")
        arrows[name] = (line, label)
    td = ax_d.set_title("", fontsize=9)

    # --- E(theta) ---
    ax_e.plot(deg, co["E_qm"].cpu(), color="#111827", lw=2.2, label="quantum $-\\cos\\theta$")
    ax_e.plot(np.degrees(co["th_mc"].cpu()), co["E_mc"].cpu(), "o", color=ACCENT[2], ms=4,
              label=f"(i) joint race, $\\Psi\\in\\mathbb{{C}}^{{{D}}}$ (Monte Carlo)")  # fmt: skip
    m_max = co["E_joint"].shape[0]
    if m_max > 1:
        ax_e.plot(deg, co["E_joint"][-1].cpu(), color=ACCENT[1], lw=1.2, ls="--", label=f"joint race $m = {m_max}$")
    for j, (E, lab) in enumerate(zip(co["E_local"].cpu(), ["$m=1$", "$m=2$", "$m\\to\\infty$ (Bell 1964)"])):
        ax_e.plot(deg, E, color=LOCAL_COLORS[j], lw=1.3, label=f"(ii) local race {lab}")
    ax_e.set_xlabel("angle between settings $\\theta$ (degrees)")
    ax_e.set_ylabel("correlation $E(a, b)$ (dimensionless)")
    ax_e.set_xlim(0, 180)
    ax_e.set_ylim(-1.1, 1.1)
    ax_e.legend(fontsize=6, loc="upper left")
    cur_e = ax_e.axvline(0, color="#f59e0b", lw=1.2)
    ax_e.set_title("correlations", fontsize=9)

    # --- S(theta) ---
    ax_s.axhspan(0, 2, color="#fee2e2", lw=0)
    ax_s.axhspan(2, T2, color="#dcfce7", lw=0)
    ax_s.axhline(T2, color="#166534", lw=0.8, ls=":")
    ax_s.plot(deg, co["S_qm"].cpu(), color="#111827", lw=2.2, label="quantum = (i), $m = 1$")
    if m_max > 1:
        ax_s.plot(deg, co["S_joint"][-1].cpu(), color=ACCENT[1], lw=1.2, ls="--", label=f"joint race $m = {m_max}$")
    for j, S in enumerate(co["S_local"].cpu()):
        ax_s.plot(deg, S, color=LOCAL_COLORS[j], lw=1.3)
    ax_s.text(178, 1.9, "local bound 2", fontsize=7, ha="right", va="top", color="#991b1b")
    ax_s.text(178, T2 + 0.04, "Tsirelson $2\\sqrt{2}$", fontsize=7, ha="right", va="bottom", color="#166534")
    ax_s.set_xlim(0, 180)
    ax_s.set_ylim(0, 4)
    ax_s.set_xlabel("$\\theta$ (degrees): $a = 0$, $b = \\theta$, $a' = 2\\theta$, $b' = 3\\theta$")
    ax_s.set_ylabel("CHSH $S$ (dimensionless)")
    ax_s.legend(fontsize=6, loc="upper right")
    dots = ax_s.plot([], [], "o", color="#f59e0b", ms=6)[0]
    ts = ax_s.set_title("", fontsize=9)

    # --- S over random settings ---
    bins = np.linspace(0, 3.0, 61)
    ax_h.hist(rc["S_joint"].cpu().numpy(), bins=bins, color="#111827", alpha=0.6, label="(i) joint, $m=1$")
    for j, S in enumerate(rc["S_local"].cpu().numpy()):
        ax_h.hist(S, bins=bins, histtype="step", color=LOCAL_COLORS[j], lw=1.3, label=f"(ii) local, max {S.max():.3f}")
    ax_h.axvline(2, color="#991b1b", lw=1, ls=":")
    ax_h.axvline(T2, color="#166534", lw=1, ls=":")
    ax_h.set_xlabel("$S$ for random settings $a, a', b, b'$ (dimensionless)")
    ax_h.set_ylabel("settings per bin (count)")
    ax_h.legend(fontsize=6, loc="upper left")
    ax_h.set_title(f"{rc['S_joint'].shape[0]} random settings: local $\\leq 2$, joint $\\leq 2\\sqrt{{2}}$", fontsize=9)

    # --- beyond Born ---
    m = bb["m"].cpu().numpy()
    ax_m.plot(m, bb["S"].cpu(), color=ACCENT[1], lw=2, label="joint race $S$ (optimal settings)")
    ax_m.axhline(T2, color="#166534", lw=0.8, ls=":")
    ax_m.axvline(1, color="#16a34a", lw=0.8)
    ax_m.set_xlabel("hidden complex dimensions per outcome $m$ (dimensionless)")
    ax_m.set_ylabel("CHSH $S$ (dimensionless)", color=ACCENT[1])
    ax_m2 = ax_m.twinx()
    ax_m2.plot(m, bb["signal"].cpu(), color=ACCENT[5], lw=1.5, ls="--", label="signalling (partly entangled)")
    ax_m2.set_ylabel("signalling $\\Delta P_A(+)$ (dimensionless)", color=ACCENT[5])
    ax_m.set_title("$m \\neq 1$: post-quantum $S$ and faster-than-light signals", fontsize=9)
    h1, l1 = ax_m.get_legend_handles_labels()
    h2, l2 = ax_m2.get_legend_handles_labels()
    ax_m.legend(h1 + h2, l1 + l2, fontsize=6, loc="upper left")

    # --- data ---
    ax_x.axhspan(1.6, 2, color="#fee2e2", lw=0)
    ax_x.axhspan(2, T2, color="#dcfce7", lw=0)
    ax_x.axhspan(T2, 3.3, color="#f3f4f6", lw=0)
    ax_x.axhline(2, color="#991b1b", lw=1, ls=":")
    ax_x.axhline(T2, color="#166534", lw=1, ls=":")
    data = [r for r in rows if isinstance(r["S"], float)]
    y = [r["S"] for r in data]
    e = [r["sigma"] for r in data]
    x = np.arange(len(data))
    ax_x.errorbar(x, y, yerr=e, fmt="o", color="#111827", capsize=3, ms=5)
    ax_x.set_xticks(x, [f"{r['experiment'].split(' ')[0]}\n{r['year']}" for r in data], fontsize=7)
    ax_x.text(len(data) - 0.6, 1.65, "(ii) local races: $S \\leq 2$", fontsize=7, ha="right", color="#991b1b")
    ax_x.text(len(data) - 0.6, 2.75, "(i) joint race = QM", fontsize=7, ha="right", color="#166534", va="top")
    ax_x.text(len(data) - 0.6, 3.25, "joint race, $m > 1$", fontsize=7, ha="right", color="#6b7280", va="top")
    ax_x.set_xlim(-0.6, len(data) - 0.4)
    ax_x.set_ylim(1.6, 3.3)
    ax_x.set_ylabel("measured CHSH $S$ (dimensionless)")
    ax_x.set_xlabel("loophole-free experiment")
    others = [r for r in rows if not isinstance(r["S"], float)]
    note = "; ".join(f"{r['experiment'].split(' ')[0]} $p$ = {r['p_value']}" for r in others)
    ax_x.set_title(f"loophole-free tests (CH-Eberhard: {note})", fontsize=7)

    th_f = np.linspace(0, np.pi / 2, N_FRAMES)

    def update(f: int):
        t = th_f[f]
        for name, phi in (("a", 0.0), ("b", t), ("a'", 2 * t), ("b'", 3 * t)):
            line, label = arrows[name]
            line.set_data([0, np.sin(phi)], [0, np.cos(phi)])
            label.set_position((1.18 * np.sin(phi), 1.18 * np.cos(phi)))
        i = int(round(t / np.pi * (len(th) - 1)))
        Sq = co["S_qm"][i].item()
        td.set_text(f"settings in one plane, $\\theta$ = {np.degrees(t):.0f}°\nAlice $a, a'$ (blue), Bob $b, b'$ (red)")
        cur_e.set_xdata([np.degrees(t)])
        ys = [Sq] + [S[i].item() for S in co["S_local"]]
        dots.set_data([np.degrees(t)] * len(ys), ys)
        ts.set_text(f"CHSH: quantum {Sq:.3f}, local {max(ys[1:]):.3f}")

    fig.subplots_adjust(left=0.06, right=0.94, bottom=0.1, top=0.92, wspace=0.6, hspace=0.45)
    return save_gif_shared_palette(fig, update, list(range(N_FRAMES)), path, fps=8)


def save_gif_shared_palette(fig, update, frames: list[int], path: Path, fps: int = 8) -> Path:
    """Like plotting.save_gif, but every frame is quantised to ONE palette built from a strip of frames, so a colour
    never drifts from frame to frame."""
    fig.set_dpi(80)
    shots = []
    for i, f in enumerate(frames):
        update(f)
        fig.canvas.draw()
        shots.append(Image.fromarray(np.asarray(fig.canvas.buffer_rgba())[..., :3]))
        report(i + 1, len(frames), "frames")
    export_plot_data(fig, path)  # numbers behind the final frame (Results page)
    plt.close(fig)
    sample = shots[:: max(1, len(shots) // 12)]  # a strip of frames spanning the run sets the palette
    strip = Image.new("RGB", (sample[0].width, sample[0].height * len(sample)))
    for k, im in enumerate(sample):
        strip.paste(im, (0, k * im.height))
    palette = strip.quantize(colors=255, method=Image.Quantize.MEDIANCUT)
    frames_p = [im.quantize(palette=palette, dither=Image.Dither.NONE) for im in shots]
    tmp = path.with_name(f".{path.stem}.tmp{path.suffix}")
    frames_p[0].save(tmp, save_all=True, append_images=frames_p[1:], duration=int(1000 / fps), loop=0)
    return tmp.replace(path)
```
