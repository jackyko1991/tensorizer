# One particle seen many times: slicing a single worldline

*cosmology · advanced* · [open in Tensorizer](https://jackyko1991.github.io/tensorizer/#/cosmology/projection_one_particle)

A speculative hypothesis, not established physics: every electron is the same object, one worldline that zig-zags back and forth in time (Wheeler's 1940 one-electron universe). A time slice cuts it many times: forward crossings look like electrons, backward ones like positrons. This step slices batches of random worldlines. All the electrons come out exactly identical, and pairs are created and annihilated at the turning points. But a closed worldline crosses every slice as often backwards as forwards, and an open one adds at most one net particle, while the universe holds about $10^{80}$ baryons and almost no antimatter. In its simplest form the idea is ruled out.

**Vector form**

$$X(s) = \big(t(s),\,x_1(s),\dots,x_{D-1}(s)\big),\qquad N_\pm(u) = \#\{s : t(s) = u,\ \pm\,t'(s) > 0\},\qquad Q(u) = N_+ - N_- = [\,t(s_1)\geq u\,] - [\,t(s_0)\geq u\,],\qquad N(u) = 2\,\#\{\text{minima} < u\} - 2\,\#\{\text{maxima} < u\},\qquad \langle N(u)\rangle = 2\sqrt{\lambda_2/\lambda_0}\;e^{-u^2/2\lambda_0},\qquad \frac{|Q|}{N} \leq \frac1N\ \ \text{vs}\ \ \eta = \frac{n_B}{n_\gamma}\approx 6.1\times10^{-10},\ \ 1 - 2f_{e^+}$$

**Index (tensor) form**

$$X_{bsi} = \sigma_k\,a_{bikc}\,\phi_{ksc},\qquad N^+_{bu} = \sum_s [\,t_{bs} < u \leq t_{b,s+1}\,],\qquad N^-_{bu} = \sum_s [\,t_{b,s+1} < u \leq t_{bs}\,],\qquad Q_{bu} = N^+_{bu} - N^-_{bu},\qquad A_E = 1 - 2 f_E,\qquad N_B = \eta\, n_\gamma\, \tfrac43\pi R_H^3$$

| Symbol | Code | Shape | Meaning |
|---|---|---|---|
| $X_{bsi}$ | `einsum('bikc,ksc->bsi', coef, basis)` | (B, S, D) | repeated $k$ (harmonic) and $c$ (cos, sin) ⇒ summed: $B$ random worldlines in $D$ dimensions; coordinate 0 is time |
| $N^\pm_{bu}$ | `crossings(t, levels, closed)` | (B, L) | forward (particle) and backward (antiparticle) crossings of every slice; the sum over segments $s$ is a scatter into a difference array plus a cumulative sum |
| $Q_{bu}$ | `up - down` | (B, L) | net charge on the slice; equals the crossing number of the two ends, $0$ for a closed loop |
| $N(u)$ | `morse_count(t, levels)` | (B, L) | the same count from the turning points alone: minima create pairs, maxima annihilate them |
| $\langle N(u)\rangle$ | `rice_mean_crossings(levels, sigmas)` | (L,) | Rice's formula with spectral moments $\lambda_0 = \sum_k\sigma_k^2$, $\lambda_2 = \sum_k k^2\sigma_k^2$ |
| $A_E$ | `1 - 2 * fraction` | (E,) | charge asymmetry of cosmic-ray $e^\pm$ from the AMS-02 positron fraction $f_E$ at energy $E$ |
| $N_B$ | `baryons_in_observable_universe(eta, n_gamma, R_H)` | () | net baryon number one worldline would have to carry ($\eta$ from Planck 2018) |

Status: speculative; ruled out in its simplest form. Step 2 of the Projection Hypothesis (the author's own speculative idea, see the series description; not established physics), for postulate P3: all particles are projections of one object. Step 1 (cosmology/projection_postulate) stated the postulates; step 3 (cosmology/projection_dimension_budget) counts the dimensions they need. Known physics first. In a 1940 telephone call John Wheeler suggested to Richard Feynman that all electrons are identical because they are one electron, whose worldline runs forwards and backwards in time; a backward stretch is a positron. Feynman told the story in his 1965 Nobel lecture, together with his own objection: there are not as many positrons as electrons. Wheeler suggested they might be hidden in protons. The idea that a positron is an electron going backwards in time survived as the Feynman–Stueckelberg picture of antiparticles (Feynman, Phys. Rev. 76, 749, 1949), but the one-electron universe did not: in quantum field theory all electrons are identical because they are excitations of one field. Here the worldline lives in a space of $D$ dimensions ($D$ is a parameter, never fixed to 4); only its time coordinate matters for the slices, as P2 says. The hypothesis' prediction: a time slice (P2) cuts the worldline at $N_+$ forward and $N_-$ backward crossings. What is computed: batches of smooth (random Fourier, $K$ harmonics, amplitudes $\propto k^{-\beta}$) and rough (Brownian) worldlines, closed or open, cut by $L$ slices. The crossings are counted by a scatter into a difference array and checked against a brute-force table. For a closed loop $N_+ = N_-$ on every slice; for an open one $Q = N_+ - N_-$ equals $[t_{end}\geq u] - [t_{start}\geq u]$, so $|Q|\leq 1$ and $N\equiv Q$ (mod 2). Tilting the slice (a moving observer) does not change $Q$. The total count equals $2(\#\text{minima} - \#\text{maxima})$ below the slice, so pairs are born at minima of $t(s)$ and die at maxima. The mean count of random smooth loops follows Rice's formula. What it gets right for free: every crossing reads the mass and charge of the same worldline, so all electrons are exactly alike, and positrons have exactly the opposite charge and the same mass (PDG 2024: $|m_{e^+} - m_{e^-}|/m < 8\times10^{-9}$, $|q_{e^+} + q_{e^-}|/e < 4\times10^{-8}$). The test against data (data/*.csv, cited numbers): AMS-02 measured the cosmic-ray positron fraction $e^+/(e^+ + e^-)$ from 0.5 GeV to 1 TeV (Aguilar et al., PRL 110, 141102, 2013; PRL 122, 101101, 2019, CC-BY 4.0, 74 bins via the CRDB database): 0.13 at 0.6 GeV, a minimum of 0.053 near 8 GeV and 0.15 to 0.16 above 200 GeV. So the asymmetry $1 - 2f$ is between 0.67 and 0.89, against $0$ for a closed loop. On cosmic scales the Planck 2018 baryon density $\Omega_b h^2 = 0.0224$ gives $\eta = n_B/n_\gamma = 6.1\times10^{-10}$ (see cosmology/baryon_asymmetry). With 411 photons per cm$^3$ in a sphere of radius 46.5 billion light years this is $N_B\approx 9\times10^{79}$ baryons, and antimatter is absent (BESS-Polar: antihelium/helium below $6.9\times10^{-8}$; Cohen, De Rújula and Glashow 1998: no antimatter domains in the visible universe). A single worldline gives at most $|Q|/N = 1/N$, which is $7\times10^{-90}$ when $N\approx 1.5\times10^{89}$, against $\eta = 6\times10^{-10}$; the net charge falls short by a factor of about $10^{80}$. A survey over the open parameters (smooth or Brownian, closed or open, 2 to 32 harmonics, three roughnesses, about $3\times10^5$ slices with $N\geq 2$) finds no configuration that reaches the AMS-02 asymmetry, and the bound shows none can: with $N\geq 2$ crossings the largest possible asymmetry is $1/3$ (at $N = 3$). Verdict: ruled out in its simplest form. Extensions, each an extra assumption and labelled as such: (A) an open worldline adds one net particle, no more. (B) A biased projection, in which observers see antiparticles with efficiency $r$, needs $r = f/(1-f)$ between 0.056 and 0.195, varying with energy, from AMS-02, but $r < 7\times10^{-8}$ from antihelium. The hidden antimatter would then weigh as much as the visible baryons, about 0.19 of the dark matter. This is Wheeler's 'hidden in protons' in modern form; it needs a mechanism, and none is given. (C) If time were a circle, a loop winding $w$ times would carry net charge $Q = w$ on every slice (checked), but $w\approx 10^{80}$ would have to be put in by hand, and compact time means closed timelike curves. None of these is a prediction. What would revive the idea is a mechanism that makes the slice count of a single object asymmetric, and that is the open question this step leaves for the series. Escape route: the simple version is ruled out, but a worldline balanced globally, with the antimatter beyond our horizon or in a CPT-mirror anti-universe, is not. Step 7 (cosmology/projection_antimatter_universe) lists the criteria such an antimatter universe must meet and tests each against data.

## 2. `equation.py` — physics → tensor translation

```python
%%writefile problems/cosmology/projection_one_particle/equation.py
"""One worldline -> tensors: slicing it in time, counting particles and antiparticles, the asymmetry it allows.

Shapes in comments: B worldlines, S samples along each, L time slices (levels u). A segment from t_a to t_b crosses
the slice t = u forwards (a particle) when t_a < u <= t_b and backwards (an antiparticle) when t_b < u <= t_a.
Summed indices: the segment index (counting crossings), the harmonic k in Rice's formula.
"""

import math

import torch


def segments(t: torch.Tensor, closed: bool):
    """Start and end times (B, S') of every straight piece of the sampled worldline; a closed one wraps round."""
    if closed:
        return t, torch.roll(t, -1, dims=1)
    return t[:, :-1], t[:, 1:]


def crossings(t: torch.Tensor, levels: torch.Tensor, closed: bool):
    """(B, L) forward (particle) and backward (antiparticle) crossings of every slice, in O(B S) memory.

    A forward segment crosses the sorted levels with index in [#(levels <= t_a), #(levels <= t_b)); each segment
    adds +1 at the first index and -1 past the last into a difference array, and a cumulative sum over the level
    index gives the counts. Backward segments are handled the same way with t_a and t_b swapped.
    """
    a, b = segments(t, closed)  # (B, S')
    ia = torch.searchsorted(levels, a.contiguous(), right=True)  # (B, S') number of levels <= t_a
    ib = torch.searchsorted(levels, b.contiguous(), right=True)
    L = len(levels)
    out = []
    for lo, hi in ((ia, ib), (ib, ia)):  # forward: a -> b; backward: b -> a
        go = (hi > lo).long()
        diff = torch.zeros(t.shape[0], L + 1, dtype=torch.long, device=t.device)
        diff.scatter_add_(1, lo, go)  # summed over segments
        diff.scatter_add_(1, hi, -go)
        out.append(torch.cumsum(diff, 1)[:, :L])
    return out[0], out[1]


def crossings_reference(t: torch.Tensor, levels: torch.Tensor, closed: bool):
    """Slow reference: the (B, S', L) table of which segment crosses which slice, summed over segments."""
    a, b = segments(t, closed)
    a, b, u = a[..., None], b[..., None], levels
    up = (a < u) & (u <= b)
    down = (b < u) & (u <= a)
    return up.sum(1), down.sum(1)


def crossings_periodic(t: torch.Tensor, levels: torch.Tensor, period: float):
    """Crossings of slices u (0 <= u < period) when time is a circle; t (B, S + 1) is unwrapped and closes up to a
    multiple of the period. Every copy u + j * period of a slice is counted, then summed over j."""
    j = torch.arange(math.floor(t.min() / period) - 1, math.ceil(t.max() / period) + 2, device=t.device)
    ext = (levels[None, :] + period * j[:, None]).reshape(-1)  # (J L,) sorted because 0 <= u < period
    up, down = crossings(t, ext, closed=False)
    return up.view(len(t), len(j), -1).sum(1), down.view(len(t), len(j), -1).sum(1)


def net_charge_from_ends(t: torch.Tensor, levels: torch.Tensor) -> torch.Tensor:
    """(B, L) the crossing number Q(u) = [t_end >= u] - [t_start >= u]: only the ends of the worldline matter."""
    return (t[:, -1:] >= levels).long() - (t[:, :1] >= levels).long()


def turning_points(t: torch.Tensor, closed: bool):
    """(B, S) masks of local minima (a pair is created) and local maxima (a pair annihilates) of t(s)."""
    d_next = torch.roll(t, -1, 1) - t  # slope after sample j
    d_prev = t - torch.roll(t, 1, 1)  # slope before sample j
    births, deaths = (d_prev < 0) & (d_next > 0), (d_prev > 0) & (d_next < 0)
    if not closed:
        births[:, [0, -1]] = False
        deaths[:, [0, -1]] = False
    return births, deaths


def morse_count(t: torch.Tensor, levels: torch.Tensor) -> torch.Tensor:
    """(B, L) crossings of a closed loop from its turning points alone: N(u) = 2 #(minima < u) - 2 #(maxima < u)."""
    births, deaths = turning_points(t, closed=True)
    below = t[..., None] < levels  # (B, S, L)
    return 2 * ((births[..., None] & below).sum(1) - (deaths[..., None] & below).sum(1))


def rice_mean_crossings(levels: torch.Tensor, sigmas: torch.Tensor) -> torch.Tensor:
    """(L,) Rice's formula for a random Fourier loop over s in [0, 2 pi]: 2 sqrt(l2 / l0) exp(-u^2 / 2 l0),
    with spectral moments l0 = sum_k sigma_k^2 and l2 = sum_k k^2 sigma_k^2 (summed over harmonics k)."""
    k = torch.arange(1, len(sigmas) + 1, dtype=sigmas.dtype, device=sigmas.device)
    l0, l2 = (sigmas**2).sum(), (k**2 * sigmas**2).sum()
    return 2 * torch.sqrt(l2 / l0) * torch.exp(-(levels**2) / (2 * l0))


def crossing_properties(t: torch.Tensor, levels: torch.Tensor, props: torch.Tensor) -> torch.Tensor:
    """(n, 2) mass and charge seen at every crossing: both are read off the one worldline (props = (mass, charge));
    a backward crossing shows the opposite charge (Feynman-Stueckelberg)."""
    a, b = segments(t, closed=True)
    a, b = a[..., None], b[..., None]
    up = ((a < levels) & (levels <= b)).nonzero()  # (n_up, 3): worldline, segment, slice
    down = ((b < levels) & (levels <= a)).nonzero()
    direction = torch.cat([torch.ones(len(up)), -torch.ones(len(down))]).to(t)  # (n,) +1 forward, -1 backward
    mass = props[0].expand(len(direction))
    return torch.stack([mass, props[1] * direction], -1)


def max_asymmetry(N: torch.Tensor, closed: bool) -> torch.Tensor:
    """Largest |Q| / N one worldline allows with N crossings: 0 if closed, (N mod 2) / N if open (|Q| <= 1)."""
    if closed:
        return torch.zeros_like(N, dtype=torch.float64)
    return (N % 2).double() / N.double()


def asymmetry_from_fraction(f: torch.Tensor) -> torch.Tensor:
    """Charge asymmetry (N_- - N_+) / (N_- + N_+) = 1 - 2 f for an antiparticle fraction f."""
    return 1 - 2 * f


def eta_from_omega_b(omega_b_h2: float) -> float:
    """Baryon-to-photon ratio eta = 273.9e-10 Omega_b h^2 (Steigman 2006)."""
    return 273.9e-10 * omega_b_h2


def baryons_in_observable_universe(eta: float, n_gamma_cm3: float, radius_gly: float) -> float:
    """N_B = eta n_gamma V for a sphere of comoving radius radius_gly (1 Gly = 9.4607e24 m)."""
    R = radius_gly * 9.4607e24
    return eta * n_gamma_cm3 * 1e6 * 4 / 3 * math.pi * R**3


def efficiency_ratio(f: torch.Tensor) -> torch.Tensor:
    """Extension (biased projection): with equal numbers N_+ = N_-, an observer who sees antiparticles with
    efficiency r relative to particles measures f = r / (1 + r), so r = f / (1 - f)."""
    return f / (1 - f)
```

## 3. `tensors.py` — PyTorch tensors

```python
%%writefile problems/cosmology/projection_one_particle/tensors.py
"""Concrete tensors: batches of worldlines in a space of `dims` dimensions, and the cited data.

A worldline is a path X(s) = (t, x_1, ..., x_{dims-1}) with a parameter s in [0, 2 pi]; t is the observer's time,
the other coordinates are space and any hidden dimensions. Smooth worldlines are random Fourier series (each
coordinate a sum of `modes` harmonics with amplitudes k^-roughness), random ones are Brownian paths. A closed
worldline returns to its start; an open one runs from one time to another.
"""

import csv
import math
from pathlib import Path

import torch

DATA = Path(__file__).parent / "data"


def parameter(S: int, closed: bool, device) -> torch.Tensor:
    """(S,) s values: 2 pi j / S for a closed loop (the last sample joins the first), linspace(0, 2 pi) if open."""
    if closed:
        return 2 * math.pi * torch.arange(S, dtype=torch.float64, device=device) / S
    return torch.linspace(0, 2 * math.pi, S, dtype=torch.float64, device=device)


def mode_sigmas(modes: int, roughness: float, device) -> torch.Tensor:
    """(K,) amplitude of harmonic k = 1..K, sigma_k ~ k^-roughness, normalised so the variance sum sigma_k^2 = 1."""
    k = torch.arange(1, modes + 1, dtype=torch.float64, device=device)
    sig = k ** (-roughness)
    return sig / sig.norm()


def smooth_worldlines(B, S, modes, roughness, dims, closed, generator, device, drift: float = 1.5) -> torch.Tensor:
    """(B, S, dims) random Fourier worldlines X_bsi = sum_k sigma_k (a_bik cos ks + c_bik sin ks).

    Open worldlines also get a time drift +-drift (s / pi - 1), so they start and end at different times.
    """
    s = parameter(S, closed, device)  # (S,)
    sig = mode_sigmas(modes, roughness, device)  # (K,)
    k = torch.arange(1, modes + 1, dtype=torch.float64, device=device)
    coef = torch.randn(B, dims, modes, 2, generator=generator, dtype=torch.float64).to(device) * sig[:, None]
    basis = torch.stack([torch.cos(k[:, None] * s), torch.sin(k[:, None] * s)], -1)  # (K, S, 2)
    X = torch.einsum("bikc,ksc->bsi", coef, basis)  # repeated k (harmonic) and c (cos/sin) => summed
    if not closed:
        sign = torch.randint(0, 2, (B,), generator=generator).to(device) * 2 - 1
        X[..., 0] += drift * sign[:, None] * (s / math.pi - 1)
    return X


def random_worldlines(B, S, dims, closed, generator, device) -> torch.Tensor:
    """(B, S, dims) Brownian worldlines; a closed one is a Brownian bridge, scaled so that the mean variance is 1."""
    steps = torch.randn(B, S, dims, generator=generator, dtype=torch.float64).to(device) / math.sqrt(S)
    W = torch.cumsum(steps, 1)  # (B, S, dims)
    if closed:
        frac = torch.arange(1, S + 1, dtype=torch.float64, device=device) / S
        W = W - frac[None, :, None] * W[:, -1:, :]  # pin the end to the start
        return W * math.sqrt(6.0)  # bridge variance s (1 - s) averages 1/6
    return W * math.sqrt(2.0)  # walk variance s averages 1/2


def worldlines(kind, B, S, modes, roughness, dims, closed, generator, device) -> torch.Tensor:
    """kind 'smooth' (Fourier) or 'random' (Brownian)."""
    if kind == "smooth":
        return smooth_worldlines(B, S, modes, roughness, dims, closed, generator, device)
    return random_worldlines(B, S, dims, closed, generator, device)


def winding_worldlines(w: torch.Tensor, S: int, period: float, wiggle: float, generator, device) -> torch.Tensor:
    """(W, S + 1) unwrapped time of loops that wind w times round a compact time of length `period` (extension).

    t(s) = w period s / 2 pi + wiggle * (periodic Fourier zig-zag); the last sample is the first plus w period,
    so the loop closes on the time circle.
    """
    s = 2 * math.pi * torch.arange(S + 1, dtype=torch.float64, device=device) / S  # includes s = 2 pi
    k = torch.arange(1, 7, dtype=torch.float64, device=device)
    coef = torch.randn(len(w), 6, 2, generator=generator, dtype=torch.float64).to(device) / k[:, None]
    zig = torch.einsum("wkc,ksc->ws", coef, torch.stack([torch.cos(k[:, None] * s), torch.sin(k[:, None] * s)], -1))
    return w.to(device)[:, None] * period * s / (2 * math.pi) + wiggle * zig


def _rows(name: str) -> list[dict]:
    with open(DATA / name, encoding="utf-8") as f:
        return list(csv.DictReader(r for r in f if not r.startswith("#")))


def positron_fraction(device=None) -> dict:
    """AMS-02 (PRL 122, 101101, 2019) positron fraction: E (GeV), bin edges, fraction, stat and syst errors (E,)."""
    rows = _rows("ams02_positron_fraction.csv")
    col = {k: torch.tensor([float(r[k]) for r in rows], dtype=torch.float64, device=device) for k in rows[0]}
    col["sigma"] = torch.sqrt(col["stat"] ** 2 + col["syst"] ** 2)
    return {"E": col["E_GeV"], "E_lo": col["E_lo_GeV"], "E_hi": col["E_hi_GeV"], "fraction": col["fraction"],
            "sigma": col["sigma"]}  # fmt: skip


def cited() -> dict:
    """Cited numbers (data/matter_antimatter.csv): quantity -> value."""
    return {r["quantity"]: float(r["value"]) for r in _rows("matter_antimatter.csv")}
```

## 4. `solve.py` — solver

```python
%%writefile problems/cosmology/projection_one_particle/solve.py
"""Solvers: slice a batch of worldlines, survey the open parameters, compare with the observed asymmetry."""

import torch

from . import equation, tensors


def slice_counts(X: torch.Tensor, levels: torch.Tensor, closed: bool) -> dict:
    """Particles (forward crossings), antiparticles (backward), net charge Q and total N on every slice (B, L)."""
    up, down = equation.crossings(X[..., 0], levels, closed)
    return {"up": up, "down": down, "Q": up - down, "N": up + down}


def batch(B, S, modes, roughness, dims, levels, generator, device) -> dict:
    """Closed smooth loops: mean counts per slice against Rice's formula, and the Morse (turning point) count."""
    X = tensors.smooth_worldlines(B, S, modes, roughness, dims, True, generator, device)
    c = slice_counts(X, levels, closed=True)
    rice = equation.rice_mean_crossings(levels, tensors.mode_sigmas(modes, roughness, device))
    morse_ok = torch.equal(c["N"], equation.morse_count(X[..., 0], levels))
    return {**c, "rice": rice, "mean_N": c["N"].double().mean(0), "sem_N": c["N"].double().std(0) / B**0.5,
            "morse_ok": morse_ok, "max_abs_Q": c["Q"].abs().max().item()}  # fmt: skip


def survey(B, S, dims, levels, A_target: float, generator, device) -> list[dict]:
    """Which worldline configurations give an asymmetry |Q| / N >= A_target on a slice with N >= 2 crossings?

    The open parameters are scanned: smooth (Fourier) or random (Brownian) worldlines, closed or open, the number of
    harmonics and the roughness. The loop runs over these modelling choices; each case is a batch of B worldlines.
    """
    cases = [("random", c, 0, 0.0) for c in (True, False)]
    cases += [("smooth", c, k, r) for c in (True, False) for k in (2, 4, 8, 16, 32) for r in (0.5, 1.0, 2.0)]
    rows = []
    for kind, closed, modes, rough in cases:
        X = tensors.worldlines(kind, B, S, max(modes, 1), rough, dims, closed, generator, device)
        c = slice_counts(X, levels, closed)
        many = c["N"] >= 2  # slices that cut the worldline at least twice
        A = c["Q"].abs().double() / c["N"].clamp_min(1).double()
        ok = many & (A >= A_target)
        rows.append({"kind": kind, "closed": closed, "modes": modes, "roughness": rough,
                     "slices": int(many.sum()), "reproduce": int(ok.sum()),
                     "max_A": A[many].max().item() if many.any() else 0.0,
                     "max_N": int(c["N"].max()), "max_abs_Q": int(c["Q"].abs().max())})  # fmt: skip
    return rows


def representative(S, modes, roughness, dims, levels, device, seed: int = 14) -> dict:
    """One closed loop for the picture: the worldline, its slices, and its turning points."""
    g = torch.Generator().manual_seed(seed)
    X = tensors.smooth_worldlines(1, S, modes, roughness, dims, True, g, device)
    c = slice_counts(X, levels, closed=True)
    births, deaths = equation.turning_points(X[..., 0], closed=True)
    return {"X": X[0], **{k: v[0] for k, v in c.items()}, "births": births[0], "deaths": deaths[0]}


def observations(device) -> dict:
    """The observed asymmetry: AMS-02 positron fraction, eta from Planck, the baryon count, BESS antihelium."""
    pf = tensors.positron_fraction(device)
    c = tensors.cited()
    eta = equation.eta_from_omega_b(c["omega_b_h2"])
    N_B = equation.baryons_in_observable_universe(eta, c["n_gamma_cm3"], c["hubble_radius_gly"])
    A = equation.asymmetry_from_fraction(pf["fraction"])
    r = equation.efficiency_ratio(pf["fraction"])
    return {**pf, "A": A, "A_min": A.min().item(), "A_max": A.max().item(), "r": r, "eta": eta, "N_B": N_B,
            "N_gamma": N_B / eta, "cited": c,
            "hidden_over_dm": c["omega_b_h2"] / c["omega_c_h2"]}  # fmt: skip


def winding_demo(device, generator) -> dict:
    """Extension: loops that wind w times round a compact time carry net charge w on every slice."""
    w = torch.tensor([1.0, 2.0, 3.0, 5.0], dtype=torch.float64)
    t = tensors.winding_worldlines(w, 2000, 1.0, 0.3, generator, device)
    levels = torch.linspace(0.05, 0.95, 19, dtype=torch.float64, device=device)
    up, down = equation.crossings_periodic(t, levels, 1.0)
    return {"w": w, "Q": up - down, "N": up + down}
```

## 5. `plot.py` — 2D/3D plot → GIF

```python
%%writefile problems/cosmology/projection_one_particle/plot.py
"""Animate a time slice sweeping one zig-zagging worldline, the counters, and the asymmetry the data demand."""

from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.collections import LineCollection
from PIL import Image

from problems._common.plotting import export_plot_data, frame_indices
from problems._common.progress import report

BLUE, RED, GREEN = "#2563eb", "#dc2626", "#16a34a"


def _cuts(t: np.ndarray, x: np.ndarray, u: float):
    """x positions where the closed loop crosses t = u, split into forward and backward crossings."""
    a, b = t, np.roll(t, -1)
    xa, xb = x, np.roll(x, -1)
    up = (a < u) & (u <= b)
    down = (b < u) & (u <= a)
    frac = (u - a) / np.where(b != a, b - a, 1)
    xc = xa + frac * (xb - xa)
    return xc[up], xc[down]


def animate(rep, bt, ob, levels, dims, path: Path) -> Path:
    X = rep["X"].cpu().numpy()
    t, x = X[:, 0], X[:, 1]
    lv = levels.cpu().numpy()
    up_r, down_r = rep["up"].cpu().numpy(), rep["down"].cpu().numpy()

    fig = plt.figure(figsize=(10.5, 6.8))
    ax_w, ax_c = fig.add_subplot(2, 2, 1), fig.add_subplot(2, 2, 2)
    ax_f, ax_a = fig.add_subplot(2, 2, 3), fig.add_subplot(2, 2, 4)

    # --- the worldline, coloured by the direction it runs in time ---
    pts = np.stack([x, t], 1)
    seg = np.stack([pts, np.roll(pts, -1, 0)], 1)
    fwd = np.roll(t, -1) > t
    ax_w.add_collection(LineCollection(seg, colors=np.where(fwd, BLUE, RED), linewidths=1.6))
    births, deaths = rep["births"].cpu().numpy(), rep["deaths"].cpu().numpy()
    ax_w.plot(x[births], t[births], "^", color=GREEN, ms=7, label="pair creation (minimum of $t$)")
    ax_w.plot(x[deaths], t[deaths], "v", color="#111827", ms=7, label="annihilation (maximum of $t$)")
    pad = 0.15 * (x.max() - x.min())
    ax_w.set_xlim(x.min() - pad, x.max() + pad)
    ax_w.set_ylim(lv[0], lv[-1])
    ax_w.set_xlabel("space $x$, one axis of the projection (a.u.)")
    ax_w.set_ylabel("time $t$ (a.u.)")
    ax_w.legend(fontsize=7, loc="upper left")
    slice_line = ax_w.axhline(lv[0], color="#f59e0b", lw=2)
    dots_p = ax_w.plot([], [], "o", color=BLUE, ms=8, mec="white")[0]
    dots_a = ax_w.plot([], [], "o", color=RED, ms=8, mec="white")[0]
    counter = ax_w.text(0.98, 0.03, "", transform=ax_w.transAxes, ha="right", fontsize=9,
                        bbox={"facecolor": "white", "edgecolor": "#d1d5db"})  # fmt: skip
    ax_w.set_title(f"one closed worldline in $D = {dims}$, seen in $(x, t)$:\nblue forward = particle, "
                   "red backward = antiparticle", fontsize=9)  # fmt: skip

    # --- counters per slice ---
    ax_c.step(lv, up_r, where="mid", color=BLUE, lw=1.6, label="particles $N_+$ (this loop)")
    ax_c.step(lv, down_r + 0.06, where="mid", color=RED, lw=1.2, ls="--", label="antiparticles $N_-$ (this loop)")
    ax_c.plot(lv, bt["mean_N"].cpu().numpy() / 2, "o", color="#6b7280", ms=3, label="batch mean $N/2$")
    ax_c.plot(lv, bt["rice"].cpu().numpy() / 2, color="#111827", lw=1, label="Rice formula / 2")
    ax_c.set_xlabel("slice time $u$ (a.u.)")
    ax_c.set_ylabel(r"count on the slice $N_\pm$ (dimensionless)")
    ax_c.legend(fontsize=7, loc="upper left")
    ax_c.set_ylim(0, max(up_r.max(), bt["mean_N"].max().item() / 2) * 1.35 + 0.5)
    cursor = ax_c.axvline(lv[0], color="#f59e0b", lw=1.5)
    ax_c.set_title(f"$N_+ = N_-$ on every slice of every closed loop (max $|Q|$ = {bt['max_abs_Q']})", fontsize=9)

    # --- AMS-02 positron fraction ---
    E, f, sg = ob["E"].cpu().numpy(), ob["fraction"].cpu().numpy(), ob["sigma"].cpu().numpy()
    ax_f.errorbar(E, f, yerr=sg, fmt="o", ms=3, color=BLUE, ecolor="#93c5fd", label="AMS-02 2019 (cited data)")
    ax_f.axhline(0.5, color=RED, lw=2, label="one closed worldline: $f = 1/2$")
    ax_f.set_xscale("log")
    ax_f.set_ylim(0, 0.6)
    ax_f.set_xlabel("kinetic energy $E$ (GeV)")
    ax_f.set_ylabel("positron fraction $f = e^+/(e^+ + e^-)$\n(dimensionless)")
    ax_f.legend(fontsize=7, loc="center left")
    ax_f.set_title("cosmic rays: positrons are only 5 to 16% of $e^\\pm$", fontsize=9)

    # --- asymmetry: what one worldline allows against what is seen ---
    N = np.logspace(0, 90, 200)
    ax_a.loglog(N, 1 / N, color=RED, lw=2, label="one open worldline: $|Q|/N \\leq 1/N$")
    ax_a.fill_between(N, 1e-95, 1 / N, color="#fecaca", alpha=0.6, lw=0)
    ax_a.axhspan(ob["A_min"], ob["A_max"], color="#bfdbfe", alpha=0.8, lw=0)
    ax_a.text(1e40, 1e-6, "blue band: cosmic-ray $e^\\pm$ (AMS-02)", fontsize=7, color="#1e40af")
    ax_a.plot(
        [ob["N_gamma"]], [ob["eta"]], "*", color="#111827", ms=12, label=f"early universe: $\\eta$ = {ob['eta']:.1e}"
    )
    ax_a.plot([ob["N_B"]], [1.0], "s", color="#9333ea", ms=8, label=f"today: {ob['N_B']:.0e} baryons, no antimatter")
    ax_a.set_xlim(1, 1e91)
    ax_a.set_ylim(1e-92, 10)
    ax_a.set_xlabel("crossings on a slice $N$ (dimensionless)")
    ax_a.set_ylabel("asymmetry $|N_+ - N_-|/N$ (dimensionless)")
    ax_a.legend(fontsize=7, loc="lower left")
    ax_a.set_title(f"net charge short by {ob['N_B']:.0e}: ruled out in its simplest form", fontsize=9)

    n = len(lv)
    sweep = list(range(n)) + list(range(n - 2, 0, -1))
    sweep = [sweep[i] for i in frame_indices(len(sweep), 64)]

    def update(i: int):
        u = lv[i]
        slice_line.set_ydata([u])
        cursor.set_xdata([u])
        xp, xa = _cuts(t, x, u)
        dots_p.set_data(xp, np.full_like(xp, u))
        dots_a.set_data(xa, np.full_like(xa, u))
        counter.set_text(f"particles {len(xp)}   antiparticles {len(xa)}   net {len(xp) - len(xa)}")

    fig.subplots_adjust(left=0.07, right=0.98, bottom=0.08, top=0.92, wspace=0.25, hspace=0.42)
    return save_gif_shared_palette(fig, update, sweep, path, fps=8)


def save_gif_shared_palette(fig, update, frames: list[int], path: Path, fps: int = 8) -> Path:
    """Like plotting.save_gif, but every frame uses ONE palette built from a strip of all frames, so colours stay
    fixed through the animation (Pillow otherwise picks a new 256-colour palette per frame)."""
    fig.set_dpi(80)
    shots = []
    for i, f in enumerate(frames):
        update(f)
        fig.canvas.draw()
        shots.append(Image.fromarray(np.asarray(fig.canvas.buffer_rgba())[..., :3]))
        report(i + 1, len(frames), "frames")
    export_plot_data(fig, path)  # numbers behind the final frame (Results page)
    plt.close(fig)
    sample = shots[:: max(1, len(shots) // 12)]
    strip = Image.new("RGB", (sample[0].width, sample[0].height * len(sample)))
    for k, im in enumerate(sample):
        strip.paste(im, (0, k * im.height))
    palette = strip.quantize(colors=255, method=Image.Quantize.MEDIANCUT)
    frames_p = [im.quantize(palette=palette, dither=Image.Dither.NONE) for im in shots]
    tmp = path.with_name(f".{path.stem}.tmp{path.suffix}")
    frames_p[0].save(tmp, save_all=True, append_images=frames_p[1:], duration=int(1000 / fps), loop=0)
    return tmp.replace(path)
```
