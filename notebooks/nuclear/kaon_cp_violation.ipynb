# Time reversal and CPT with neutral kaons

*nuclear · advanced* · [open in Tensorizer](https://jackyko1991.github.io/tensorizer/#/nuclear/kaon_cp_violation)

A neutral kaon is a two-state system ($K^0$, $\bar K^0$) that oscillates and decays, so it evolves under a non-Hermitian $2\times 2$ Hamiltonian $H = M - i\Gamma/2$. Built from PDG values and exponentiated for a batch of worlds at every time at once, it shows strangeness oscillating at $\Delta m$, the CP-violating $\pi\pi$ interference, and the CPLEAR test of time reversal: $K^0 \to \bar K^0$ and $\bar K^0 \to K^0$ differ by $A_T \approx 4\,\mathrm{Re}\,\varepsilon \approx 6.5\times 10^{-3}$, while CPT keeps $K^0$ and $\bar K^0$ equal in mass and lifetime.

**Vector form**

$$i\frac{d}{dt}\begin{pmatrix}a\\ \bar a\end{pmatrix} = \Big(M - \tfrac{i}{2}\Gamma\Big)\begin{pmatrix}a\\ \bar a\end{pmatrix},\qquad A_T = \frac{P(\bar K^0\to K^0) - P(K^0\to\bar K^0)}{P(\bar K^0\to K^0) + P(K^0\to\bar K^0)} \approx 4\,\mathrm{Re}\,\varepsilon$$

**Index (tensor) form**

$$H_{ab} = V_{ai}\,\lambda_i\,[V^{-1}]_{ib},\qquad U_{tab} = V_{ai}\,e^{-i\lambda_i t}\,[V^{-1}]_{ib},\qquad P_{tab} = |U_{tab}|^2,\qquad R_{tb} = \big|\,a_x U_{txb}\big|^2$$

| Symbol | Code | Shape | Meaning |
|---|---|---|---|
| $V_{ai}$ | `eigenvectors(eps, delta)` | (B, 2, 2) complex | columns $K_S \propto (1+\varepsilon_S, 1-\varepsilon_S)$, $K_L \propto (1+\varepsilon_L, -(1-\varepsilon_L))$ with $\varepsilon_{S,L} = \varepsilon \pm \delta$ |
| $\lambda_i$ | `eigenvalues(dm, gamma_s, gamma_l)` | (2,) complex | $\mp\Delta m/2 - i\Gamma_{S,L}/2$ in units of $\hbar/\tau_S$ (the common mass is a global phase) |
| $H_{ab}$ | `einsum('...ai,i,...ib->...ab', v, lam, inv(v))` | (B, 2, 2) complex | repeated $i$ ⇒ summed: one effective Hamiltonian per world |
| $U_{tab}$ | `einsum('bxi,bti,biy->btxy', v, phase, inv(v))` | (B, T, 2, 2) complex | $e^{-iHt}$ at every time, checked against matrix_exp |
| $P_{tab}$ | `u.abs() ** 2` | (B, T, 2, 2) | probability that a kaon born as $b$ is found as $a$ ($0 = K^0$, $1 = \bar K^0$) |
| $\Gamma = i(H - H^\dagger)$ | `decay_matrix(h)` | (2, 2) complex | Hermitian and $\ge 0$: unitarity, the Bell–Steinberger relation |
| $a_x$ | `pipi_amplitude(v, eta)` | (B, 2) complex | row vector with $a\cdot K_S = 1$, $a\cdot K_L = \eta_{+-}$ |
| $R_{tb}$ | `einsum('bx,btxy->bty', a, u).abs() ** 2` | (B, T, 2) | repeated $x$ ⇒ summed: $\pi^+\pi^-$ decay rate of a beam born as $K^0$ or $\bar K^0$ |

C (charge conjugation) swaps particles and antiparticles, P (parity) mirrors space and T reverses time. The CPT theorem (Lüders, Pauli, Schwinger) says any local, Lorentz-invariant quantum field theory is symmetric under the three together, so a particle and its antiparticle have exactly equal masses and lifetimes. In 1964 Christenson, Cronin, Fitch and Turlay saw the long-lived $K_L$ decay to two pions, which CP forbids: CP is violated, at the level $|\varepsilon| \approx 2.2\times 10^{-3}$ (Nobel Prize 1980 to Cronin and Fitch). If CPT holds, CP violation means T violation too, and CPLEAR at CERN saw it directly in 1998: starting from tagged $K^0$ and $\bar K^0$ and tagging the strangeness at decay with semileptonic decays, $\bar K^0 \to K^0$ happens slightly more often than $K^0 \to \bar K^0$, $\langle A_T\rangle = (6.6 \pm 1.3 \pm 1.0)\times 10^{-3}$ (the interpretation as T violation assumes the $\Delta S = \Delta Q$ rule and no CPT violation in the decays). Here $H$ is built from its eigenstates, so the CP parameter $\varepsilon$ and the CPT parameter $\delta$ are dials: with $\delta = 0$, $H_{11} = H_{22}$ (equal masses and lifetimes), $P(K^0\to K^0) = P(\bar K^0\to\bar K^0)$ and $A_T$ is a flat plateau at $(|p/q|^2 - |q/p|^2)/(|p/q|^2 + |q/p|^2) \approx 4\,\mathrm{Re}\,\varepsilon$; a nonzero $\mathrm{Re}\,\delta$ makes $A_{CPT} = [P(\bar K^0\to\bar K^0) - P(K^0\to K^0)]/[\ldots]$ climb to $4\,\mathrm{Re}\,\delta$ after a few $\tau_S$ (CPLEAR found $\mathrm{Re}\,\delta$ consistent with zero at the $10^{-4}$ level). The bottom panel shows the classic $\pi^+\pi^-$ interference between $K_S$ and $K_L$ decays, largest near $12\,\tau_S$, from which $|\eta_{+-}|$ and $\phi_{+-}$ are measured. Unitarity is checked with the Bell–Steinberger relation $(i\Delta m + \bar\Gamma)\langle K_S|K_L\rangle = \sum_f A_S^* A_L$; with $\pi\pi$ dominance the measured $\eta_{+-}$ predicts $\mathrm{Re}\,\varepsilon$ to 0.2%. Values: PDG 2024 ($\tau_S = 0.8954\times10^{-10}$ s, $\tau_L = 5.116\times10^{-8}$ s, $\Delta m = 0.5293\times10^{10}\,\hbar$/s, $|\varepsilon| = 2.228\times 10^{-3}$, $\phi_\varepsilon = 43.52°$, $|\eta_{+-}| = 2.232\times10^{-3}$); CPLEAR, Phys. Lett. B 444 (1998) 43. Simplifications: direct CP violation ($\varepsilon'$, $10^{-3}$ of $\varepsilon$) is neglected so $\eta_{+-} = \varepsilon$, the semileptonic tagging is idealised as a perfect strangeness measurement, and no regeneration in matter.

## 1. `main.py` — main loop

```python
%%writefile problems/nuclear/kaon_cp_violation/main.py
"""Main loop: parameters → tensors → solve → plot."""

from pathlib import Path

import torch

from problems._common.runtime import cli, device

from . import equation, plot, solve, tensors


def run(params: dict, out_dir: Path) -> Path:
    dev = device()
    eps = tensors.epsilon(params["eps_abs"] * 1e-3, params["eps_phase"])
    delta = complex(params["re_delta"] * 1e-4, 0.0)
    dm = params["dm"] * 1e10 * tensors.TAU_S  # ħ/s → ħ/τ_S
    t = tensors.times(params["t_max"], 600, dev)  # (T,) in units of τ_S
    e, d = tensors.scenario_parameters(eps, delta, dev)  # (3,): chosen, δ = 0, ε = δ = 0

    out = solve.simulate(dm, tensors.GAMMA_S, tensors.GAMMA_L, e, d, t)
    h = out["h"]
    print(f"Δm τ_S = {dm:.4f},  Γ_L/Γ_S = {tensors.GAMMA_L:.3e},  ε = {abs(eps):.3e} ∠ {params['eps_phase']:.1f}°")
    print(f"H (K⁰, K̄⁰ basis, units ħ/τ_S, common mass removed):\n{h[0].cpu().numpy().round(5)}")
    print(f"CPT check: H₁₁ − H₂₂ = {complex((h[0, 0, 0] - h[0, 1, 1]).item()):.2e}  (0 when δ = 0)")
    gamma = equation.decay_matrix(h[0])
    print(f"unitarity: eigenvalues of Γ = i(H − H†) = {torch.linalg.eigvalsh(gamma).cpu().numpy().round(6)} (≥ 0)")
    v = equation.eigenvectors(e[0], d[0])
    overlap = torch.vdot(v[:, 0], v[:, 1])
    lhs = (1j * dm + (tensors.GAMMA_S + tensors.GAMMA_L) / 2) * overlap
    rhs = torch.vdot(v[:, 0], gamma @ v[:, 1])
    print(f"Bell–Steinberger: ⟨K_S|K_L⟩ = {complex(overlap.item()):.3e},  |LHS − RHS| = {abs(lhs - rhs).item():.1e}")
    a_t = out["a_t"][0, 1:].mean().item()
    print(f"Kabir T asymmetry A_T = {a_t:.4e}   (4 Re ε = {4 * eps.real:.4e};  CPLEAR 1998: (6.6 ± 1.6) × 10⁻³)")
    a_cpt = out["a_cpt"][0, -1].item()
    print(f"CPT asymmetry at t = {params['t_max']:g} τ_S: {a_cpt:.3e}   (4 Re δ = {4 * delta.real:.1e})")
    print(f"no CP violation (ε = δ = 0): max |A_T| = {out['a_t'][2, 1:].abs().max().item():.1e}")
    return plot.animate(t, out, (eps, delta), out_dir / "kaon_cp_violation.gif")


if __name__ == "__main__":
    cli(run, __file__)
```

## 2. `equation.py` — physics → tensor translation

```python
%%writefile problems/nuclear/kaon_cp_violation/equation.py
r"""Physics → tensor translation.

A neutral kaon is a two-state system in the strangeness basis (K⁰, K̄⁰). Because it decays, its evolution is
generated by an effective non-Hermitian Hamiltonian

    i dψ/dt = H ψ,      H = M − iΓ/2,      M = M†, Γ = Γ† ≥ 0      (2 × 2 complex).

Its eigenstates are the short- and long-lived kaons K_S and K_L with eigenvalues λ = m − iΓ/2. In the
CPLEAR convention, with ε measuring CP violation and δ CPT violation (ε_S = ε + δ, ε_L = ε − δ),

    K_S ∝ (1 + ε_S) K⁰ + (1 − ε_S) K̄⁰,      K_L ∝ (1 + ε_L) K⁰ − (1 − ε_L) K̄⁰,

so H = V Λ V⁻¹ with V = [K_S K_L] and Λ = diag(λ_S, λ_L), and the evolution operator is

    U_{ab}(t) = V_{ai} e^{−iλ_i t} [V⁻¹]_{ib}      (i summed),      P(b → a, t) = |U_{ab}(t)|².

The common mass m_K only adds a global phase and is left out (λ_{S,L} = ∓Δm/2 − iΓ_{S,L}/2).
"""

import torch


def eigenvectors(eps: torch.Tensor, delta: torch.Tensor) -> torch.Tensor:
    r"""V_{ai}: columns K_S, K_L in the (K⁰, K̄⁰) basis, normalised.   eps, delta: (...) → (..., 2, 2) complex"""
    eps, delta = (torch.as_tensor(x, dtype=torch.complex128) for x in (eps, delta))
    e_s, e_l = eps + delta, eps - delta
    k_s = torch.stack([1 + e_s, 1 - e_s], dim=-1)  # (..., 2)
    k_l = torch.stack([1 + e_l, -(1 - e_l)], dim=-1)
    v = torch.stack([k_s, k_l], dim=-1)  # (..., a, i)
    return v / torch.linalg.vector_norm(v, dim=-2, keepdim=True)


def eigenvalues(dm: float, gamma_s: float, gamma_l: float) -> torch.Tensor:
    r"""λ_i = m_i − iΓ_i/2 for i = S, L (common mass removed): (2,) complex"""
    return torch.tensor([-dm / 2 - 0.5j * gamma_s, dm / 2 - 0.5j * gamma_l], dtype=torch.complex128)


def hamiltonian(dm: float, gamma_s: float, gamma_l: float, eps: torch.Tensor, delta: torch.Tensor) -> torch.Tensor:
    r"""H_{ab} = V_{ai} λ_i [V⁻¹]_{ib}   (i summed).   → (..., 2, 2) complex, one H per (ε, δ)"""
    v = eigenvectors(eps, delta)
    lam = eigenvalues(dm, gamma_s, gamma_l).to(v.device)
    return torch.einsum("...ai,i,...ib->...ab", v, lam, torch.linalg.inv(v))


def evolution(h: torch.Tensor, t: torch.Tensor) -> torch.Tensor:
    r"""U_{tab} = V_{ai} e^{−iλ_i t} [V⁻¹]_{ib} = [e^{−iHt}]_{ab}.   h: (B, 2, 2), t: (T,) → (B, T, 2, 2)"""
    lam, v = torch.linalg.eig(h)  # (B, 2), (B, 2, 2)
    phase = torch.exp(-1j * lam[:, None, :] * t[None, :, None])  # (B, T, i)
    return torch.einsum("bxi,bti,biy->btxy", v, phase, torch.linalg.inv(v))


def probabilities(u: torch.Tensor) -> torch.Tensor:
    r"""P(b → a, t) = |U_{ab}(t)|²: index [..., t, a (final), b (initial)], 0 = K⁰, 1 = K̄⁰."""
    return u.abs() ** 2


def decay_matrix(h: torch.Tensor) -> torch.Tensor:
    r"""Γ = i(H − H†): Hermitian and positive semi-definite if probability only leaks out (unitarity)."""
    return 1j * (h - h.conj().transpose(-1, -2))


def kabir_asymmetry(p: torch.Tensor) -> torch.Tensor:
    r"""A_T(t) = [P(K̄⁰→K⁰) − P(K⁰→K̄⁰)] / [P(K̄⁰→K⁰) + P(K⁰→K̄⁰)]: T violation (≈ 4 Re ε)."""
    forward, backward = p[..., 0, 1], p[..., 1, 0]
    return (forward - backward) / (forward + backward)


def cpt_asymmetry(p: torch.Tensor) -> torch.Tensor:
    r"""A_CPT(t) = [P(K̄⁰→K̄⁰) − P(K⁰→K⁰)] / [P(K̄⁰→K̄⁰) + P(K⁰→K⁰)]: zero whenever CPT holds."""
    anti, kaon = p[..., 1, 1], p[..., 0, 0]
    return (anti - kaon) / (anti + kaon)


def flavour_asymmetry(p: torch.Tensor) -> torch.Tensor:
    r"""[P(K⁰→K⁰) − P(K⁰→K̄⁰)] / [sum] = 2 e^{−Γ̄t} cos(Δm t) / (e^{−Γ_S t} + e^{−Γ_L t}) without CP violation."""
    same, flipped = p[..., 0, 0], p[..., 1, 0]
    return (same - flipped) / (same + flipped)


def pipi_amplitude(v: torch.Tensor, eta: torch.Tensor) -> torch.Tensor:
    r"""Row vector a_b with a·K_S = 1 and a·K_L = η₊₋:  a_b = (1, η)_i [V⁻¹]_{ib}.   → (B, 2) complex"""
    eta = torch.as_tensor(eta, dtype=torch.complex128, device=v.device)
    ones = torch.ones_like(eta)
    return torch.einsum("bi,bix->bx", torch.stack([ones, eta], dim=-1), torch.linalg.inv(v))
```

## 3. `tensors.py` — PyTorch tensors

```python
%%writefile problems/nuclear/kaon_cp_violation/tensors.py
"""Build the concrete tensors: PDG constants of the neutral kaons (in units of τ_S and ħ = 1) and the time grid.

PDG 2024 (Review of Particle Physics, Phys. Rev. D 110, 030001):
    τ_S = 0.8954 × 10⁻¹⁰ s,  τ_L = 5.116 × 10⁻⁸ s,  Δm = m_L − m_S = 0.5293 × 10¹⁰ ħ s⁻¹,
    |ε| = 2.228 × 10⁻³, φ_ε = 43.52°,  |η₊₋| = 2.232 × 10⁻³, φ₊₋ = 43.4°,  m_K = 497.611 MeV.
CPLEAR 1998 (Phys. Lett. B 444, 43): Kabir asymmetry ⟨A_T⟩ = (6.6 ± 1.3 stat ± 1.0 syst) × 10⁻³.
"""

import cmath
import math

import torch

TAU_S, TAU_L = 0.8954e-10, 5.116e-8  # s
DM_PER_S = 0.5293e10  # Δm in ħ/s
GAMMA_S = 1.0  # time unit τ_S, so Γ_S = 1
GAMMA_L = TAU_S / TAU_L  # ≈ 1.75 × 10⁻³
DM = DM_PER_S * TAU_S  # ≈ 0.474: Δm in units of ħ/τ_S
EPS_ABS, EPS_PHASE = 2.228e-3, 43.52  # |ε|, φ_ε in degrees
ETA_PM_ABS, ETA_PM_PHASE = 2.232e-3, 43.4  # |η₊₋|, φ₊₋ in degrees
M_K = 497.611  # MeV, the common mass (only differences enter the evolution)
CPLEAR_AT = (6.6e-3, math.hypot(1.3e-3, 1.0e-3))  # value, total error


def epsilon(modulus: float, phase_deg: float) -> complex:
    """A complex CP parameter from its modulus and phase in degrees."""
    return cmath.rect(modulus, math.radians(phase_deg))


def times(t_max: float, n: int, device: torch.device) -> torch.Tensor:
    """Proper times 0 … t_max in units of τ_S: shape (T,)."""
    return torch.linspace(0.0, t_max, n, dtype=torch.float64, device=device)


def scenario_parameters(eps: complex, delta: complex, device: torch.device) -> tuple[torch.Tensor, torch.Tensor]:
    """Three kaon worlds side by side: the chosen (ε, δ), CPT conserved (δ = 0) and no CP violation at all.

    Returns ε and δ, each (3,) complex, so every later tensor carries a leading batch axis B = 3.
    """
    e = torch.tensor([eps, eps, 0.0], dtype=torch.complex128, device=device)
    d = torch.tensor([delta, 0.0, 0.0], dtype=torch.complex128, device=device)
    return e, d
```

## 4. `solve.py` — solver

```python
%%writefile problems/nuclear/kaon_cp_violation/solve.py
"""Evolve every kaon world on the whole time grid at once and read off the observables."""

import torch

from . import equation


def pipi_rate(h: torch.Tensor, v: torch.Tensor, eta: torch.Tensor, t: torch.Tensor) -> torch.Tensor:
    """Decay rate to π⁺π⁻ (in units of Γ_S(ππ)) for a beam that was pure K⁰ or pure K̄⁰ at t = 0.

    h: (B, 2, 2), v: (B, 2, 2), eta: (B,), t: (T,) → (B, T, 2), last axis = starting K⁰, K̄⁰.
    """
    a = equation.pipi_amplitude(v, eta)  # (B, 2)
    u = equation.evolution(h, t)  # (B, T, 2, 2)
    return torch.einsum("bx,btxy->bty", a, u).abs() ** 2  # x summed: project the state onto ππ


def simulate(dm: float, gamma_s: float, gamma_l: float, eps: torch.Tensor, delta: torch.Tensor, t: torch.Tensor):
    """All observables for a batch of (ε, δ) worlds.  eps, delta: (B,) → dict of (B, T, ...) tensors."""
    h = equation.hamiltonian(dm, gamma_s, gamma_l, eps, delta)  # (B, 2, 2)
    v = equation.eigenvectors(eps, delta)
    p = equation.probabilities(equation.evolution(h, t))  # (B, T, final, initial)
    return {
        "h": h,
        "p": p,
        "a_t": equation.kabir_asymmetry(p),
        "a_cpt": equation.cpt_asymmetry(p),
        "pipi": pipi_rate(h, v, eps, t),  # η₊₋ = ε (direct CP violation ε' neglected)
    }
```

## 5. `plot.py` — 2D/3D plot → GIF

```python
%%writefile problems/nuclear/kaon_cp_violation/plot.py
"""Top: what a K⁰ beam is made of as it flies. Middle: the T-violating asymmetry plateau. Bottom: ππ interference."""

from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import torch

from problems._common.plotting import frame_indices, save_gif

from . import tensors

BLUE, RED, GREY, AMBER = "#2563eb", "#dc2626", "#9ca3af", "#f59e0b"


def animate(t: torch.Tensor, out: dict, eps_delta: tuple, path: Path) -> Path:
    eps, delta = eps_delta
    t = t.cpu().numpy()
    p = out["p"].cpu().numpy()  # (B, T, final, initial)
    kaon, anti = p[0, :, 0, 0], p[0, :, 1, 0]  # a beam that started as K⁰
    total = kaon + anti
    frac = kaon / total
    a_t = 1e3 * out["a_t"].cpu().numpy()  # (B, T) in 10⁻³
    a_cpt = 1e3 * out["a_cpt"].cpu().numpy()
    pipi = out["pipi"].cpu().numpy()  # (B, T, start)
    a_pipi = (pipi[..., 1] - pipi[..., 0]) / (pipi[..., 1] + pipi[..., 0])  # (B, T)

    fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(8, 6.4), sharex=True)
    ax1.set_ylim(0, 1)
    ax1.set_xlim(t[0], t[-1])
    ax1.set_ylabel("beam content")
    ax1.set_title(r"a pure K$^0$ beam: strangeness oscillates at $\Delta m$ while K$_S$ dies out", fontsize=10)
    ax1b = ax1.twinx()
    ax1b.set_yscale("log")
    ax1b.set_ylim(max(total.min(), 1e-12) / 2, 1.5)
    ax1b.set_ylabel("kaons left", color=GREY)
    ax1b.grid(False)
    ax1.plot([], [], color=BLUE, lw=6, label=r"K$^0$")
    ax1.plot([], [], color=RED, lw=6, alpha=0.6, label=r"$\bar{\mathrm{K}}^0$")
    ax1.legend(loc="upper right", fontsize=8, ncol=2, framealpha=0.9)
    (survive,) = ax1b.plot([], [], color="#374151", lw=1.2)

    value, error = tensors.CPLEAR_AT
    ax2.axhspan(1e3 * (value - error), 1e3 * (value + error), color=AMBER, alpha=0.25, lw=0, label="CPLEAR 1998")
    ax2.axhline(4e3 * eps.real, color=AMBER, lw=1, ls="--", label=r"$4\,\mathrm{Re}\,\varepsilon$")
    (line_t,) = ax2.plot([], [], color=BLUE, lw=2, label=r"$A_T$ (Kabir)")
    (line_cpt,) = ax2.plot([], [], color="#9333ea", lw=1.5, label=r"$A_{CPT}$")
    (line_zero,) = ax2.plot([], [], color=GREY, lw=1.5, label=r"$\varepsilon=\delta=0$")
    lo = min(-1.0, a_cpt[0].min() - 1)
    hi = max(1e3 * (value + error), a_t[0, 1:].max(), a_cpt[0].max()) + 2
    ax2.set_ylim(lo, hi)
    ax2.set_ylabel(r"asymmetry ($10^{-3}$)")
    ax2.legend(loc="center right", fontsize=7, ncol=3, framealpha=0.9)

    ax3.axhline(0, color=GREY, lw=0.8)
    (line_pp,) = ax3.plot([], [], color=BLUE, lw=2, label=r"$(\bar R-R)/(\bar R+R)$ to $\pi^+\pi^-$")
    (line_pp0,) = ax3.plot([], [], color=GREY, lw=1.5, label=r"$\varepsilon = 0$")
    span = np.abs(a_pipi[0]).max()
    ax3.set_ylim(-1.15 * span, 1.15 * span)
    ax3.set_ylabel("ππ asymmetry")
    ax3.set_xlabel(r"proper time $t$ ($\tau_S$ = 0.0895 ns)")
    ax3.legend(loc="upper left", fontsize=8, framealpha=0.9)
    cursors = [ax.axvline(t[0], color="#111827", lw=1) for ax in (ax1, ax2, ax3)]
    readout = ax2.text(0.01, 0.06, "", transform=ax2.transAxes, fontsize=8)
    fills = []

    def update(step: int):
        for f in fills:
            f.remove()
        fills.clear()
        s = slice(0, step + 1)
        fills.append(ax1.fill_between(t[s], 0, frac[s], color=BLUE, lw=0))
        fills.append(ax1.fill_between(t[s], frac[s], 1, color=RED, alpha=0.6, lw=0))
        survive.set_data(t[s], total[s])
        line_t.set_data(t[1 : step + 1], a_t[0, 1 : step + 1])
        line_cpt.set_data(t[s], a_cpt[0, s])
        line_zero.set_data(t[1 : step + 1], a_t[2, 1 : step + 1])
        line_pp.set_data(t[s], a_pipi[0, s])
        line_pp0.set_data(t[s], a_pipi[2, s])
        for c in cursors:
            c.set_xdata([t[step]])
        if step > 0:
            readout.set_text(f"$A_T$ = {a_t[0, step]:.2f}e-3   $A_{{CPT}}$ = {a_cpt[0, step]:.2f}e-3")

    fig.tight_layout()
    return save_gif(fig, update, frame_indices(len(t), 80), path)
```
