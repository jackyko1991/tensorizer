# Prime spirals and the Hardy–Littlewood conjectures

*maths · intermediate* · [open in Tensorizer](https://jackyko1991.github.io/tensorizer/#/maths/prime_spirals)

Sieve the primes into one boolean tensor and draw them on the Ulam and Sacks spirals, where they crowd onto lines. Each line is a quadratic such as Euler's $k^2 + k + 41$, and the Hardy–Littlewood conjectures predict how many primes it holds. The same sieve then tests their twin-prime and Goldbach predictions and Cramér's random model of prime gaps. These are conjectures: the program checks them numerically, it does not prove them.

**Vector form**

$$Q_f(K) \sim C(f)\sum_{k\le K}\frac{1}{\ln f(k)},\quad C(f) = \prod_p \frac{1 - w_f(p)/p}{1 - 1/p},\qquad \pi_2(x) \sim 2C_2\int_2^x \frac{dt}{\ln^2 t},\qquad r(N) \sim 2C_2\prod_{p\mid N,\,p>2}\frac{p-1}{p-2}\int_2^{N-2}\frac{dt}{\ln t\,\ln(N-t)}$$

**Index (tensor) form**

$$F_{ik} = a_i k^2 + b_i k + c_i,\quad Q_{iK} = \sum_{k\le K} \mathbb 1_P(F_{ik}),\qquad w_{ip} = 1 + \Big(\frac{D_i}{p}\Big),\quad \log C_i = \sum_p \log\frac{1 - w_{ip}/p}{1 - 1/p},\qquad r_N = \sum_{m} \mathbb 1_P(m)\,\mathbb 1_P(N - m)$$

| Symbol | Code | Shape | Meaning |
|---|---|---|---|
| $\mathbb 1_P(n)$ | `is_prime[p * p :: p] = False` | (n_max + 1,) bool | sieve of Eratosthenes: one strided slice per prime $p \le \sqrt{n_\text{max}}$ |
| $(x_n, y_n)$ | `ulam_xy(n, start)` | (N, 2) | Ulam spiral position of every number at once, from its ring $k = \lceil(\sqrt m - 1)/2\rceil$ and its distance to the corner $(2k+1)^2$ |
| $(\sqrt n\cos 2\pi\sqrt n,\ \sqrt n\sin 2\pi\sqrt n)$ | `sacks_xy(n)` | (N, 2) | Sacks spiral: the squares $k^2$ sit on the positive $x$-axis |
| $F_{ik}$ | `quadratic(coef, k)` | (F, K) | the values of $F$ quadratics $4k^2 + bk + c$ (the diagonal rays of the Ulam spiral) at every $k$ |
| $Q_{iK}$ | `cumsum(is_prime[values], dim=1)` | (F, K) | look every value up in the sieve; the running sum counts primes on each ray |
| $w_{ip} = 1 + (D_i/p)$ | `_power_mod(D % p, (p - 1) // 2, p)` | (F, P) | roots of $f_i$ mod $p$ from Euler's criterion $(D/p) \equiv D^{(p-1)/2}$, with $D = b^2 - 4ac$; tried directly for $p < 50$ |
| $\log C_i$ | `log((1 - w / p) / (1 - 1 / p)).sum(dim=1)` | (F,) | repeated $p$ ⇒ summed: the Euler product as a sum of logs over primes $p < 10^5$ |
| $r_N$ | `irfft(rfft(is_prime)**2)` | (2X - 1,) | repeated $m$ ⇒ summed: the Goldbach count as a self-convolution, done by FFT |
| $\pi_2(x)$ | `cumsum(is_prime[:-2] & is_prime[2:])` | (X,) | twin pairs $(p, p+2)$ with $p \le x$ |
| $g_n/\ln p_n$ | `(p[1:] - p[:-1]) / log(p[:-1])` | (G,) | normalised gaps, compared with the $e^{-s}$ law of Cramér's random model |

Ulam (1963) wrote the integers on a square spiral and noticed that the primes line up along diagonals; the Sacks spiral (1994) puts $n$ at radius $\sqrt n$ and angle $2\pi\sqrt n$ and shows curves instead. Every diagonal ray of the Ulam spiral holds the values of a quadratic $4k^2 + bk + c$, and on the spiral started at 41 the main diagonal is Euler's $k^2 + k + 41$, prime for $k = 0, \dots, 39$ (proved: Rabinowitsch showed $k^2 + k + p$ is prime for all $k < p - 1$ exactly when $\mathbb Q(\sqrt{1 - 4p})$ has class number one, so $p \in \{2, 3, 5, 11, 17, 41\}$, a list completed by Heegner and Stark). Some rays are dense and others empty because of the residues: $f$ has no roots modulo many small primes when its discriminant is a non-residue for them. Hardy and Littlewood (1923) turned this into Conjecture F, which the Bateman–Horn conjecture (1962) restates as $Q_f(K) \sim C(f)\sum_{k \le K} 1/\ln f(k)$. The top-right panel tests it on several hundred quadratics at once: as $K$ grows, the measured constants approach the predicted $C(f)$ (the diagonal line). No quadratic is even known to take infinitely many prime values: that is Landau's fourth problem (1912) for $n^2 + 1$, still open. For Euler's polynomial Hardy and Littlewood quote the odd-prime product $3.3197732$; the Bateman–Horn constant includes the factor 2 from $p = 2$ (the values are always odd), so it is $6.6395$. The twin-prime conjecture with constant $C_2 = 0.6601618\ldots$ and the Goldbach conjecture with its singular series are the same idea for $p$ and $p + 2$, and for $p$ and $N - p$. The sieve reproduces the known counts $\pi_2(10^6) = 8169$ and $\pi_2(10^7) = 58980$, and Goldbach's comet collapses onto the prediction: the bands of the comet are the singular series (green: $3 \mid N$). Both conjectures are open. Goldbach is verified for all even numbers up to $4 \cdot 10^{18}$ (Oliveira e Silva, Herzog and Pardi, 2014); Zhang (2014) and the Polymath and Maynard work proved infinitely many prime gaps of at most 246, and Chen (1973) that every large even number is a prime plus a product of at most two primes. Cramér's model (1936) treats each $n$ as prime with probability $1/\ln n$; it predicts exponential gaps with mean $\ln p$, close to what we see, and maximal gaps of size about $\ln^2 p$, but it gets the twin count wrong by exactly the factor $2C_2 \approx 1.32$ (dotted line) because it ignores divisibility by 2, 3, 5, … Maier (1985) showed that it fails in other ways too, and Granville suggested maximal gaps up to $2e^{-\gamma}\ln^2 p \approx 1.12\ln^2 p$.

## 1. `main.py` — main loop

```python
%%writefile problems/maths/prime_spirals/main.py
"""Main loop: parameters → tensors → solve → plot."""

import math
from pathlib import Path

import torch

from problems._common.runtime import cli, device, fast_mode

from . import equation, plot, solve, tensors


def run(params: dict, out_dir: Path) -> Path:
    dev = device()
    side = 61 if fast_mode() else params["side"]
    start = params["start"]
    x_max = 10 ** (5 if fast_mode() else params["log10_x"])
    g_max = 20000 if fast_mode() else params["goldbach_max"]
    k_max = 300 if fast_mode() else params["k_poly"]

    # 1. one sieve big enough for everything
    coef = tensors.quadratic_family(8, 60, dev)  # (F, 3): 4k² + bk + c
    q_max = int(equation.quadratic(coef, torch.tensor([k_max], device=dev)).max())
    n_max = max(x_max, g_max, q_max, start + side * side)
    is_prime = tensors.sieve(n_max, dev)
    primes = torch.nonzero(is_prime).flatten()
    pi = torch.cumsum(is_prime.long(), 0)
    print(f"sieve up to {n_max:,}: {len(primes):,} primes")
    print(f"{'x':>11} {'π(x)':>9} {'π₂(x)':>8} {'HL twins':>9} {'ratio':>6}")
    c2 = equation.twin_constant(primes[primes < 10**6])
    twins = equation.twin_count(is_prime[: x_max + 1])
    for e in range(3, int(math.log10(x_max)) + 1):
        pred = equation.twin_prediction(torch.tensor([10.0**e], dtype=torch.float64, device=dev), c2).item()
        print(f"{10**e:>11,} {int(pi[10**e]):>9,} {int(twins[10**e]):>8,} {pred:>9.0f} {twins[10**e] / pred:>6.3f}")
    print(f"twin-prime constant C₂ = {c2:.8f} (product over p < 10⁶)")

    # 2. the spirals
    n = torch.arange(start, start + side * side, device=dev)
    ulam = tensors.ulam_xy(n, start)  # (N, 2)
    sacks = tensors.sacks_xy(n)  # (N, 2)
    k = torch.arange(0, side // 2 + 1, device=dev)
    diag = torch.cat([k * k + k + start, k * k - k + start])  # the two halves of the main diagonal
    diag_frac = is_prime[diag].double().mean().item()
    print(
        f"Ulam spiral {side}×{side} from {start}: {is_prime[n].double().mean():.1%} primes overall, "
        f"{diag_frac:.1%} on the main diagonal k² + k + {start}"
    )
    if start == 41:
        print(
            "  Euler: k² + k + 41 is prime for k = 0 … 39 (Rabinowitsch: k² + k + p works for p = 2, 3, 5, 11, 17, 41)"
        )

    # 3. Conjecture F on every diagonal-type quadratic 4k² + bk + c
    small_primes = primes[primes < 10**5]
    fit = solve.conjecture_f(coef, is_prime, k_max, small_primes)
    alive = fit["constant"] > 0
    ratio = (fit["observed"][:, -1] / fit["predicted"][:, -1])[alive]
    print(
        f"Conjecture F on {int(alive.sum())} quadratics 4k² + bk + c (k ≤ {k_max}): observed / predicted primes "
        f"median {ratio.median():.3f}, middle 90% {ratio.quantile(0.05):.3f} … {ratio.quantile(0.95):.3f}"
    )
    pair = torch.stack([fit["observed"][alive, -1].double(), fit["constant"][alive]])
    print(f"  correlation of the prime count with C(f): {torch.corrcoef(pair)[0, 1]:.3f}")
    euler = solve.conjecture_f(torch.tensor([[1, 1, 41]], device=dev), is_prime, 2 * k_max, small_primes)
    print(
        f"  Euler n² + n + 41: C = {euler['constant'][0]:.5f} (= 2 × 3.31977…), {int(euler['observed'][0, -1])} primes "
        f"for 3 ≤ n ≤ {2 * k_max}, predicted {euler['predicted'][0, -1]:.0f}"
    )

    # 4. Goldbach's comet
    r = equation.goldbach_count(is_prime[: g_max + 1])
    n_even = torch.arange(6, g_max + 1, 2, device=dev)
    r_even = r[n_even].double()
    hl = equation.goldbach_prediction(n_even, primes, c2)
    upper = n_even > g_max // 2
    print(
        f"Goldbach: every even 4 ≤ N ≤ {g_max:,} has r(N) ≥ 1 (min {int(r_even.min())}); r / prediction for "
        f"N > {g_max // 2:,}: mean {(r_even / hl)[upper].mean():.4f}, std {(r_even / hl)[upper].std():.4f}"
    )

    # 5. gaps against Cramér's model
    gaps, norm = equation.normalised_gaps(is_prime[: x_max + 1])
    cramer = equation.cramer_model(x_max, 7, dev)
    _, norm_c = equation.normalised_gaps(cramer)
    twins_c = equation.twin_count(cramer)
    p_rec, rec_ratio = equation.ln_squared_ratio(is_prime[: x_max + 1])
    print(f"gaps up to {x_max:,}: mean g/ln p = {norm.mean():.3f} (Cramér model {norm_c.mean():.3f})")
    print(
        f"  record gaps: largest g/ln² p = {rec_ratio[p_rec > 100].max():.3f} (Cramér's conjecture ≤ 1, "
        f"Granville's 2e^(−γ) = {equation.GRANVILLE:.3f})"
    )
    print(
        f"  twins in Cramér's model: {int(twins_c[-1]):,} vs {int(twins[-1]):,} real: the model misses the factor "
        f"2C₂ = {2 * c2:.3f} (it ignores that p and p + 2 are both odd)"
    )

    data = {
        "side": side,
        "start": start,
        "ulam": ulam,
        "sacks": sacks,
        "prime_mask": is_prime[n],
        "diag_mask": torch.isin(n, diag),
        "fit": fit,
        "alive": alive,
        "x_max": x_max,
        "twins": twins,
        "twins_cramer": twins_c,
        "c2": c2,
        "n_even": n_even,
        "r_even": r_even,
        "hl": hl,
        "primes": primes[primes <= x_max],
        "norm": norm,
        "norm_cramer": norm_c,
        "cramer_primes": torch.nonzero(cramer).flatten(),
    }
    return plot.animate(data, out_dir / "prime_spirals.gif")


if __name__ == "__main__":
    cli(run, __file__)
```

## 2. `equation.py` — physics → tensor translation

```python
%%writefile problems/maths/prime_spirals/equation.py
r"""Mathematics → tensor translation: the Hardy–Littlewood conjectures as tensor operations.

1. Conjecture F / Bateman–Horn. For an irreducible quadratic f(k) = a k² + b k + c the number of primes among
   f(1), …, f(K) should be

        Q_f(K) ≈ C(f) Σ_{k≤K} 1 / ln f(k),      C(f) = Π_p (1 − w_f(p)/p) / (1 − 1/p),

   where w_f(p) is the number of roots of f mod p. For an odd prime p ∤ a, w_f(p) = 1 + (D/p) with D = b² − 4ac and
   the Legendre symbol from Euler's criterion (D/p) ≡ D^{(p−1)/2} mod p. The (F, P) table of w over F quadratics and
   P primes is built in one go and the product over p is a sum of logs over the prime axis.

2. Twin primes (Hardy–Littlewood): π₂(x) ≈ 2 C₂ ∫₂^x dt / ln² t,  C₂ = Π_{p>2} p(p−2)/(p−1)² = 0.6601618…

3. Goldbach: r(N) = #{(p, q) : p + q = N} is the self-convolution of the prime indicator, r = 1_P * 1_P (done by FFT),
   predicted to be r(N) ≈ 2 C₂ Π_{p | N, p > 2} (p−1)/(p−2) · ∫₂^{N−2} dt / (ln t · ln(N − t)).

4. Cramér's model: n is "prime" independently with probability 1/ln n, so gaps g_n near x are roughly exponential
   with mean ln x: P(g/ln p > s) ≈ e^{−s}.
"""

import math

import torch


def quadratic(coef: torch.Tensor, k: torch.Tensor) -> torch.Tensor:
    """f_i(k_j) = a_i k_j² + b_i k_j + c_i for every quadratic i and every k_j.   (F, 3), (K,) → (F, K) int64"""
    a, b, c = coef[:, 0:1], coef[:, 1:2], coef[:, 2:3]
    return a * k[None, :] ** 2 + b * k[None, :] + c


def _power_mod(base: torch.Tensor, exp: torch.Tensor, mod: torch.Tensor) -> torch.Tensor:
    """base^exp mod mod, elementwise, by repeated squaring (all entries < 2^31 so products fit in int64)."""
    result = torch.ones_like(base)
    base = base % mod
    exp = exp.clone()
    while bool((exp > 0).any()):
        odd = (exp % 2) == 1
        result = torch.where(odd, (result * base) % mod, result)
        base = (base * base) % mod
        exp = exp // 2
    return result


def root_counts(coef: torch.Tensor, primes: torch.Tensor, brute_below: int = 50) -> torch.Tensor:
    """w_f(p): number of x mod p with f(x) ≡ 0, for every quadratic and prime.   (F, 3), (P,) → (F, P)

    Small primes (where p might divide 2a) by trying every residue; larger ones by Euler's criterion."""
    coef, primes = coef.long(), primes.long()
    small = primes[primes < brute_below]
    x = torch.arange(brute_below, device=coef.device)
    vals = quadratic(coef, x)[:, None, :] % small[None, :, None]  # (F, P_small, X)
    in_range = x[None, None, :] < small[None, :, None]  # only residues 0 … p−1
    w_small = ((vals == 0) & in_range).sum(dim=2)
    large = primes[primes >= brute_below]
    disc = coef[:, 1] ** 2 - 4 * coef[:, 0] * coef[:, 2]  # D = b² − 4ac   (F,)
    d_mod = disc[:, None] % large[None, :]  # (F, P_large), in 0 … p−1
    euler = _power_mod(d_mod, ((large - 1) // 2)[None, :].expand_as(d_mod), large[None, :].expand_as(d_mod))
    legendre = torch.where(euler == large[None, :] - 1, -1, euler)  # 1, 0 or −1
    return torch.cat([w_small, 1 + legendre], dim=1)


def bateman_horn_constant(coef: torch.Tensor, primes: torch.Tensor) -> torch.Tensor:
    r"""C(f) = Π_p (1 − w_f(p)/p) / (1 − 1/p), truncated at the largest prime given.   (F, 3) → (F,) float64

    Zero when some prime divides every value (w_f(p) = p)."""
    w = root_counts(coef, primes).double()
    p = primes.double()[None, :]
    factor = (1 - w / p) / (1 - 1 / p)  # (F, P)
    dead = (factor <= 0).any(dim=1)
    log_c = torch.log(factor.clamp(min=1e-300)).sum(dim=1)  # product over p as a sum of logs
    return torch.where(dead, 0.0, torch.exp(log_c))


def twin_constant(primes: torch.Tensor) -> float:
    r"""C₂ = Π_{p>2} p(p − 2)/(p − 1)²."""
    p = primes[primes > 2].double()
    return torch.exp(torch.log(p * (p - 2) / (p - 1) ** 2).sum()).item()


def log_integral(x: torch.Tensor, power: int, n_grid: int = 4000) -> torch.Tensor:
    """∫₂^x dt / ln^power t for every x (trapezoid rule on a log-spaced grid).   (X,) → (X,)"""
    u = torch.linspace(0, 1, n_grid, dtype=torch.float64, device=x.device)[None, :]
    t = 2 * (x.double()[:, None] / 2) ** u  # (X, G) from 2 to x
    return torch.trapezoid(1 / torch.log(t) ** power, t, dim=1)


def twin_prediction(x: torch.Tensor, c2: float) -> torch.Tensor:
    """Hardy–Littlewood: π₂(x) ≈ 2 C₂ ∫₂^x dt / ln² t."""
    return 2 * c2 * log_integral(x, 2)


def twin_count(is_prime: torch.Tensor) -> torch.Tensor:
    """π₂(x) for every x: number of p ≤ x with p and p + 2 both prime.   (X,) bool → (X,) int64"""
    starts = torch.zeros_like(is_prime)
    starts[:-2] = is_prime[:-2] & is_prime[2:]
    return torch.cumsum(starts.long(), dim=0)


def goldbach_count(is_prime: torch.Tensor) -> torch.Tensor:
    """r(N) = Σ_p 1_P(p) 1_P(N − p): the self-convolution of the prime indicator, by FFT.   (X,) → (2X − 1,) int64"""
    n = 2 * len(is_prime) - 1
    spectrum = torch.fft.rfft(is_prime.double(), n=n)
    return torch.fft.irfft(spectrum * spectrum, n=n).round().long()


def goldbach_prediction(n_even: torch.Tensor, primes: torch.Tensor, c2: float, n_grid: int = 400) -> torch.Tensor:
    r"""Hardy–Littlewood prediction for r(N) at every even N.   (E,) int64 → (E,) float64

    The singular series Π_{p|N, p>2} (p−1)/(p−2) is built by multiplying in (p−1)/(p−2) on every multiple of p."""
    n_max = int(n_even.max())
    series = torch.ones(n_max + 1, dtype=torch.float64, device=n_even.device)
    for p in primes[(primes > 2) & (primes <= n_max)].tolist():
        series[p::p] *= (p - 1) / (p - 2)
    n = n_even.double()[:, None]
    t = 2 + (n - 4) * torch.linspace(0, 1, n_grid, dtype=torch.float64, device=n_even.device)[None, :]  # (E, G)
    integral = torch.trapezoid(1 / (torch.log(t) * torch.log(n - t)), t, dim=1)
    return 2 * c2 * series[n_even] * integral


def cramer_model(n_max: int, seed: int, device: torch.device) -> torch.Tensor:
    """Cramér's random primes: n ≥ 3 is kept with probability 1/ln n (2 is kept).   → (n_max + 1,) bool"""
    gen = torch.Generator(device="cpu").manual_seed(seed)
    n = torch.arange(n_max + 1, dtype=torch.float64)
    prob = torch.where(n >= 3, 1 / torch.log(n.clamp(min=2)), 0.0)
    keep = torch.rand(n_max + 1, generator=gen, dtype=torch.float64) < prob
    keep[2] = True
    return keep.to(device)


def normalised_gaps(is_prime: torch.Tensor) -> tuple[torch.Tensor, torch.Tensor]:
    """Gaps g_n = p_{n+1} − p_n and g_n / ln p_n for every consecutive pair.   → (G,), (G,)"""
    p = torch.nonzero(is_prime).flatten()
    gaps = p[1:] - p[:-1]
    return gaps, gaps.double() / torch.log(p[:-1].double().clamp(min=2))


def ln_squared_ratio(is_prime: torch.Tensor) -> tuple[torch.Tensor, torch.Tensor]:
    """Record (maximal) gaps and their ratio g / ln² p (Cramér's conjecture: limsup ≤ 1, Granville: 2e^{−γ}?)."""
    p = torch.nonzero(is_prime).flatten()
    gaps = p[1:] - p[:-1]
    record = gaps > torch.cummax(torch.cat([gaps.new_zeros(1), gaps[:-1]]), dim=0).values
    p_rec = p[:-1][record].double()
    return p_rec, gaps[record].double() / torch.log(p_rec) ** 2


EULER_GAMMA = 0.5772156649015329
GRANVILLE = 2 * math.exp(-EULER_GAMMA)  # 1.1229…
```

## 3. `tensors.py` — PyTorch tensors

```python
%%writefile problems/maths/prime_spirals/tensors.py
"""Build the concrete tensors: the prime sieve, Ulam and Sacks spiral coordinates and the family of quadratics."""

import math

import torch

# Well-established values used only to check the computation (OEIS A007508: twin pairs (p, p+2) with p ≤ 10^k).
KNOWN_TWINS = {10**3: 35, 10**4: 205, 10**5: 1224, 10**6: 8169, 10**7: 58980, 10**8: 440312}


def sieve(n_max: int, device: torch.device) -> torch.Tensor:
    """Sieve of Eratosthenes as a boolean tensor is_prime[n] for n = 0 … n_max: (n_max + 1,)."""
    is_prime = torch.ones(n_max + 1, dtype=torch.bool, device=device)
    is_prime[:2] = False
    for p in range(2, math.isqrt(n_max) + 1):
        if is_prime[p]:
            is_prime[p * p :: p] = False  # strike every multiple from p² on in one strided slice
    return is_prime


def ulam_xy(n: torch.Tensor, start: int) -> torch.Tensor:
    """Grid position of every number n on the Ulam spiral that starts with `start` at (0, 0), all at once.

    With m = n − start + 1 the ring index is k = ⌈(√m − 1)/2⌉; ring k ends at (2k+1)² in the corner (k, −k) and is
    walked anticlockwise from (k, −k+1): up the right side, left along the top, down the left, right along the bottom.
    (N,) int → (N, 2) int64
    """
    m = (n - start + 1).long()
    k = torch.ceil((torch.sqrt(m.double()) - 1) / 2).long()
    side = 2 * k  # cells per side of ring k
    corner = (2 * k + 1) ** 2  # last number of ring k, at (k, −k)
    back = corner - m  # how far behind the corner, 0 … 8k − 1
    x = torch.where(back < side, k - back, torch.zeros_like(k))  # bottom side
    y = torch.where(back < side, -k, torch.zeros_like(k))
    on_left = (back >= side) & (back < 2 * side)
    x, y = torch.where(on_left, -k, x), torch.where(on_left, -k + back - side, y)
    on_top = (back >= 2 * side) & (back < 3 * side)
    x, y = torch.where(on_top, -k + back - 2 * side, x), torch.where(on_top, k, y)
    on_right = back >= 3 * side
    x, y = torch.where(on_right, k, x), torch.where(on_right, k - back + 3 * side, y)
    return torch.stack([x, y], dim=1)


def sacks_xy(n: torch.Tensor) -> torch.Tensor:
    """Sacks spiral: n at radius √n and angle 2π√n, so the squares k² line up on the positive x-axis.   (N,) → (N, 2)"""
    r = torch.sqrt(n.double())
    return torch.stack([r * torch.cos(2 * math.pi * r), r * torch.sin(2 * math.pi * r)], dim=1)


def quadratic_family(b_max: int, c_max: int, device: torch.device) -> torch.Tensor:
    """Coefficients (a, b, c) of f(k) = 4k² + bk + c, |b| ≤ b_max, 1 ≤ c ≤ c_max: every diagonal ray of the Ulam
    spiral is such a quadratic. Reducible ones (b² − 16c a perfect square) are dropped.   → (F, 3) int64"""
    b, c = torch.meshgrid(
        torch.arange(-b_max, b_max + 1, device=device), torch.arange(1, c_max + 1, device=device), indexing="ij"
    )
    b, c = b.flatten(), c.flatten()
    disc = b * b - 16 * c
    root = torch.sqrt(disc.clamp(min=0).double()).round().long()
    keep = ~((disc >= 0) & (root * root == disc))
    return torch.stack([torch.full_like(b, 4), b, c], dim=1)[keep]
```

## 4. `solve.py` — solver

```python
%%writefile problems/maths/prime_spirals/solve.py
"""Measure the primes against the Hardy–Littlewood predictions: quadratics, twins, Goldbach and gaps."""

import torch

from .equation import bateman_horn_constant, quadratic


def conjecture_f(coef: torch.Tensor, is_prime: torch.Tensor, k_max: int, primes: torch.Tensor) -> dict:
    """For every quadratic: the running prime count Q_f(K) and the prediction C(f) Σ_{3≤k≤K} 1/ln f(k).

    Returns the constants (F,), the observed counts (F, K) and the predicted counts (F, K) for K = 3 … k_max."""
    k = torch.arange(3, k_max + 1, device=coef.device)
    values = quadratic(coef, k)  # (F, K) all > 1 for k ≥ 3
    hits = is_prime[values].long()  # look every value up in the sieve at once
    observed = torch.cumsum(hits, dim=1)
    constant = bateman_horn_constant(coef, primes)  # (F,)
    predicted = constant[:, None] * torch.cumsum(1 / torch.log(values.double()), dim=1)
    return {"k": k, "constant": constant, "observed": observed, "predicted": predicted}


def cumulative_on(x: torch.Tensor, counts: torch.Tensor) -> torch.Tensor:
    """counts[x] for an integer grid x (a running count looked up at many x at once)."""
    return counts[x.long()].double()
```

## 5. `plot.py` — 2D/3D plot → GIF

```python
%%writefile problems/maths/prime_spirals/plot.py
"""Ulam and Sacks spirals growing, with the Hardy–Littlewood checks (quadratics, twins, Goldbach, gaps) alongside."""

import math
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import torch
from matplotlib.colors import ListedColormap

from problems._common.plotting import ACCENT, frame_indices, save_gif

from .equation import twin_prediction

N_FRAMES = 45


def animate(d: dict, path: Path) -> Path:
    side, half = d["side"], d["side"] // 2
    ulam = d["ulam"].cpu().numpy() + half
    prime = d["prime_mask"].cpu().numpy()
    diag = d["diag_mask"].cpu().numpy()
    sacks = d["sacks"].cpu().numpy()

    fig, axes = plt.subplots(2, 3, figsize=(12, 7.2))
    (ax_u, ax_s, ax_f), (ax_t, ax_g, ax_h) = axes

    # Ulam spiral: 0 not yet drawn, 1 composite, 2 prime, 3 prime on the highlighted diagonal
    cmap = ListedColormap(["#ffffff", "#f3f4f6", "#1e3a8a", "#dc2626"])
    code = np.where(prime, np.where(diag, 3, 2), 1)
    img = np.zeros((side, side))
    im = ax_u.imshow(img, cmap=cmap, vmin=0, vmax=3, origin="lower", interpolation="nearest")
    ax_u.set_xticks([])
    ax_u.set_yticks([])
    ax_u.grid(False)
    ax_u.set_title(f"Ulam spiral from {d['start']}: red = $k^2+k+{d['start']}$", fontsize=9)

    pr = sacks[prime]
    on_diag = sacks[prime & diag]
    sc_s = ax_s.scatter(pr[:1, 0], pr[:1, 1], s=1.2, color="#1e3a8a", lw=0)
    sc_d = ax_s.scatter(on_diag[:1, 0], on_diag[:1, 1], s=4, color="#dc2626", lw=0)
    lim = math.sqrt(d["start"] + side * side)
    ax_s.set_xlim(-lim, lim)
    ax_s.set_ylim(-lim, lim)
    ax_s.set_aspect("equal")
    ax_s.set_xticks([])
    ax_s.set_yticks([])
    ax_s.grid(False)
    ax_s.set_title(r"Sacks spiral: $n$ at $r=\sqrt{n}$, $\theta = 2\pi\sqrt{n}$", fontsize=9)
    idx_prime = np.flatnonzero(prime)
    idx_diag = np.flatnonzero(prime & diag)

    fit = d["fit"]
    alive = d["alive"].cpu().numpy()
    const = fit["constant"].cpu().numpy()[alive]
    obs = fit["observed"].cpu().numpy()[alive].astype(float)
    pred = fit["predicted"].cpu().numpy()[alive]
    ks = fit["k"].cpu().numpy()
    emp = obs / (pred / const[:, None])  # observed / Σ 1/ln f = empirical constant
    sc_f = ax_f.scatter(const, emp[:, 0], s=6, color=ACCENT[0], alpha=0.6, lw=0)
    top = const.max() * 1.1
    ax_f.plot([0, top], [0, top], color=ACCENT[1], lw=1.2, label="Conjecture F: equal")
    ax_f.set_xlim(0, top)
    ax_f.set_ylim(0, top * 1.15)
    ax_f.set_xlabel("predicted constant $C(f)$")
    ax_f.set_ylabel(r"primes / $\sum 1/\ln f(k)$")
    ax_f.legend(fontsize=7, loc="upper left")
    title_f = ax_f.set_title("", fontsize=9)

    x_max = d["x_max"]
    xs = np.unique(np.logspace(3, math.log10(x_max), 300).astype(np.int64))
    hl_t = twin_prediction(torch.tensor(xs, dtype=torch.float64), d["c2"]).numpy()
    real = d["twins"].cpu().numpy()[xs] / hl_t
    model = d["twins_cramer"].cpu().numpy()[xs] / hl_t
    (ln_real,) = ax_t.semilogx(xs[:1], real[:1], color=ACCENT[0], lw=1.5, label=r"real $\pi_2(x)$")
    (ln_model,) = ax_t.semilogx(xs[:1], model[:1], color=ACCENT[4], lw=1.2, label="Cramér model")
    ax_t.axhline(1, color=ACCENT[1], lw=1, ls="--")
    ax_t.axhline(1 / (2 * d["c2"]), color="#9ca3af", lw=1, ls=":", label=r"$1/2C_2$")
    ax_t.set_xlim(xs[0], xs[-1])
    ax_t.set_ylim(0.5, 1.3)
    ax_t.set_xlabel("$x$")
    ax_t.set_ylabel(r"count / $2C_2\int dt/\ln^2 t$")
    ax_t.legend(fontsize=7, loc="lower right")
    ax_t.set_title("twin primes vs Hardy–Littlewood", fontsize=9)

    ne, re_, hl = d["n_even"].cpu().numpy(), d["r_even"].cpu().numpy(), d["hl"].cpu().numpy()
    keep = np.unique(np.linspace(0, len(ne) - 1, min(len(ne), 4000)).astype(int))
    ne, re_, hl = ne[keep], re_[keep], hl[keep]
    band = np.where(ne % 3 == 0, ACCENT[2], "#6b7280")
    sc_g = ax_g.scatter(ne[:1], re_[:1], s=1.5, c=band[:1], lw=0)
    sc_p = ax_g.scatter(ne[:1], hl[:1], s=1.0, color=ACCENT[1], lw=0)
    ax_g.set_xlim(0, ne[-1])
    ax_g.set_ylim(0, re_.max() * 1.05)
    ax_g.set_xlabel("even $N$")
    ax_g.set_ylabel("$r(N)$ = #{p + q = N}")
    title_g = ax_g.set_title("", fontsize=9)

    primes = d["primes"].cpu().numpy()
    norm, norm_c = d["norm"].cpu().numpy(), d["norm_cramer"].cpu().numpy()
    bins = np.linspace(0, 6, 31)
    centres = 0.5 * (bins[1:] + bins[:-1])
    (st_r,) = ax_h.semilogy(
        centres, np.ones_like(centres), drawstyle="steps-mid", color=ACCENT[0], lw=1.5, label="primes"
    )
    (st_c,) = ax_h.semilogy(
        centres, np.ones_like(centres), drawstyle="steps-mid", color=ACCENT[4], lw=1, label="Cramér model"
    )
    ax_h.semilogy(centres, np.exp(-centres), color=ACCENT[1], lw=1, ls="--", label="$e^{-s}$")
    ax_h.set_ylim(1e-4, 2)
    ax_h.set_xlabel(r"normalised gap $s = g/\ln p$")
    ax_h.set_ylabel("density")
    ax_h.legend(fontsize=7, loc="upper right")
    title_h = ax_h.set_title("", fontsize=9)
    cramer_primes = d["cramer_primes"].cpu().numpy()

    def hist(v: np.ndarray) -> np.ndarray:
        h, _ = np.histogram(v, bins=bins, density=True)
        return np.where(h > 0, h, np.nan)

    def update(f: int):
        s = (f + 1) / N_FRAMES
        ring = max(1, round(half * s))
        count = (2 * ring + 1) ** 2  # numbers in the first `ring` rings
        img[:] = 0
        img[ulam[:count, 1], ulam[:count, 0]] = code[:count]
        im.set_data(img)
        sc_s.set_offsets(sacks[idx_prime[idx_prime < count]])
        sc_d.set_offsets(sacks[idx_diag[idx_diag < count]] if (idx_diag < count).any() else np.empty((0, 2)))

        j = min(len(ks) - 1, max(0, round((len(ks) - 1) * s**2)))
        sc_f.set_offsets(np.column_stack([const, emp[:, j]]))
        title_f.set_text(f"{len(const)} quadratics $4k^2+bk+c$, $k \\leq {ks[j]}$")

        m = max(2, round(len(xs) * s))
        ln_real.set_data(xs[:m], real[:m])
        ln_model.set_data(xs[:m], model[:m])

        g = max(2, round(len(ne) * s))
        sc_g.set_offsets(np.column_stack([ne[:g], re_[:g]]))
        sc_g.set_color(band[:g])
        sc_p.set_offsets(np.column_stack([ne[:g], hl[:g]]))
        title_g.set_text(f"Goldbach's comet to {ne[g - 1]:,} (red: HL prediction)")

        x_f = 10 ** (3 + (math.log10(x_max) - 3) * s)
        n_r = max(2, int(np.searchsorted(primes, x_f)))
        st_r.set_ydata(hist(norm[: n_r - 1]))
        st_c.set_ydata(hist(norm_c[: max(1, int(np.searchsorted(cramer_primes, x_f)) - 1)]))
        title_h.set_text(f"prime gaps up to $x$ = {x_f:,.0f}")

    fig.tight_layout()
    return save_gif(fig, update, frame_indices(N_FRAMES, N_FRAMES), path, fps=5)
```
